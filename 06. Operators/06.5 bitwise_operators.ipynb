# 06.5 Bitwise Operators

Six operators working on the **individual bits** of an integer. Most everyday
Python never needs them — but they are essential for flags, permissions, hashing,
network protocols and low-level work, and they appear in code you will read.

**40 numbered examples.**

## Theory

### The six operators

| Operator | Name | Rule |
|---|---|---|
| `&` | AND | 1 only when **both** bits are 1 |
| `\|` | OR | 1 when **either** bit is 1 |
| `^` | XOR | 1 when the bits **differ** |
| `~` | NOT | Inverts every bit |
| `<<` | Left shift | Move bits left, filling with 0 |
| `>>` | Right shift | Move bits right |

### Shifts are multiplication and division

```
x << n   is   x * (2 ** n)
x >> n   is   x // (2 ** n)
```

Shifting is how you multiply or divide by powers of two cheaply. Note `>>` floors,
matching `//`.

### `~` and two's complement

`~x` equals `-(x + 1)`. That looks arbitrary until you know Python uses **two's
complement** with conceptually infinite sign bits:

```
~5  ==  -6
~0  ==  -1
```

### Where they are actually used

- **Flag sets** — permissions packed into one integer
- **Masks** — extracting parts of a value
- **Hashing** — mixing bits
- **Protocols** — parsing binary formats
- **Sets** — `&`, `|`, `^` are overloaded for set operations (Chapter 13)

### The critical warning

`&` is **not** `and`. It does not short-circuit, it binds **tighter** than
comparison, and on integers it does arithmetic. `1 & 2` is `0`, while `1 and 2`
is `2`.

````python
# EXAMPLE 1-6: the six operators on one pair.
print("EXAMPLE 1-6: the six bitwise operators")
print("")

left = 12    # 1100
right = 10   # 1010

print(f"   left  = {left:>3}  binary {left:08b}")
print(f"   right = {right:>3}  binary {right:08b}")
print("")
print(f"   1. left &  right = {left & right:>4}  {left & right:08b}  (AND)")
print(f"   2. left |  right = {left | right:>4}  {left | right:08b}  (OR)")
print(f"   3. left ^  right = {left ^ right:>4}  {left ^ right:08b}  (XOR)")
print(f"   4. ~left         = {~left:>4}            (NOT)")
print(f"   5. left << 1     = {left << 1:>4}  {left << 1:08b}  (shift left)")
print(f"   6. left >> 1     = {left >> 1:>4}  {left >> 1:08b}  (shift right)")
````

````python
# EXAMPLE 7-12: bit-by-bit truth tables.
print("EXAMPLE 7-12: what each operator does per bit")
print("")

print("   A  B  |  A&B  A|B  A^B")
print("   " + "-" * 26)
for bit_a in [0, 1]:
    for bit_b in [0, 1]:
        print(f"   {bit_a}  {bit_b}  |   {bit_a & bit_b}    "
              f"{bit_a | bit_b}    {bit_a ^ bit_b}")

print("")
print("   7.  AND: 1 only when BOTH are 1")
print("   8.  OR:  1 when EITHER is 1")
print("   9.  XOR: 1 when they DIFFER")
print("")
print("   10-12. Applied to full numbers, column by column:")
print(f"        12 = {12:04b}")
print(f"        10 = {10:04b}")
print(f"        --   ----")
print(f"        &  = {12 & 10:04b}  = {12 & 10}")
print(f"        |  = {12 | 10:04b}  = {12 | 10}")
print(f"        ^  = {12 ^ 10:04b}  = {12 ^ 10}")
````

````python
# EXAMPLE 13-18: shifts are powers of two.
print("EXAMPLE 13-18: shifting")
print("")

value = 5
print(f"   starting from {value} ({value:08b})")
print("")
for shift in range(4):
    shifted = value << shift
    print(f"   13-16. {value} << {shift} = {shifted:>3}  {shifted:08b}"
          f"   = {value} * 2**{shift} = {value * 2 ** shift}")

print("")
value = 80
for shift in range(4):
    shifted = value >> shift
    print(f"   17. {value} >> {shift} = {shifted:>3}  {shifted:08b}"
          f"   = {value} // 2**{shift} = {value // 2 ** shift}")

# 18. Right shift floors, including for negatives.
print("")
print("   18. >> floors, matching //:")
for number in [7, -7]:
    print(f"       {number:>3} >> 1 = {number >> 1:>3}"
          f"   {number:>3} // 2 = {number // 2:>3}")
````

````python
# EXAMPLE 19-22: ~ and two's complement.
print("EXAMPLE 19-22: the NOT operator")
print("")

print("   19. ~x is always -(x + 1):")
for value in [0, 1, 5, 10, -1, -6]:
    print(f"       ~{value:>3} = {~value:>3}   and -({value} + 1) = {-(value + 1):>3}")

# 20. Double negation returns the original.
print("")
print("   20. ~~x == x ->", all(~~value == value for value in range(-5, 6)))

# 21. Why: Python integers have conceptually infinite sign bits.
print("")
print("   21. Python has no fixed width, so ~ flips an infinite sign extension.")
print("       To emulate an 8-bit NOT, mask the result:")
value = 5
print(f"       ~{value}        = {~value}")
print(f"       ~{value} & 0xFF = {~value & 0xFF}  ({~value & 0xFF:08b})")

# 22. Masking to a fixed width generally.
print("")
print("   22. masking to common widths:")
for bits in [8, 16, 32]:
    mask = (1 << bits) - 1
    print(f"       {bits:>2}-bit mask = {mask:>12}  (~5 & mask = {~5 & mask})")
````

## The practical use: flags

````python
# EXAMPLE 23-30: packing flags into one integer.
print("EXAMPLE 23-30: flag sets")
print("")

# Each permission is a single distinct bit.
READ = 0b0001      # 1
WRITE = 0b0010     # 2
EXECUTE = 0b0100   # 4
DELETE = 0b1000    # 8

names = {READ: "READ", WRITE: "WRITE", EXECUTE: "EXECUTE", DELETE: "DELETE"}

print("   23. each permission is one bit:")
for flag, name in names.items():
    print(f"       {name:<9} = {flag:>2}  {flag:04b}")

# 24. Combine with OR.
permissions = READ | WRITE
print("")
print(f"   24. READ | WRITE = {permissions}  {permissions:04b}")

# 25. Test with AND.
print("")
print("   25. testing a flag with &:")
for flag, name in names.items():
    has_it = bool(permissions & flag)
    print(f"       has {name:<9} {has_it}")

# 26. Add a flag.
permissions |= EXECUTE
print("")
print(f"   26. after |= EXECUTE: {permissions}  {permissions:04b}")

# 27. Remove a flag.
permissions &= ~WRITE
print(f"   27. after &= ~WRITE:  {permissions}  {permissions:04b}")

# 28. Toggle a flag.
permissions ^= READ
print(f"   28. after ^= READ:    {permissions}  {permissions:04b}")
permissions ^= READ
print(f"   29. after ^= READ:    {permissions}  {permissions:04b}  <- back again")

# 30. List which flags are set.
print("")
active = [name for flag, name in names.items() if permissions & flag]
print("   30. active permissions:", active)
````

````python
# EXAMPLE 31-34: the modern way - enum.Flag.
print("EXAMPLE 31-34: enum.Flag does this properly")
print("")

from enum import Flag, auto


class Permission(Flag):
    """Permissions as a proper flag enum."""
    READ = auto()
    WRITE = auto()
    EXECUTE = auto()
    DELETE = auto()


# 31. Combining reads naturally.
permissions = Permission.READ | Permission.WRITE
print("   31. Permission.READ | Permission.WRITE ->", permissions)

# 32. Testing uses `in`.
print("")
print("   32. Permission.READ in permissions ->", Permission.READ in permissions)
print("       Permission.DELETE in permissions ->",
      Permission.DELETE in permissions)

# 33. Adding and removing.
permissions |= Permission.EXECUTE
print("")
print("   33. after |= EXECUTE ->", permissions)
permissions &= ~Permission.WRITE
print("       after &= ~WRITE  ->", permissions)

# 34. The underlying value is still an integer.
print("")
print("   34. underlying value:", permissions.value,
      f"({permissions.value:04b})")
print("       Same bit tricks, readable names, type safety.")
print("       Enums are covered fully in Chapter 28.")
````

````python
# EXAMPLE 35-40: other real uses, and the & vs and warning.
print("EXAMPLE 35-40: other uses")
print("")

# 35. Check if a number is even.
print("   35. even check with &:")
for number in [4, 7]:
    print(f"       {number} & 1 == 0 -> {number & 1 == 0}")

# 36. Check for a power of two.
print("")
print("   36. power-of-two check:")
for number in [8, 12, 16, 1]:
    is_power = number > 0 and (number & (number - 1)) == 0
    print(f"       {number:>3} -> {is_power}")

# 37. Swap without a temporary, using XOR.
a, b = 5, 9
a ^= b
b ^= a
a ^= b
print("")
print("   37. XOR swap: 5, 9 ->", a, b)
print("       (a real technique, but Python's a, b = b, a is clearer)")

# 38. Extract bytes from a packed integer.
packed = 0xDEADBEEF
print("")
print("   38. extracting bytes from 0xDEADBEEF:")
for index in range(4):
    byte = (packed >> (index * 8)) & 0xFF
    print(f"       byte {index}: {byte:>3}  0x{byte:02X}")

# 39. Set operations reuse these symbols.
print("")
print("   39. the same symbols work on sets:")
print("       {1,2,3} & {2,3,4} ->", {1, 2, 3} & {2, 3, 4})
print("       {1,2,3} | {2,3,4} ->", {1, 2, 3} | {2, 3, 4})
print("       {1,2,3} ^ {2,3,4} ->", {1, 2, 3} ^ {2, 3, 4})

# 40. THE WARNING.
print("")
print("   40. & is NOT and:")
print("       1 and 2  ->", 1 and 2, " (returns an operand)")
print("       1 & 2    ->", 1 & 2, " (bitwise AND of 01 and 10)")
print("       & does not short-circuit, and binds tighter than comparison:")
print("       1 & 2 == 2 parses as 1 & (2 == 2) ->", 1 & (2 == 2))
````

## Takeaways

1. Six operators working on individual bits: `&`, `|`, `^`, `~`, `<<`, `>>`.
2. `x << n` is `x * 2**n`; `x >> n` is `x // 2**n` — and `>>` floors like `//`.
3. `~x` is always `-(x + 1)`, because Python uses two's complement with no fixed
   width. Mask with `& 0xFF` to emulate a byte.
4. **Flags** are the main everyday use: `|` to add, `& flag` to test, `& ~flag`
   to remove, `^` to toggle.
5. Prefer **`enum.Flag`** for real code — same bit operations, readable names.
6. `n & (n - 1) == 0` tests for a power of two; `n & 1` tests oddness.
7. `&`, `|` and `^` are **overloaded for sets** (Chapter 13).
8. `&` is **not** `and` — no short-circuiting, tighter binding, different result.
