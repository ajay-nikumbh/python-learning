# 06.1 Arithmetic Operators

Seven operators, and every one of them has behaviour that surprises people at
least once. This notebook works through all of them with **30 numbered
examples**, including the edge cases that cause real bugs.

## Theory

### The seven arithmetic operators

| Operator | Name | `7` and `2` give |
|---|---|---|
| `+` | Addition | `9` |
| `-` | Subtraction | `5` |
| `*` | Multiplication | `14` |
| `/` | True division | `3.5` |
| `//` | Floor division | `3` |
| `%` | Modulo | `1` |
| `**` | Exponentiation | `49` |

### Operators are method calls

Every operator is syntactic sugar for a **dunder method**:

```python
a + b        is     a.__add__(b)
a ** b       is     a.__pow__(b)
```

This is why `+` concatenates strings and lists as well as adding numbers — those
types define `__add__` differently. It is also how you make operators work on
your own classes (Chapter 27).

### The type-promotion rule

When operands differ, Python widens to the more general type:

```
int  →  float  →  complex
```

No information is lost widening that direction, so it is safe to do
automatically. Python never narrows.

### The three that surprise people

1. `/` **always** returns a float — `10 / 5` is `2.0`
2. `//` rounds towards **negative infinity** — `-7 // 2` is `-4`
3. `%` takes the sign of the **divisor** — `-7 % 2` is `1`

All three came up in 05.1. Here we see why they are mutually consistent.

````python
# The seven operators, on the same pair of values.
first = 7
second = 2

print("EXAMPLE 1-7: the seven arithmetic operators")
print("")
print(f"   1.  {first} +  {second}  = {first + second}")
print(f"   2.  {first} -  {second}  = {first - second}")
print(f"   3.  {first} *  {second}  = {first * second}")
print(f"   4.  {first} /  {second}  = {first / second}   <- always a float")
print(f"   5.  {first} // {second}  = {first // second}")
print(f"   6.  {first} %  {second}  = {first % second}")
print(f"   7.  {first} ** {second}  = {first ** second}")
````

````python
# EXAMPLE 8-12: type promotion along int -> float -> complex.
print("EXAMPLE 8-12: mixed types promote to the wider one")
print("")

cases = [
    ("8.  int + int    ", 3 + 4),
    ("9.  int + float  ", 3 + 4.0),
    ("10. int + complex", 3 + 4j),
    ("11. float + complex", 3.5 + 4j),
    ("12. bool + int   ", True + 4),
]

for label, result in cases:
    print(f"   {label} = {str(result):<12} {type(result).__name__}")

print("")
print("Promotion only ever widens. Python never narrows automatically.")
````

````python
# EXAMPLE 13-16: every operator is a dunder method call.
print("EXAMPLE 13-16: operators are method calls")
print("")

value_a = 7
value_b = 2

print("   13. 7 + 2          ->", value_a + value_b)
print("       (7).__add__(2) ->", value_a.__add__(value_b))
print("")
print("   14. 7 * 2          ->", value_a * value_b)
print("       (7).__mul__(2) ->", value_a.__mul__(value_b))
print("")
print("   15. 7 ** 2         ->", value_a ** value_b)
print("       (7).__pow__(2) ->", value_a.__pow__(value_b))

# When the left operand cannot handle it, Python tries the RIGHT operand's
# reflected method.
print("")
print("   16. reflected operations:")
print("       (2).__radd__ exists?", hasattr(value_b, "__radd__"))
print("       This is how 1 + Fraction(1,2) works - int cannot add a")
print("       Fraction, so Fraction.__radd__ handles it.")

from fractions import Fraction
print("       1 + Fraction(1, 2) =", 1 + Fraction(1, 2))
````

## Division in depth

The three division-related operators answer different questions.

````python
# EXAMPLE 17-20: the division family compared.
print("EXAMPLE 17-20: / vs // vs % vs divmod")
print("")
print("   Expression      Result        Type")
print("   " + "-" * 44)

pairs = [(17, 5), (10, 5), (7, 2), (1, 3)]

for numerator, denominator in pairs:
    true_div = numerator / denominator
    floor_div = numerator // denominator
    remainder = numerator % denominator

    print(f"   {numerator} / {denominator}".ljust(18),
          f"{true_div:<13} {type(true_div).__name__}")
    print(f"   {numerator} // {denominator}".ljust(18),
          f"{floor_div:<13} {type(floor_div).__name__}")
    print(f"   {numerator} % {denominator}".ljust(18),
          f"{remainder:<13} {type(remainder).__name__}")
    print(f"   divmod({numerator}, {denominator})".ljust(18),
          f"{str(divmod(numerator, denominator)):<13} tuple")
    print("")
````

````python
# EXAMPLE 21-24: negative numbers, where intuition fails.
print("EXAMPLE 21-24: floor division with negatives")
print("")
print("   Expression    //      %      int(a/b)   Check: (a//b)*b + a%b")
print("   " + "-" * 66)

cases = [(7, 2), (-7, 2), (7, -2), (-7, -2)]

for numerator, denominator in cases:
    floor_result = numerator // denominator
    mod_result = numerator % denominator
    truncated = int(numerator / denominator)
    reconstructed = floor_result * denominator + mod_result

    expression = f"{numerator} and {denominator}"
    print(f"   {expression:<13} {floor_result:<7} {mod_result:<6} "
          f"{truncated:<10} {reconstructed} == {numerator}")

print("")
print("21. -7 // 2 is -4, not -3 - it rounds DOWN, not towards zero")
print("22. -7 % 2 is 1, not -1  - the sign follows the DIVISOR")
print("23. int(-7/2) is -3      - int() truncates towards zero")
print("24. The invariant (a//b)*b + (a%b) == a holds in every case")
````

````python
# EXAMPLE 25-27: why the divisor-sign rule is useful.
print("EXAMPLE 25-27: modulo for cyclic values")
print("")

# 25. Wrapping a clock face, including backwards.
print("   25. 12-hour clock, going backwards:")
for hours in [-5, -1, 0, 11, 13, 25]:
    print(f"       hour {hours:>3} -> {hours % 12}")

# 26. Wrapping list indices.
items = ["a", "b", "c"]
print("")
print("   26. Wrapping list indices:")
for index in [-4, -1, 0, 3, 5]:
    print(f"       index {index:>3} -> {items[index % len(items)]}")

# 27. Alternating between two states.
print("")
print("   27. Alternating rows:")
for row in range(5):
    shade = "dark" if row % 2 else "light"
    print(f"       row {row} -> {shade}")

print("")
print("In C, -5 % 12 would be -5, and every one of these would break.")
````

## Exponentiation

````python
# EXAMPLE 28-32: the ** operator.
print("EXAMPLE 28-32: exponentiation")
print("")

# 28. Basic powers, staying exact for integers.
print("   28. 2 ** 10        =", 2 ** 10)
print("       2 ** 100       =", 2 ** 100)
print("       type stays int:", type(2 ** 100).__name__)

# 29. Fractional exponents give roots.
print("")
print("   29. 16 ** 0.5      =", 16 ** 0.5, "<- square root")
print("       27 ** (1/3)    =", 27 ** (1 / 3), "<- cube root")

# 30. Negative exponents force a float.
print("")
print("   30. 2 ** -1        =", 2 ** -1, type(2 ** -1).__name__)
print("       2 ** -3        =", 2 ** -3)

# 31. ** is RIGHT associative - the only one that is.
print("")
print("   31. 2 ** 3 ** 2    =", 2 ** 3 ** 2, "<- 2**(3**2), not (2**3)**2")
print("       (2 ** 3) ** 2  =", (2 ** 3) ** 2)

# 32. Unary minus binds LOOSER than **.
print("")
print("   32. -2 ** 2        =", -2 ** 2, "<- -(2**2), not (-2)**2")
print("       (-2) ** 2      =", (-2) ** 2)
print("")
print("Example 32 is a genuine source of bugs. Bracket it when unsure.")
````

````python
# EXAMPLE 33-35: pow() and its three-argument form.
print("EXAMPLE 33-35: pow()")
print("")

# 33. pow(a, b) is the same as a ** b.
print("   33. pow(2, 10)      =", pow(2, 10))
print("       2 ** 10         =", 2 ** 10)

# 34. The third argument computes a modulus efficiently.
print("")
print("   34. pow(2, 10, 100) =", pow(2, 10, 100))
print("       (2 ** 10) % 100 =", (2 ** 10) % 100, "<- same answer")

# The three-argument form avoids building the huge intermediate value.
import time

start = time.perf_counter()
pow(7, 100000, 13)
fast = time.perf_counter() - start

start = time.perf_counter()
(7 ** 100000) % 13
slow = time.perf_counter() - start

print("")
print("   35. Why the third argument matters:")
print(f"       pow(7, 100000, 13)     took {fast * 1000:.3f} ms")
print(f"       (7 ** 100000) % 13     took {slow * 1000:.3f} ms")
print("       The first never builds the enormous intermediate number.")
print("       This is the basis of RSA and Diffie-Hellman.")
````

## `+` and `*` on non-numbers

These two operators are overloaded for sequences, which is where most of their
everyday use actually lives.

````python
# EXAMPLE 36-42: + and * on sequences.
print("EXAMPLE 36-42: sequence operations")
print("")

# 36-38. Concatenation with +.
print("   36. 'ab' + 'cd'         ->", repr("ab" + "cd"))
print("   37. [1, 2] + [3]        ->", [1, 2] + [3])
print("   38. (1, 2) + (3,)       ->", (1, 2) + (3,))

# 39-41. Repetition with *.
print("")
print("   39. 'ab' * 3            ->", repr("ab" * 3))
print("   40. [0] * 5             ->", [0] * 5)
print("   41. '-' * 20            ->", "-" * 20)

# 42. What does NOT work.
print("")
print("   42. What + refuses:")
for left, right in [("ab", 1), ([1], (2,)), ({1: 2}, {3: 4})]:
    try:
        left + right
        print(f"       {type(left).__name__} + {type(right).__name__} -> worked")
    except TypeError as error:
        print(f"       {type(left).__name__} + {type(right).__name__} ->", error)

print("")
print("Sets and dicts use different operators - covered in 06.5 and Ch13.")
````

````python
# EXAMPLE 43-45: the repetition trap from Chapter 04.
print("EXAMPLE 43-45: * repeats REFERENCES, not objects")
print("")

# 43. Safe with immutable contents.
numbers = [0] * 5
numbers[0] = 99
print("   43. [0] * 5, then set [0] = 99")
print("       ->", numbers, "<- fine, ints are immutable")

# 44. Dangerous with mutable contents.
grid = [[0] * 3] * 3
grid[0][0] = 99
print("")
print("   44. [[0] * 3] * 3, then set [0][0] = 99")
print("       ->", grid, "<- every row changed")
print("       all rows the same object?", grid[0] is grid[1] is grid[2])

# 45. The correct way.
safe_grid = [[0] * 3 for _ in range(3)]
safe_grid[0][0] = 99
print("")
print("   45. [[0] * 3 for _ in range(3)], then set [0][0] = 99")
print("       ->", safe_grid, "<- correct")
print("       all rows the same object?",
      safe_grid[0] is safe_grid[1] is safe_grid[2])
````

## Performance notes

````python
import time

# EXAMPLE 46-48: which forms are faster.
print("EXAMPLE 46-48: performance")
print("")


def time_it(operation, repetitions=200000):
    """Time a repeated operation in milliseconds."""
    start = time.perf_counter()
    for _ in range(repetitions):
        operation()
    return (time.perf_counter() - start) * 1000


# 46. ** vs math.pow for integers.
import math
operator_time = time_it(lambda: 2 ** 10)
function_time = time_it(lambda: math.pow(2, 10))

print("   46. 2 ** 10 vs math.pow(2, 10):")
print(f"       operator: {operator_time:.1f} ms")
print(f"       math.pow: {function_time:.1f} ms")
print("       Prefer ** - it keeps ints exact as well as being faster.")

# 47. divmod vs separate // and %.
both_time = time_it(lambda: divmod(17, 5))
separate_time = time_it(lambda: (17 // 5, 17 % 5))

print("")
print("   47. divmod(a, b) vs (a // b, a % b):")
print(f"       divmod:   {both_time:.1f} ms")
print(f"       separate: {separate_time:.1f} ms")
print("       divmod computes both in one pass.")

# 48. String building.
print("")
print("   48. Building a string in a loop:")

def with_plus():
    """Concatenate repeatedly - creates a new string each time."""
    result = ""
    for index in range(100):
        result = result + str(index)
    return result


def with_join():
    """Collect parts and join once."""
    return "".join(str(index) for index in range(100))


plus_time = time_it(with_plus, 2000)
join_time = time_it(with_join, 2000)

print(f"       repeated + : {plus_time:.1f} ms")
print(f"       ''.join()  : {join_time:.1f} ms")
print("       Strings are immutable, so + builds a new one every pass.")
print("       Covered fully in Chapter 07.")
````

## Takeaways

1. Seven arithmetic operators, each a **dunder method call** underneath —
   which is why `+` works on strings and lists too.
2. Types promote along `int → float → complex`. Python never narrows.
3. `/` **always** returns a float; `//` returns an int only when both operands
   are ints.
4. `//` rounds towards **negative infinity**; `int()` truncates towards **zero**.
5. `%` takes the sign of the **divisor**, which makes it correct for clocks,
   indices and alternation.
6. `(a // b) * b + (a % b) == a` always holds.
7. `**` is **right-associative**, and `-2 ** 2` is `-4` because unary minus binds
   looser.
8. `pow(a, b, m)` is dramatically faster than `(a ** b) % m` — the basis of
   modern cryptography.
9. `*` on a list repeats **references** — `[[0] * 3] * 3` gives three views of one
   row.
