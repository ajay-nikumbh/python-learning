# 06.7 Operator Precedence and Associativity

`2 + 3 * 4` is `14`, not `20`. Everyone knows that one. But `-2 ** 2` is `-4`,
`not a == b` is `not (a == b)`, and `1 & 2 == 2` is `1` — and those catch people
constantly.

**30 numbered examples**, plus the full table.

## Theory

### Two separate rules

**Precedence** decides which operator binds tighter when they differ:

```
2 + 3 * 4    →    2 + (3 * 4)
```

**Associativity** decides the order when precedence is equal:

```
10 - 3 - 2   →    (10 - 3) - 2      left-associative
2 ** 3 ** 2  →    2 ** (3 ** 2)     right-associative
```

Almost everything in Python is **left**-associative. The exceptions are `**` and
assignment.

### The full table, tightest first

| Level | Operators |
|---|---|
| 1 | `()` `[]` `{}` — grouping and literals |
| 2 | `x[i]` `x.attr` `f()` — access and calls |
| 3 | `await` |
| 4 | `**` — **right**-associative |
| 5 | `+x` `-x` `~x` — unary |
| 6 | `*` `/` `//` `%` `@` |
| 7 | `+` `-` |
| 8 | `<<` `>>` |
| 9 | `&` |
| 10 | `^` |
| 11 | `\|` |
| 12 | `in` `not in` `is` `is not` `<` `<=` `>` `>=` `!=` `==` |
| 13 | `not x` |
| 14 | `and` |
| 15 | `or` |
| 16 | `if ... else` — conditional expression |
| 17 | `lambda` |
| 18 | `:=` — **loosest** |

### The three that actually bite

1. **`**` binds tighter than unary minus** — `-2 ** 2` is `-(2**2)` = `-4`
2. **Bitwise binds tighter than comparison** — `1 & 2 == 2` is `1 & (2 == 2)`
3. **`not` binds looser than comparison** — `not a == b` is `not (a == b)`

### The practical rule

Memorising all eighteen levels is not the goal. Know the common ones, and
**bracket anything that is not instantly obvious** to a reader.

````python
# EXAMPLE 1-8: precedence basics.
print("EXAMPLE 1-8: precedence in action")
print("")

cases = [
    ("1. 2 + 3 * 4", 2 + 3 * 4, "2 + (3 * 4)", 2 + (3 * 4)),
    ("2. (2 + 3) * 4", (2 + 3) * 4, "brackets override", (2 + 3) * 4),
    ("3. 10 - 3 - 2", 10 - 3 - 2, "(10 - 3) - 2", (10 - 3) - 2),
    ("4. 2 ** 3 ** 2", 2 ** 3 ** 2, "2 ** (3 ** 2)", 2 ** (3 ** 2)),
    ("5. -2 ** 2", -2 ** 2, "-(2 ** 2)", -(2 ** 2)),
    ("6. (-2) ** 2", (-2) ** 2, "brackets override", (-2) ** 2),
    ("7. 100 / 10 / 2", 100 / 10 / 2, "(100 / 10) / 2", (100 / 10) / 2),
    ("8. 2 * 3 ** 2", 2 * 3 ** 2, "2 * (3 ** 2)", 2 * (3 ** 2)),
]

print("   Expression          Result    Parsed as              Check")
print("   " + "-" * 66)
for label, result, explanation, verify in cases:
    print(f"   {label:<19} {str(result):<9} {explanation:<22} {result == verify}")
````

````python
# EXAMPLE 9-14: associativity.
print("EXAMPLE 9-14: left vs right associativity")
print("")

print("   LEFT-associative (almost everything):")
print("   9.  10 - 3 - 2   =", 10 - 3 - 2, " parsed (10-3)-2 =", (10 - 3) - 2)
print("   10. 100 / 5 / 2  =", 100 / 5 / 2, " parsed (100/5)/2 =", (100 / 5) / 2)
print("   11. 2 // 3 // 4  =", 2 // 3 // 4, " parsed (2//3)//4 =", (2 // 3) // 4)

print("")
print("   RIGHT-associative (only ** and assignment):")
print("   12. 2 ** 3 ** 2  =", 2 ** 3 ** 2, "parsed 2**(3**2) =", 2 ** (3 ** 2))
print("       if it were left: (2**3)**2 =", (2 ** 3) ** 2)

# 13-14. Assignment chains right to left.
print("")
print("   13. a = b = c = 5 binds right to left")
a = b = c = 5
print("       all three are 5:", a, b, c)

print("")
print("   14. why ** is right-associative: it matches mathematics.")
print("       2^3^2 in maths means 2^(3^2) = 2^9 = 512")
````

## The three traps

````python
# EXAMPLE 15-20: trap 1 - unary minus and **.
print("EXAMPLE 15-20: ** binds tighter than unary minus")
print("")

print("   15. -2 ** 2      =", -2 ** 2, "  <- -(2**2)")
print("   16. (-2) ** 2    =", (-2) ** 2, "   <- what most people expect")
print("   17. -2 ** 3      =", -2 ** 3, "  <- -(2**3)")
print("   18. (-2) ** 3    =", (-2) ** 3, "  <- same here, by coincidence")

# 19. It matters in real formulas.
print("")
print("   19. a real case - the quadratic discriminant:")
b_value = 4
wrong = -b_value ** 2
right = (-b_value) ** 2
print(f"       -b ** 2   with b=4 -> {wrong}")
print(f"       (-b) ** 2 with b=4 -> {right}")

# 20. But the exponent side is fine.
print("")
print("   20. on the RIGHT of **, unary minus works normally:")
print("       2 ** -1 =", 2 ** -1)
````

````python
# EXAMPLE 21-25: trap 2 - bitwise vs comparison.
print("EXAMPLE 21-25: bitwise binds TIGHTER than comparison")
print("")

print("   21. 1 & 2 == 2")
print("       parses as 1 & (2 == 2) ->", 1 & (2 == 2))
print("       you probably meant (1 & 2) == 2 ->", (1 & 2) == 2)

print("")
print("   22. flags & READ == READ is a classic bug:")
READ = 0b0001
WRITE = 0b0010
flags = READ | WRITE
print("       flags & READ == READ   ->", flags & READ == READ)
print("       (flags & READ) == READ ->", (flags & READ) == READ)
print("       Both True here by luck. Try with a flag that is absent:")
DELETE = 0b1000
print("       flags & DELETE == DELETE   ->", flags & DELETE == DELETE)
print("       (flags & DELETE) == DELETE ->", (flags & DELETE) == DELETE)

print("")
print("   23. always bracket flag tests: if (flags & FLAG):")
print("   24. or better, use bool(flags & FLAG) ->", bool(flags & READ))
print("   25. or best, use enum.Flag and `in` (see 06.5)")
````

````python
# EXAMPLE 26-30: trap 3 - `not` and comparison.
print("EXAMPLE 26-30: `not` binds LOOSER than comparison")
print("")

a, b = 1, 2

print("   26. not a == b")
print("       parses as not (a == b) ->", not (a == b))
print("       NOT (not a) == b       ->", (not a) == b)

print("")
print("   27. this usually happens to work, which hides the issue:")
print("       not a == b  ->", not a == b)
print("       a != b      ->", a != b, "<- clearer, say what you mean")

print("")
print("   28. but combined with `in` it can confuse:")
items = [1, 2, 3]
print("       not 5 in items    ->", not 5 in items)
print("       5 not in items    ->", 5 not in items, "<- preferred")

print("")
print("   29. and with `is`:")
value = None
print("       not value is None ->", not value is None)
print("       value is not None ->", value is not None, "<- preferred")

print("")
print("   30. Python provides `not in` and `is not` as single operators")
print("       precisely so you never need the ambiguous form.")
````

## Takeaways

1. **Precedence** decides which operator binds tighter; **associativity** decides
   ties.
2. Almost everything is **left**-associative. `**` and assignment are **right**.
3. `-2 ** 2` is `-4` — `**` binds tighter than unary minus.
4. `1 & 2 == 2` parses as `1 & (2 == 2)` — bitwise binds **tighter** than
   comparison. Always bracket flag tests.
5. `not a == b` is `not (a == b)` — `not` binds **looser** than comparison.
6. Prefer `x not in items` and `x is not None` over `not x in items` and
   `not x is None`.
7. The practical rule: **bracket anything not instantly obvious**. Brackets cost
   nothing.
