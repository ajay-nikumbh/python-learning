## Section C — Find and fix the bug

```python
print("C1. Checking membership against several values")
print("")

value = 5

print("   BROKEN: value == 1 or 2 or 3")
print("      ->", value == 1 or 2 or 3)
print("      parses as (value == 1) or 2 or 3, which is truthy for ANY value")

print("")
print("   FIXED: value in (1, 2, 3)")
print("      ->", value in (1, 2, 3))

# Prove the broken version is always truthy.
print("")
print("   proof the broken form never returns False:")
for candidate in [1, 5, 99]:
    print(f"      candidate={candidate:<4} -> {candidate == 1 or 2 or 3}")
```

```python
print("C2. A default that swallows zero")
print("")


def set_timeout_broken(timeout=None):
    """Apply a default - broken for 0."""
    return timeout or 30


def set_timeout_fixed(timeout=None):
    """Apply a default only when genuinely absent."""
    return 30 if timeout is None else timeout


print("   Input    Broken   Fixed")
print("   " + "-" * 30)
for timeout in [None, 10, 0]:
    print(f"   {str(timeout):<8} {set_timeout_broken(timeout):<8} "
          f"{set_timeout_fixed(timeout)}")

print("")
print("   BUG: timeout=0 means 'do not wait', not 'use the default'.")
```

```python
print("C3. A flag test without brackets")
print("")

READ = 0b0001
WRITE = 0b0010
DELETE = 0b1000

flags = READ | WRITE

print("   flags has READ and WRITE, not DELETE")
print("")
print("   BROKEN: flags & DELETE == DELETE")
print("      ->", flags & DELETE == DELETE, "<- claims DELETE is present")
print("      parses as flags & (DELETE == DELETE) = flags & True = flags & 1")

print("")
print("   FIXED: (flags & DELETE) == DELETE")
print("      ->", (flags & DELETE) == DELETE)
print("   ALSO FIXED: bool(flags & DELETE)")
print("      ->", bool(flags & DELETE))
```

```python
print("C4. A function that damages its argument")
print("")


def add_tax_broken(prices):
    """Use += on a list parameter - modifies the caller's data."""
    prices += [sum(prices) * 0.18]
    return prices


def add_tax_fixed(prices):
    """Build a new list - leaves the caller's data alone."""
    return prices + [sum(prices) * 0.18]


original = [100.0, 200.0]
add_tax_broken(original)
print("   BROKEN - after the call, the caller's list is:", original)

original = [100.0, 200.0]
result = add_tax_fixed(original)
print("")
print("   FIXED  - caller's list:", original)
print("            returned:     ", result)
```

## Section D — Precedence drills

Work each out on paper first, then run.

```python
print("D1-D10: add the brackets")
print("")

drills = [
    ("D1.  2 + 3 * 4", 2 + 3 * 4, "2 + (3 * 4)"),
    ("D2.  2 * 3 ** 2", 2 * 3 ** 2, "2 * (3 ** 2)"),
    ("D3.  -2 ** 2", -2 ** 2, "-(2 ** 2)"),
    ("D4.  10 - 3 - 2", 10 - 3 - 2, "(10 - 3) - 2"),
    ("D5.  2 ** 3 ** 2", 2 ** 3 ** 2, "2 ** (3 ** 2)"),
    ("D6.  1 | 2 & 3", 1 | 2 & 3, "1 | (2 & 3)"),
    ("D7.  1 << 2 + 3", 1 << 2 + 3, "1 << (2 + 3)"),
    ("D8.  True or False and False", True or False and False,
     "True or (False and False)"),
    ("D9.  not True == False", not True == False, "not (True == False)"),
    ("D10. 1 + 2 > 2 and 3 < 4", 1 + 2 > 2 and 3 < 4,
     "((1+2) > 2) and (3 < 4)"),
]

print("   Expression                Result   Parsed as")
print("   " + "-" * 62)
for label, result, parsed in drills:
    print(f"   {label:<25} {str(result):<8} {parsed}")
```
