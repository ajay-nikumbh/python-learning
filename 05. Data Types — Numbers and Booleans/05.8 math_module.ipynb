# 05.8 The `math` Module

`math` wraps the C standard library's mathematical functions. It is fast, always
available, and covers most of what everyday code needs.

## Theory

### What `math` is for

Functions on **real numbers**. Three related modules exist:

| Module | For | Notes |
|---|---|---|
| `math` | Real numbers | Rejects complex input |
| `cmath` | Complex numbers | See 05.3 |
| `statistics` | Data summaries | Mean, median, stdev |

### Why some functions duplicate operators

`math.pow(2, 3)` and `2 ** 3` both compute 8 — but not identically:

- `**` keeps integers exact and supports arbitrary precision
- `math.pow()` always converts to `float`, so it loses precision on large values

Prefer the operator unless you specifically want float semantics.

### The precision-preserving functions

Several functions exist purely to avoid the float errors from 05.2:

- `math.fsum()` — accurate summation
- `math.hypot()` — avoids overflow computing the hypotenuse
- `math.expm1()` / `math.log1p()` — accurate near zero
- `math.isclose()` — the correct float comparison

Knowing these exist is the main value of this notebook.

```python
import math

# Constants.
print("Constants:")
print("   math.pi   =", math.pi)
print("   math.e    =", math.e)
print("   math.tau  =", math.tau, "<- 2*pi")
print("   math.inf  =", math.inf)
print("   math.nan  =", math.nan)

# Rounding family - each rounds differently.
print("")
print("Rounding, compared:")
print("   Value    floor   ceil    trunc   round")
print("   " + "-" * 44)
for value in [3.7, 3.2, -3.7, -3.2, 2.5, 3.5]:
    print(f"   {value:<8} {math.floor(value):<7} {math.ceil(value):<7} "
          f"{math.trunc(value):<7} {round(value)}")

print("")
print("floor goes down, ceil goes up, trunc goes towards zero,")
print("and round uses banker's rounding (halves to even).")
```

```python
import math

# Powers, roots and logarithms.
print("Powers and roots:")
print("   math.sqrt(16)    ->", math.sqrt(16))
print("   math.isqrt(17)   ->", math.isqrt(17), "<- integer square root, exact")
print("   math.cbrt(27)    ->", math.cbrt(27) if hasattr(math, "cbrt") else "3.12+ only")
print("   math.pow(2, 10)  ->", math.pow(2, 10), "<- always a float")
print("   2 ** 10          ->", 2 ** 10, "<- stays an int")

# The difference matters for large values.
print("")
print("Why ** is usually better:")
print("   2 ** 100      ->", 2 ** 100)
print("   math.pow(2,100)->", math.pow(2, 100), "<- precision lost")

print("")
print("Logarithms:")
print("   math.log(math.e) ->", math.log(math.e), "<- natural log")
print("   math.log(100, 10)->", math.log(100, 10))
print("   math.log10(1000) ->", math.log10(1000), "<- more accurate")
print("   math.log2(1024)  ->", math.log2(1024))
```

## The precision-preserving functions

These exist specifically to work around the float issues from 05.2.

```python
import math

# fsum is exact where sum() accumulates error.
values = [0.1] * 10
print("Accurate summation:")
print("   sum([0.1] * 10)       ->", sum(values))
print("   math.fsum([0.1] * 10) ->", math.fsum(values))

# A harder case with mixed magnitudes.
mixed = [1e16, 1.0, -1e16]
print("")
print("With mixed magnitudes", mixed, ":")
print("   sum()       ->", sum(mixed), "<- the 1.0 was lost")
print("   math.fsum() ->", math.fsum(mixed), "<- correct")

# hypot avoids overflow.
print("")
print("Hypotenuse without overflow:")
large = 1e200
print("   naive sqrt(x*x + y*y) ->", end=" ")
try:
    naive = math.sqrt(large * large + large * large)
    print(naive)
except OverflowError as error:
    print("OverflowError:", error)
print("   math.hypot(x, y)      ->", math.hypot(large, large))

# expm1 and log1p are accurate near zero.
tiny = 1e-16
print("")
print("Accuracy near zero:")
print("   math.exp(1e-16) - 1 ->", math.exp(tiny) - 1, "<- catastrophic loss")
print("   math.expm1(1e-16)   ->", math.expm1(tiny), "<- correct")
print("   math.log(1 + 1e-16) ->", math.log(1 + tiny))
print("   math.log1p(1e-16)   ->", math.log1p(tiny), "<- correct")
```

## Integer-specific functions

```python
import math

print("Exact integer operations:")
print("   math.factorial(20) ->", math.factorial(20))
print("   math.gcd(48, 18)   ->", math.gcd(48, 18))
print("   math.lcm(4, 6)     ->", math.lcm(4, 6))
print("   math.comb(5, 2)    ->", math.comb(5, 2), "<- combinations, 5 choose 2")
print("   math.perm(5, 2)    ->", math.perm(5, 2), "<- permutations")
print("   math.isqrt(1000)   ->", math.isqrt(1000), "<- floor of the square root")

# isqrt is exact where sqrt is not.
big = 10 ** 40
print("")
print("Why isqrt matters for large integers:")
print("   math.isqrt(10**40) ->", math.isqrt(big))
print("   int(math.sqrt(10**40)) ->", int(math.sqrt(big)), "<- wrong")
print("   isqrt squared == original?", math.isqrt(big) ** 2 == big)

# gcd and lcm accept several arguments.
print("")
print("Multiple arguments:")
print("   math.gcd(12, 18, 24) ->", math.gcd(12, 18, 24))
print("   math.lcm(2, 3, 4)    ->", math.lcm(2, 3, 4))
```

## Trigonometry

All trigonometric functions work in **radians**, not degrees.

```python
import math

# Conversion is explicit.
print("Degrees and radians:")
print("   math.radians(180) ->", math.radians(180))
print("   math.degrees(math.pi) ->", math.degrees(math.pi))

print("")
print("Angle    sin      cos      tan")
print("-" * 40)
for degrees in [0, 30, 45, 60, 90]:
    radians = math.radians(degrees)
    sine = math.sin(radians)
    cosine = math.cos(radians)

    # tan(90) is undefined and produces an enormous number.
    tangent = math.tan(radians)
    tangent_text = f"{tangent:8.4f}" if abs(tangent) < 1e10 else "   huge"

    print(f"{degrees:>5}    {sine:7.4f}  {cosine:7.4f}  {tangent_text}")

print("")
print("Note sin(180 degrees) is not exactly zero:")
print("   math.sin(math.radians(180)) ->", math.sin(math.radians(180)))
print("   This is float precision (05.2), not a bug in math.")
print("   Use math.isclose(value, 0, abs_tol=1e-15) to test it.")

# The practical one: atan2 handles all four quadrants.
print("")
print("atan2 gets the quadrant right where atan cannot:")
for x_value, y_value in [(1, 1), (-1, 1), (-1, -1), (1, -1)]:
    angle = math.degrees(math.atan2(y_value, x_value))
    print(f"   point ({x_value:>2}, {y_value:>2}) -> {angle:>7.1f} degrees")
```

## Takeaways

1. `math` covers **real numbers** only. Use `cmath` for complex, `statistics` for
   data summaries.
2. Prefer `**` over `math.pow()` — the operator keeps integers exact, `math.pow`
   always returns a float.
3. `math.isqrt()` is exact for large integers where `int(math.sqrt(n))` is wrong.
4. The precision-preserving family is the real value here: `fsum`, `hypot`,
   `expm1`, `log1p`, `isclose`.
5. `floor`, `ceil`, `trunc` and `round` all round **differently** — know which
   you need.
6. Trigonometric functions use **radians**; convert with `radians()` and
   `degrees()`.
7. `atan2(y, x)` handles all four quadrants; `atan(y/x)` cannot.
8. `gcd`, `lcm`, `comb`, `perm` and `factorial` are exact integer functions.
