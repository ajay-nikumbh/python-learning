# 01.2 How Computers Execute Code

Everything a computer stores is a number. Everything a CPU does is
fetch an instruction, decode it, execute it, repeat.

This file makes those invisible facts visible.


## PART 1: Everything is bits


```python
# A bit is a single 0 or 1. Eight bits make a byte.
# Let's take an ordinary number and look at how it is stored.
number = 65

# bin() shows the binary form. The "0b" prefix just marks it as binary.
print("The number", number, "in binary is:", bin(number))

# format() with "08b" pads the binary out to a full 8 bits (one byte),
# which is how it actually sits in memory.
print("As a full byte:", format(number, "08b"))
```

## PART 2: The same bits mean different things


```python
# chr() takes a number and interprets it as a text character.
# The number 65 interpreted as text is the letter "A".
print("65 interpreted as a NUMBER:", 65)
print("65 interpreted as TEXT:    ", chr(65))

# ord() does the reverse - it shows the number behind a character.
print("The letter 'A' is really the number:", ord("A"))

# This is the key insight: the hardware stores 01000001 either way.
# Nothing in the bits says "this is text" or "this is a number".
# The TYPE of a value is what keeps that interpretation straight, which
# is why Python cares about types at all.
print("Same bits (01000001), two meanings - the type decides which.")
```

## PART 3: Memory - values live at addresses


```python
# When we create a value, Python puts it somewhere in RAM.
age = 25
city = "Mumbai"

# id() returns the memory address where the object lives.
# The exact numbers differ every run - that is expected and fine.
# What matters is that each value HAS a location.
print("The value 25 lives at memory address:", id(age))
print("The text 'Mumbai' lives at address:  ", id(city))

# A variable name is a LABEL pointing at an address - not a box holding
# a value. Proof: point a second name at the same object.
same_age = age

# Both names now report the identical address, because both point at
# the one object. Nothing was copied.
print("age      points to address:", id(age))
print("same_age points to address:", id(same_age))
print("Same address? ", id(age) == id(same_age))
print("Nothing was copied - two labels, one object. More in Chapter 04.")
```

## PART 4: How much memory does a value take?


```python
# We import sys to use getsizeof(), which reports an object's size in bytes.
import sys

# Even a tiny integer carries bookkeeping overhead in Python, because
# every value is a full object, not a bare machine number.
print("Bytes used by the integer 0:      ", sys.getsizeof(0))
print("Bytes used by the integer 1000000:", sys.getsizeof(1000000))
print("Bytes used by the text 'hi':      ", sys.getsizeof("hi"))
print("Bytes used by an empty list:      ", sys.getsizeof([]))

# Bigger numbers genuinely need more space, because Python integers can
# grow without limit (covered in Chapter 05).
huge_number = 10 ** 100
print("Bytes used by 10 to the power 100:", sys.getsizeof(huge_number))
```

## PART 5: The fetch-decode-execute cycle, made visible


```python
# The CPU repeats one loop: fetch the next instruction, decode what it
# means, execute it, move on. Python works the same way over bytecode.

# We import dis, the disassembler, to see the actual instruction list.
import dis


# This function is deliberately tiny so its instruction list stays readable.
def add_two_numbers(first, second):
    # One line of Python that becomes several machine-level steps.
    result = first + second
    return result


# dis.dis() prints the bytecode instructions Python generated.
# Read the output as the CPU would: one instruction per line, in order.
print("The Python source was ONE line: result = first + second")
print("Here are the instructions it actually became:")
dis.dis(add_two_numbers)

# Notice the pattern in the output above:
#   LOAD_FAST  - fetch a value from memory into the working area
#   BINARY_OP  - execute the addition
#   STORE_FAST - write the result back to memory
#   RETURN_VALUE - hand the answer back
# Load, operate, store. That is what all computation looks like underneath.
```

## PART 6: The speed gap between memory and storage


```python
# We import time so we can actually measure the difference.
import time

# Build a list in memory to read from.
numbers_in_memory = list(range(100000))

# Record the time before reading from memory.
start_time = time.perf_counter()

# Read every value from memory. This is the fast path.
total = 0
for number in numbers_in_memory:
    total = total + number

# Record the time after, and work out how long it took.
memory_duration = time.perf_counter() - start_time

print("Summing 100,000 numbers held in MEMORY took:")
print("   ", round(memory_duration * 1000, 3), "milliseconds")
print("Reading the same data from disk would be roughly 100x slower,")
print("and from a network, thousands of times slower. This is why")
print("programs load data into memory instead of re-reading files.")
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. All data is bits; 8 bits make a byte.")
print("2. The same bits mean different things depending on interpretation.")
print("3. Values live at memory addresses; names are labels pointing at them.")
print("4. Every value occupies measurable space - check with sys.getsizeof().")
print("5. The CPU loops: fetch, decode, execute - see it with dis.dis().")
print("6. Memory is far faster than disk, which shapes how we write code.")
```
