# 01.5 Programming Paradigms

A paradigm is a style of organising code. Python supports several and
forces none.

To make the comparison fair, this file solves THE SAME PROBLEM four times,
once in each paradigm.

THE PROBLEM: given a list of employees, find the total salary cost of
everyone in the engineering department.

Do not worry about understanding every line yet - the chapters that teach
each style are noted. The goal here is to recognise the shapes.


## THE SHARED DATA


```python
# Each employee is a dictionary holding three facts about one person.
# Dictionaries are covered properly in Chapter 14.
employees = [
    {"name": "Asha", "department": "engineering", "salary": 95000},
    {"name": "Ben", "department": "sales", "salary": 62000},
    {"name": "Chen", "department": "engineering", "salary": 88000},
    {"name": "Dara", "department": "marketing", "salary": 71000},
    {"name": "Eli", "department": "engineering", "salary": 104000},
]
```

## PARADIGM 1: PROCEDURAL - describe the steps


```python
# Procedural code says HOW to do the job, step by step, and groups
# those steps into procedures (functions). Taught in Chapters 09, 10, 16.


def total_engineering_salary_procedural(employee_list):
    """Add up engineering salaries by walking the list step by step."""
    # Start the running total at zero before we begin.
    running_total = 0

    # Visit each employee in turn.
    for employee in employee_list:
        # Check whether this person is in engineering.
        if employee["department"] == "engineering":
            # They are, so add their salary to the running total.
            running_total = running_total + employee["salary"]

    # Hand back the finished total.
    return running_total


# Call it and show the result.
print("Total engineering salary:", total_engineering_salary_procedural(employees))
print("Style: explicit steps - set up a total, loop, test, accumulate.")
print("Good for: scripts, straightforward logic, anything you read top to bottom.")
```

## PARADIGM 2: OBJECT-ORIENTED - model the problem as things


```python
# OOP bundles data together with the behaviour that acts on it.
# Instead of loose values and separate functions, you get objects that
# know their own data. Taught in Chapters 24 to 28.


class Employee:
    """One employee: their own data plus their own behaviour."""

    def __init__(self, name, department, salary):
        # Store this employee's own facts on the object itself.
        self.name = name
        self.department = department
        self.salary = salary

    def works_in(self, department_name):
        """Ask this employee whether they work in a given department."""
        # The object answers a question about itself.
        return self.department == department_name


class Company:
    """A company that holds employees and can answer questions about them."""

    def __init__(self, employee_objects):
        # The company owns its list of employees.
        self.employees = employee_objects

    def total_salary_for(self, department_name):
        """Add up salaries for everyone in one department."""
        running_total = 0

        # Ask each employee object whether it belongs to the department.
        for employee in self.employees:
            if employee.works_in(department_name):
                running_total = running_total + employee.salary

        return running_total


# Build Employee objects from the raw dictionaries.
employee_objects = []
for record in employees:
    # Create one object per record and collect it.
    employee_objects.append(
        Employee(record["name"], record["department"], record["salary"])
    )

# Wrap them in a Company object that knows how to answer our question.
company = Company(employee_objects)

print("Total engineering salary:", company.total_salary_for("engineering"))
print("Style: data and behaviour live together inside objects.")
print("Good for: domains with clear entities - users, orders, accounts.")
```

## PARADIGM 3: FUNCTIONAL - transform data without changing it


```python
# Functional code builds results by transforming data through functions,
# avoiding changing anything in place. Taught in Chapters 17 and 31.


def is_engineer(employee):
    """Return True if this employee is in engineering."""
    # A small, pure function: same input always gives the same output,
    # and it changes nothing outside itself.
    return employee["department"] == "engineering"


def get_salary(employee):
    """Return this employee's salary."""
    # Another pure function - it only reads, never writes.
    return employee["salary"]


# filter() keeps only the employees that is_engineer() approves of.
# It produces a NEW collection; the original list is untouched.
engineers = filter(is_engineer, employees)

# map() turns each employee into just their salary number.
engineer_salaries = map(get_salary, engineers)

# sum() collapses those salaries into one total.
functional_total = sum(engineer_salaries)

print("Total engineering salary:", functional_total)

# Prove the original data was never modified.
print("Original list still has", len(employees), "employees - nothing changed.")
print("Style: filter, map, reduce - build new values instead of mutating old ones.")
print("Good for: data pipelines and transformations. Fewer moving parts, fewer bugs.")
```

## PARADIGM 4: DECLARATIVE - describe the result you want


```python
# Declarative code states WHAT you want rather than HOW to compute it.
# In Python, comprehensions are the everyday declarative tool.
# Taught in Chapter 15.

# Read this out loud: "the salary of each employee, for each employee
# in the list, where the department is engineering". That sentence IS
# the code - no counter, no accumulator, no loop mechanics.
declarative_total = sum(
    employee["salary"]
    for employee in employees
    if employee["department"] == "engineering"
)

print("Total engineering salary:", declarative_total)
print("Style: describe the result; let Python handle the looping.")
print("Good for: anywhere it reads more clearly than a manual loop.")
```

## ALL FOUR AGREE


```python
print("Same problem, four styles, one answer")

# Collect all four results to confirm they match.
procedural_result = total_engineering_salary_procedural(employees)
oop_result = company.total_salary_for("engineering")

print("Procedural: ", procedural_result)
print("OOP:        ", oop_result)
print("Functional: ", functional_total)
print("Declarative:", declarative_total)

# Check they are genuinely identical, rather than trusting our eyes.
all_match = (
    procedural_result == oop_result == functional_total == declarative_total
)
print("All four agree?", all_match)
```

## WHICH ONE SHOULD YOU USE?


```python
print("Choosing a paradigm")
print("Do not pick a team. Real Python mixes all four, often in one file.")
print("  PROCEDURAL  - scripts and plain logic you read top to bottom")
print("  OOP         - when you have entities with state AND behaviour")
print("  FUNCTIONAL  - data transformation pipelines")
print("  DECLARATIVE - whenever it simply reads more clearly")
print("For THIS problem the declarative version is the best Python:")
print("it is the shortest and states the intent most directly.")
print("For a system with hundreds of employee rules, OOP would win.")
print("The skill is matching the style to the problem - not loyalty.")
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. A paradigm is a style of organising code, not a set of features.")
print("2. Python is multi-paradigm - it supports several and forces none.")
print("3. Procedural describes HOW; declarative describes WHAT.")
print("4. OOP bundles data with the behaviour that acts on it.")
print("5. Functional avoids changing state, so less can go wrong.")
print("6. Good code mixes paradigms to match each problem.")
```
