# 01.7 Where Python is Used

Python is not the fastest language, nor the safest. It is the most
broadly useful, and its libraries are its real advantage.

This file surveys the domains honestly - including where Python is the
wrong tool, because knowing that is part of engineering judgement.


## PART 1: The domains where Python leads


```python
# Each entry is (domain, key libraries, why Python won there).
# We store them as tuples in a list so we can loop and print consistently.
strong_domains = [
    (
        "Data science, ML and AI",
        "NumPy, pandas, scikit-learn, PyTorch, TensorFlow",
        "the heavy maths runs in C/C++; Python is the steering wheel",
    ),
    (
        "Automation and scripting",
        "os, pathlib, shutil, subprocess",
        "the original use case - shortest path from problem to solution",
    ),
    (
        "Web backends",
        "Django, Flask, FastAPI",
        "Instagram, Spotify and Dropbox run substantial Python",
    ),
    (
        "DevOps and infrastructure",
        "Ansible, boto3, cloud SDKs",
        "deployment tooling leans on it heavily; Ansible IS Python",
    ),
    (
        "Scientific computing",
        "SciPy, Astropy, Biopython, Jupyter",
        "notebooks are the standard medium for computational research",
    ),
    (
        "Testing",
        "pytest, Selenium, Playwright",
        "widely used to test systems written in OTHER languages",
    ),
]

# Print each domain as a small block.
for domain, libraries, reason in strong_domains:
    print("")
    print("  " + domain)
    print("     libraries:", libraries)
    print("     why:      ", reason)
```

## PART 2: Where Python is used but does not lead


```python
# Same structure, different verdict: (domain, what exists, the catch).
mixed_domains = [
    (
        "Desktop GUI apps",
        "Tkinter, PyQt, Kivy",
        "works well, but packaging and distribution are awkward",
    ),
    (
        "Game development",
        "Pygame, Panda3D",
        "excellent for learning; commercial games use C++ or C#",
    ),
    (
        "Mobile apps",
        "Kivy, BeeWare",
        "they exist, but Swift and Kotlin dominate the platforms",
    ),
]

for domain, what_exists, catch in mixed_domains:
    print("")
    print("  " + domain)
    print("     available:", what_exists)
    print("     the catch:", catch)
```

## PART 3: Where you should pick something else


```python
# Being honest about limits is part of the job. A developer who claims
# their favourite language is right for everything cannot be trusted
# to choose tools.
wrong_tool = [
    (
        "Systems programming (OS, drivers)",
        "C, C++, Rust",
        "needs direct memory control and zero runtime overhead",
    ),
    (
        "Hard real-time embedded work",
        "C, Rust",
        "garbage collection can pause at an unpredictable moment",
    ),
    (
        "Browser frontend",
        "JavaScript, TypeScript",
        "the only language browsers natively run",
    ),
    (
        "Maximum-throughput CPU work",
        "C++, Rust, Go",
        "or call into them FROM Python, which is the usual answer",
    ),
]

for domain, use_instead, reason in wrong_tool:
    print("")
    print("  " + domain)
    print("     use instead:", use_instead)
    print("     because:    ", reason)
```

## PART 4: A real automation example


```python
# This is the kind of task Python gets used for constantly: a boring,
# repetitive job that would take an hour by hand.

# A list of messy filenames, as they might come off a camera or scanner.
messy_filenames = [
    "  Holiday Photo 1.JPG  ",
    "Holiday Photo 2.JPG",
    "  invoice MARCH.pdf",
    "Invoice April.PDF  ",
]

print("Before cleaning:")
for filename in messy_filenames:
    # repr() shows the quotes and spaces, making the mess visible.
    print("   ", repr(filename))

# Clean every name with a few clear steps.
cleaned_filenames = []
for filename in messy_filenames:
    # Remove the leading and trailing spaces.
    cleaned = filename.strip()

    # Make everything lowercase so the names are consistent.
    cleaned = cleaned.lower()

    # Replace spaces with underscores so the names are shell-friendly.
    cleaned = cleaned.replace(" ", "_")

    # Collect the cleaned-up name.
    cleaned_filenames.append(cleaned)

print("After cleaning:")
for filename in cleaned_filenames:
    print("   ", repr(filename))

print("Four files here; the same six lines handle forty thousand.")
print("That is the automation case in one example.")
```

## PART 5: A real data analysis example


```python
# Monthly sales figures, as you might read from a spreadsheet.
monthly_sales = [
    ("January", 45000),
    ("February", 52000),
    ("March", 48000),
    ("April", 61000),
    ("May", 58000),
    ("June", 67000),
]

# Pull just the numbers out so we can do arithmetic on them.
sales_figures = []
for month_name, amount in monthly_sales:
    sales_figures.append(amount)

# Total sales across the whole period.
total_sales = sum(sales_figures)

# Average sales per month.
average_sales = total_sales / len(sales_figures)

# The best and worst months. max() and min() find the extreme values.
best_month_amount = max(sales_figures)
worst_month_amount = min(sales_figures)

# Growth from the first month to the last, as a percentage.
first_month = sales_figures[0]
last_month = sales_figures[-1]
growth_percent = ((last_month - first_month) / first_month) * 100

print("Total sales:  ", total_sales)
print("Average month:", round(average_sales, 2))
print("Best month:   ", best_month_amount)
print("Worst month:  ", worst_month_amount)
print("Growth:       ", round(growth_percent, 1), "percent")

print("All of that used zero installed libraries. With pandas it would be")
print("three lines - which is why data work gravitated to Python.")
```

## PART 6: The honest summary


```python
print("Python's niche is problems where DEVELOPER time matters more than")
print("MACHINE time - which is most problems, most of the time.")
print("When machine time genuinely dominates, Python is still often the")
print("right choice as the orchestration layer wrapped around fast native")
print("libraries. NumPy is Python on the outside and C on the inside.")
print("Pick the tool for the problem. Python is a very wide tool, but it")
print("is not every tool, and saying so is a sign of judgement.")
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. Python leads in data/ML, automation, web backends and DevOps.")
print("2. Its libraries, not its syntax, are the real advantage.")
print("3. It is usable but not dominant for GUI, games and mobile.")
print("4. Use C, C++ or Rust for systems, real-time and max-throughput work.")
print("5. Python's niche: developer time matters more than machine time.")
print("6. For heavy numeric work, Python orchestrates fast native libraries.")
```
