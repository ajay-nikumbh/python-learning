## SECTION B: Predict the output


```python
# B1 - order of operations.
print("B1. What does this print?")
print("    x = 10")
print("    x = 20")
print("    print(x)")

# Run it so the answer is demonstrated rather than asserted.
x = 10
x = 20
print("    RESULT:", x)
print("    WHY: the second assignment repointed the name. The first value")
print("         is simply no longer referenced.")

# B2 - the same bits, two interpretations.
print("B2. What do these two lines print?")
print("    print(65)")
print("    print(chr(65))")
print("    RESULT:", 65, "and", chr(65))
print("    WHY: identical bits (01000001). The TYPE decides whether they")
print("         are read as a number or as a character.")

# B3 - strong typing.
print("B3. Does this work: 5 + ' apples' ?")

# Catch the error so the file keeps running.
try:
    broken = 5 + " apples"
    print("    RESULT:", broken)
except TypeError:
    print("    RESULT: TypeError - it does NOT work.")
    print("    WHY: Python is strongly typed and will not guess. You must")
    print("         convert explicitly: str(5) + ' apples'.")

# B4 - integer division vs true division.
print("B4. What is the difference between 7 / 2 and 7 // 2 ?")
print("    7 / 2  gives", 7 / 2, "(true division, always a float)")
print("    7 // 2 gives", 7 // 2, "(floor division, rounds down)")
print("    WHY: two different operators for two different questions.")
print("         Covered fully in Chapter 06.")
```

## SECTION C: Pseudocode practice


```python
# C1 - a worked example to show the expected format.
print("C1. WORKED EXAMPLE - count the even numbers in a list.")
print("    PSEUDOCODE:")
print("      START")
print("        SET count TO 0")
print("        FOR each number in the list")
print("          IF the number divides by 2 with no remainder THEN")
print("            ADD 1 to count")
print("        RETURN count")
print("      END")
print("    TRANSLATION:")

# The translation, line for line against the pseudocode above.
numbers = [4, 7, 2, 9, 10, 3, 8]

# PSEUDOCODE: "SET count TO 0"
count = 0

# PSEUDOCODE: "FOR each number in the list"
for number in numbers:
    # PSEUDOCODE: "IF the number divides by 2 with no remainder"
    # The % operator gives the remainder after division.
    if number % 2 == 0:
        # PSEUDOCODE: "ADD 1 to count"
        count = count + 1

print("      numbers:", numbers)
print("      even count:", count)

# C2, C3, C4 - for the reader to do on paper.
print("C2. YOUR TURN (on paper first): write pseudocode to find the")
print("    SMALLEST number in a list. Then translate it.")
print("C3. YOUR TURN: write pseudocode to count how many words in a")
print("    sentence are longer than 4 characters.")
print("C4. YOUR TURN: write pseudocode to decide whether a word is a")
print("    palindrome (reads the same forwards and backwards).")
```
