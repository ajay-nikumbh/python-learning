# 01.6 What is Python

Python is a high-level, general-purpose, dynamically typed, interpreted
language that values readability above almost everything else.

This file inspects the Python you are actually running, and demonstrates
what its design philosophy means in practice.


## PART 1: Which Python are you running?


```python
# sys gives us information about the interpreter running this file.
import sys

# version_info is a structured tuple, best for comparing versions in code.
version = sys.version_info
print("Version:", version.major, ".", version.minor, ".", version.micro)

# sys.version is the full human-readable string, including build details.
print("Full version string:")
print("   ", sys.version)

# The path to the actual python executable running right now. Useful when
# you have several Pythons installed and need to know which one you got.
print("Executable location:")
print("   ", sys.executable)

# This course targets 3.12 or newer. Check and report honestly.
if version.major == 3 and version.minor >= 12:
    print("")
    print("Good - this course targets Python 3.12+ and you are on it.")
else:
    print("")
    print("Note: this course targets Python 3.12+. Some later chapters")
    print("use features your version may not have.")
```

## PART 2: Dynamically typed - types belong to values, not names


```python
# In a statically typed language you declare a type up front:
#     int count = 5;      <- C, Java: this name holds an int, forever
#
# In Python you never declare a type. The VALUE carries the type, and
# the same name can point at different types over time.

# Start by pointing the name at an integer.
flexible_value = 42
print("flexible_value =", flexible_value, "  type:", type(flexible_value).__name__)

# Point the same name at text instead. Completely legal.
flexible_value = "now I am text"
print("flexible_value =", flexible_value, "  type:", type(flexible_value).__name__)

# And now at a list.
flexible_value = [1, 2, 3]
print("flexible_value =", flexible_value, "  type:", type(flexible_value).__name__)

print("Convenient, but it cuts both ways: nothing stops you passing text")
print("to a function expecting a number, and the error only shows up when")
print("that line runs. Type hints (Chapter 32) add the safety back.")
```

## PART 3: Strongly typed - Python still refuses nonsense


```python
# Python is dynamically typed but STRONGLY typed: it will not silently
# guess what you meant when the types do not fit together.

# We use try/except (Chapter 23) to catch the error rather than crash.
try:
    # In JavaScript this quietly produces "5 apples". Python refuses.
    nonsense = 5 + " apples"
except TypeError as error:
    print("Trying 5 + ' apples' gives a TypeError:")
    print("   ", error)

print("This is the Zen line 'errors should never pass silently' in action.")
print("If you want the two joined, you must say so explicitly:")

# Explicit conversion - we state exactly what we want to happen.
explicit_result = str(5) + " apples"
print("   str(5) + ' apples' =", explicit_result)
```

## PART 4: The Zen of Python


```python
# `import this` prints Tim Peters' 19 aphorisms - the language's design
# philosophy, shipped inside the language itself as an easter egg.
print("Running `import this`:")

# Importing the module is what triggers the printing.
import this

# The module also stores the text, so we can point at specific lines.
print("The ones that change how you write code day to day:")
print("   Explicit is better than implicit.")
print("      -> say what you mean; hidden magic costs more than it saves")
print("   Readability counts.")
print("      -> code is read far more often than it is written")
print("   Errors should never pass silently.")
print("      -> never write a bare `except: pass` (Chapter 23)")
print("   If the implementation is hard to explain, it's a bad idea.")
print("      -> if you cannot describe your function in one sentence, rewrite it")
```

## PART 5: What "Pythonic" means


```python
# Both versions below produce identical output. One follows Python's
# idioms; the other transliterates habits from C-style languages.

fruits = ["apple", "banana", "cherry"]

# NOT PYTHONIC: managing an index counter by hand.
print("Not Pythonic - manual index management:")
index = 0
while index < len(fruits):
    # Reach into the list using the counter we are maintaining ourselves.
    print("   ", index, fruits[index])
    # Remember to move the counter, or loop forever.
    index = index + 1

# PYTHONIC: say what you actually mean - look at each item.
print("Pythonic - say what you mean:")
for position, fruit in enumerate(fruits):
    # enumerate() hands us the position and the item together.
    # No counter to maintain, no off-by-one error possible.
    print("   ", position, fruit)

print("Both work. The second is shorter, safer, and states the intent.")
print("You are not expected to have this instinct yet - it builds over time.")
```

## PART 6: Batteries included


```python
# Python ships with a large standard library, so a great deal is possible
# with no installation at all. This is the "batteries included" philosophy.

# Count how many modules are built directly into the interpreter.
print("Modules compiled into this interpreter:", len(sys.builtin_module_names))

# Count the names available with no import whatsoever.
print("Built-in names available with zero imports:", len(dir(__builtins__)))

print("A few standard library modules you will meet in this course:")
print("   json     - read and write JSON        (Chapter 22)")
print("   datetime - dates and times            (Chapter 21)")
print("   pathlib  - file paths done properly   (Chapter 21)")
print("   re       - regular expressions        (Chapter 30)")
print("   sqlite3  - a full database, built in  (Chapter 39)")
print("   asyncio  - asynchronous programming   (Chapter 35)")
print("None of those need installing. That is a large part of why Python")
print("is so productive for scripting and automation.")
```

## PART 7: A short version history


```python
# A list of (version, year, headline) tuples we can loop over and print.
# Tuples are covered in Chapter 12.
milestones = [
    ("0.9.0", 1991, "first public release - classes, exceptions, functions"),
    ("1.0", 1994, "lambda, map, filter, reduce"),
    ("2.0", 2000, "list comprehensions, garbage collection"),
    ("3.0", 2008, "deliberate backwards-incompatible cleanup"),
    ("3.6", 2016, "f-strings"),
    ("3.10", 2021, "match statement, much better error messages"),
    ("3.11", 2022, "major speed improvements (10-60%)"),
    ("3.12", 2023, "improved f-strings, better typing"),
    ("3.13", 2024, "experimental free-threaded build (no GIL)"),
]

# Print each milestone on its own line, neatly aligned.
for release, year, headline in milestones:
    # ljust() pads the version out so the columns line up.
    print("   ", release.ljust(6), year, "-", headline)

print("Guido van Rossum started Python in December 1989 as a holiday")
print("project. It is named after Monty Python's Flying Circus, not the")
print("snake - which is why the docs are full of spam and eggs.")
print("Python 2 reached end of life on 1 January 2020. If a tutorial uses")
print("`print \"hello\"` without brackets, it is Python 2 - close the tab.")
```
