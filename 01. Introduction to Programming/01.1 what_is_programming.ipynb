# 01.1 What is Programming

A program is a precise sequence of instructions that turns input into output.

This file demonstrates the single most important idea in this chapter:
the computer does exactly what you SAY, never what you MEAN.


## PART 1: A program is just instructions, executed in order


```python
# Print a heading so the output sections are easy to tell apart.

# Each of these three lines is one instruction.
# Python runs the first, finishes it, then runs the second, and so on.
print("Step 1: wake up")
print("Step 2: make coffee")
print("Step 3: start working")

# Order is not decoration - it changes the result.
# The lines below are the same three instructions in a different order,
# and they describe a completely different (and worse) morning.
print("Same instructions, different order:")
print("Step 1: start working")
print("Step 2: wake up")
print("Step 3: make coffee")
```

## PART 2: Input -> Processing -> Output


```python
# INPUT: the data the program starts with.
# Normally this comes from a user or a file. Here we hardcode it so the
# example runs without anyone having to type anything.
price_before_tax = 100.0
tax_rate = 0.18

# PROCESSING: the calculation that turns input into the answer we want.
# This is the part that carries the actual logic of the program.
tax_amount = price_before_tax * tax_rate
price_after_tax = price_before_tax + tax_amount

# OUTPUT: showing the result to a human.
# A program that computes the right answer and never reports it is useless.
print("Input  - price before tax:", price_before_tax)
print("Input  - tax rate:", tax_rate)
print("Process- tax amount:", tax_amount)
print("Output - price after tax:", price_after_tax)
```

## PART 3: Precision - the computer does what you SAY, not what you MEAN


```python
# A human reading "add up these numbers and average them" knows what to do.
# A computer needs every step spelled out.
scores = [85, 92, 78, 95, 88]

# We must say explicitly: add them all together.
total_of_scores = sum(scores)

# We must say explicitly: count how many there are.
number_of_scores = len(scores)

# We must say explicitly: divide one by the other.
average_score = total_of_scores / number_of_scores

print("Scores:", scores)
print("Total:", total_of_scores)
print("Count:", number_of_scores)
print("Average:", average_score)

# Here is the classic "say vs mean" trap, demonstrated safely.
# A human would say the average of an empty list is "nothing" or zero.
# The computer, told to divide by the count, would divide by zero and crash.
empty_scores = []

# So we have to state the special case ourselves. The computer will not
# infer it. This single line is the difference between working software
# and a crash.
if len(empty_scores) == 0:
    print("Average of an empty list: cannot compute - no scores to average")
else:
    print("Average of an empty list:", sum(empty_scores) / len(empty_scores))
```

## PART 4: Program vs process


```python
# A PROGRAM is this file sitting on disk. It is static text. It does nothing.
# A PROCESS is what exists right now: this file loaded into memory and
# actively running, with real values stored in real memory locations.

# We import the sys module so we can look at the running process itself.
# An import makes another module's tools available in this file.
import sys

# sys.argv holds the command line arguments this process was started with.
# The first entry is always the path of the script being run - proof that
# a specific file was loaded and turned into a live process.
print("This process is running the program stored at:")
print("   ", sys.argv[0])

# A single program file can be run many times at once, as separate
# processes, each with its own independent memory. The recipe is one
# document; the cooking is a separate event each time.
print("Run this same file in two terminals and you get two processes,")
print("each with its own copy of every value - one program, two processes.")
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. A program is an ordered sequence of precise instructions.")
print("2. Order matters - the same instructions rearranged do something else.")
print("3. Every program is input -> processing -> output.")
print("4. The computer does what you say, not what you mean.")
print("5. Edge cases (like an empty list) must be handled explicitly.")
print("6. A program is the file on disk; a process is it running in memory.")
```
