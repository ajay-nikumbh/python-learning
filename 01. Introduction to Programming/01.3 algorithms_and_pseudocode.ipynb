# 01.3 Algorithms and Pseudocode

An algorithm is a finite, unambiguous sequence of steps that solves a problem.
Pseudocode is how you write one down before worrying about syntax.

This file shows the workflow that professional programmers actually use:
write the algorithm in English first, then translate it to code.


## PART 1: Pseudocode first, then the translation


```python
# THE PROBLEM: find the largest number in a list.
#
# THE PSEUDOCODE (written before any Python):
#
#   START
#     SET largest TO the first number in the list
#     FOR each number in the list
#       IF the number is greater than largest THEN
#         SET largest TO that number
#     RETURN largest
#   END
#
# Read that and you can tell it works. Now the translation, line for line.

# The input our algorithm will work on.
numbers = [23, 67, 12, 89, 45, 91, 34]

# PSEUDOCODE: "SET largest TO the first number in the list"
# We assume the first number is the winner until something beats it.
largest = numbers[0]

# PSEUDOCODE: "FOR each number in the list"
# Look at every number, one at a time.
for number in numbers:
    # PSEUDOCODE: "IF the number is greater than largest THEN"
    # Compare this number against the current record holder.
    if number > largest:
        # PSEUDOCODE: "SET largest TO that number"
        # This number wins, so it becomes the new record holder.
        largest = number

# PSEUDOCODE: "RETURN largest"
# The loop is finished, so largest now holds the biggest value.
print("Numbers:", numbers)
print("Largest:", largest)

# The structure of the pseudocode survived the translation completely.
# That is the point - the thinking was already done before typing.
```

## PART 2: The four properties every algorithm must have


```python
# An algorithm must be FINITE (it stops), UNAMBIGUOUS (one meaning per step),
# EFFECTIVE (each step is doable), and DEFINED (clear inputs and outputs).
def count_down_from(start_number):
    """Count down to zero. Demonstrates the FINITE property."""
    # Keep a list of the numbers we visit so we can show the result.
    visited_numbers = []

    # Start at the given number.
    current = start_number

    # Keep going while we are above zero. This condition is what
    # guarantees the algorithm STOPS - without it we loop forever.
    while current > 0:
        # Record where we are.
        visited_numbers.append(current)

        # Move one step closer to the stopping condition.
        # Forgetting this line is the single most common cause of a
        # program that hangs and never finishes.
        current = current - 1

    # Return the finished result. Clear output = the DEFINED property.
    return visited_numbers


# Run it and see that it terminates.
print("count_down_from(5) gives:", count_down_from(5))
print("It stops because each pass moves closer to the stop condition.")
```

## PART 3: The three building blocks - sequence, selection, iteration


```python
# BUILDING BLOCK 1 - SEQUENCE: do this, then this, then this.
print("SEQUENCE - steps in order:")

# Each line runs once, in the order written.
item_price = 250.0
quantity = 3
subtotal = item_price * quantity
print("   price:", item_price, "x quantity:", quantity, "= subtotal:", subtotal)

# BUILDING BLOCK 2 - SELECTION: choose between paths.
print("SELECTION - choosing a path:")

# The customer's status decides which calculation runs.
is_member = True

# Only one of these two branches will execute.
if is_member:
    # This path runs when is_member is True.
    final_total = subtotal * 0.90
    print("   member discount applied, total:", final_total)
else:
    # This path runs when is_member is False.
    final_total = subtotal
    print("   no discount, total:", final_total)

# BUILDING BLOCK 3 - ITERATION: repeat something.
print("ITERATION - repeating a step:")

# A cart with several items to add up.
cart_prices = [120.0, 45.5, 330.0]

# Start the running total at zero before the loop begins.
cart_total = 0

# Repeat the same "add it on" step once per item.
for price in cart_prices:
    # Each pass adds one price to the running total.
    cart_total = cart_total + price
    print("   added", price, "- running total is now", cart_total)

print("   final cart total:", cart_total)

print("Sequence, selection and iteration are the complete set.")
print("Functions and classes organise these three - they add no new powers.")
```

## PART 4: Decomposition - breaking a big problem into small ones


```python
# THE BIG PROBLEM: "analyse this sentence."
# That is far too vague to start typing. So we break it into pieces
# that are each obviously solvable, and solve them one at a time.

sentence = "the quick brown fox jumps over the lazy dog"


# SMALL PROBLEM 1: split the sentence into individual words.
def split_into_words(text):
    """Break a sentence into a list of words."""
    # .split() cuts the text wherever there is a space.
    return text.split()


# SMALL PROBLEM 2: count how many words there are.
def count_words(words):
    """Count the words in a list."""
    # len() reports how many items a list holds.
    return len(words)


# SMALL PROBLEM 3: find the longest word.
def find_longest_word(words):
    """Return the longest word in a list."""
    # Assume the first word is longest until one beats it.
    longest = words[0]

    # Check each word's length against the current record.
    for word in words:
        # len(word) is the number of characters in that word.
        if len(word) > len(longest):
            longest = word

    return longest


# SMALL PROBLEM 4: calculate the average word length.
def average_word_length(words):
    """Return the mean number of characters per word."""
    # Add up the lengths of all the words.
    total_characters = 0
    for word in words:
        total_characters = total_characters + len(word)

    # Divide by how many words there were.
    return total_characters / len(words)


# Now the "big" problem is just four easy calls in sequence.
words = split_into_words(sentence)

print("Sentence:", sentence)
print("Word count:", count_words(words))
print("Longest word:", find_longest_word(words))
print("Average word length:", round(average_word_length(words), 2))

print("Nothing here was hard. Decomposition is what made it easy.")
```

## PART 5: Different algorithms, same answer, different cost


```python
# THE PROBLEM: add up the numbers from 1 to n.

# ALGORITHM A - loop through every number and add it on.
# This does n additions, so the work grows with n.
def sum_by_looping(n):
    """Add 1 + 2 + ... + n by visiting every number."""
    total = 0

    # Visit each number from 1 up to and including n.
    for number in range(1, n + 1):
        total = total + number

    return total


# ALGORITHM B - use the arithmetic series formula.
# This does the same job in a fixed number of operations, whatever n is.
def sum_by_formula(n):
    """Add 1 + 2 + ... + n using n * (n + 1) / 2."""
    # Gauss's formula. Three operations, regardless of how big n gets.
    return n * (n + 1) // 2


# Both give the identical answer.
target = 1000
print("Sum of 1 to", target)
print("   by looping:", sum_by_looping(target))
print("   by formula:", sum_by_formula(target))
print("   same answer?", sum_by_looping(target) == sum_by_formula(target))

# But their cost is very different as the input grows.
import time

# Time the looping version on a large input.
start = time.perf_counter()
sum_by_looping(1000000)
loop_time = time.perf_counter() - start

# Time the formula version on the same input.
start = time.perf_counter()
sum_by_formula(1000000)
formula_time = time.perf_counter() - start

print("For n = 1,000,000:")
print("   looping took: ", round(loop_time * 1000, 4), "ms")
print("   formula took: ", round(formula_time * 1000, 4), "ms")
print("Choosing the right algorithm beats any amount of micro-optimisation.")
print("This idea is measured properly in Chapter 37 (Big-O).")
```
