# 03.6 PEP 8 Introduction

**PEP 8** is Python's style guide. It has no force at runtime — code that ignores
it runs perfectly — but effectively every Python project follows it, which makes
it the closest thing the language has to a shared dialect.

## Theory

### What a PEP is

A **PEP** is a Python Enhancement Proposal: a numbered design document. Some
propose language features, some describe process, and a few define conventions.

The ones worth knowing by number:

- **PEP 8** — style guide for code
- **PEP 20** — the Zen of Python (`import this`)
- **PEP 257** — docstring conventions
- **PEP 484** — type hints

PEP 8 was written by Guido van Rossum, Barry Warsaw and Nick Coghlan in 2001.

### Why it matters more than it looks

Code is read far more often than it is written. When every project shares a
layout, you can open an unfamiliar codebase and immediately know what is a class,
what is a constant, and where a block begins.

PEP 8 also removes an entire category of pointless argument. The style question
is already answered, so review can focus on whether the code is *correct*.

### The rule PEP 8 states about itself

The guide explicitly says consistency with the surrounding code wins:

> A style guide is about consistency. Consistency with this style guide is
> important. Consistency within a project is more important. Consistency within
> one module or function is the most important.

So: when editing existing code, match what is already there.

### Do not apply it by hand

Formatters apply PEP 8 mechanically and linters flag what remains. Learn the
rules so you can *read* code fluently — then let `ruff` or `black` do the work.

```python
# The layout rules, and what each one prevents.
layout_rules = [
    ("Indentation", "4 spaces per level", "never tabs - see 03.2"),
    ("Line length", "79 chars (88 with black)", "side-by-side diffs stay readable"),
    ("Blank lines", "2 between top-level defs", "visual separation of units"),
    ("Blank lines", "1 between methods", "same, at a smaller scale"),
    ("Imports", "one per line, at the top", "easy to scan and edit"),
    ("Import order", "stdlib, third-party, local", "origin is obvious at a glance"),
    ("Encoding", "UTF-8, no declaration", "the default since Python 3"),
    ("File ending", "one newline at the end", "POSIX tools expect it"),
]

print("Area          Rule                          Why")
print("-" * 82)
for area, rule, why in layout_rules:
    print(area.ljust(13), rule.ljust(29), why)
```

## Whitespace: where it goes and where it does not

Most PEP 8 violations are whitespace. The underlying principle is that spaces
should show *structure*, so adding them where there is no structural boundary
makes code harder to read, not easier.

```python
# Each entry is (wrong, right, the rule).
whitespace_rules = [
    ("spam( ham[ 1 ] )", "spam(ham[1])", "no space just inside brackets"),
    ("x , y = y , x", "x, y = y, x", "no space before a comma"),
    ("print ('hello')", "print('hello')", "no space before a call's bracket"),
    ("dict ['key']", "dict['key']", "no space before an index bracket"),
    ("x = 1; y = 2", "x = 1\ny = 2", "one statement per line"),
    ("if x == 4 :", "if x == 4:", "no space before a colon"),
    ("x=1", "x = 1", "spaces around assignment"),
    ("x<1", "x < 1", "spaces around comparison"),
    ("def f(a = 1)", "def f(a=1)", "NO spaces around a default value"),
    ("f(a = 1)", "f(a=1)", "NO spaces around a keyword argument"),
]

print("Wrong                   Right                  Rule")
print("-" * 78)
for wrong, right, rule in whitespace_rules:
    # Show the newline escape literally rather than breaking the table.
    display_right = right.replace(chr(10), " / ")
    print(wrong.ljust(24), display_right.ljust(23), rule)

print("")
print("The last two are the ones people get wrong most:")
print("   spaces around = for ASSIGNMENT, none for DEFAULTS and KEYWORDS.")
```

```python
# The default-value rule, demonstrated with real code.

# CORRECT: no spaces around = in the signature or the call.
def create_user(name, role="member", active=True):
    """Build a user record."""
    return {"name": name, "role": role, "active": active}


# Calling it - again, no spaces around = for keyword arguments.
user = create_user("Ajay", role="admin")
print("Created:", user)

# The exception: when a parameter has a type ANNOTATION, spaces return.
def create_user_typed(name: str, role: str = "member") -> dict:
    """Build a user record, with type hints."""
    return {"name": name, "role": role}


print("Typed version:", create_user_typed("Ajay"))

print("")
print("Rule summary:")
print("   role='member'          no annotation -> no spaces")
print("   role: str = 'member'   annotation    -> spaces around =")
```

### Operators: use spacing to show precedence

PEP 8 allows you to *omit* spaces around the highest-priority operator in an
expression, to make the grouping visible. This is a rare case where the guide
prefers judgement over a fixed rule.

```python
x = 3
y = 4
i = 2
n = 10

# PEP 8 encourages this - the tighter binding is visually tighter too.
good_a = x*2 - 1
good_b = x*x + y*y
good_c = (x+y) * (x-y)

# Technically fine, but the uniform spacing hides the precedence.
poor_a = x * 2 - 1
poor_b = x * x + y * y

print("x*2 - 1      =", good_a)
print("x*x + y*y    =", good_b)
print("(x+y)*(x-y)  =", good_c)

print("")
print("Both spacings are valid. The first shows which operation binds")
print("tighter, so the reader does not have to recall the precedence table.")

# Never mix in a way that MISLEADS about precedence.
misleading = 1+2 * 3        # looks like (1+2)*3, is actually 1+(2*3)
print("")
print("1+2 * 3 evaluates to", misleading, "- the spacing lies about the grouping.")
print("Write 1 + 2*3 instead.")
```

## Imports

Import style is the most mechanical part of PEP 8, and the easiest to get right
with tooling.

```python
# CORRECT ORDER, with a blank line between each group:
#
#   import os                    <- 1. standard library
#   import sys
#
#   import requests              <- 2. third party
#   from django.db import models
#
#   from myapp.utils import helper   <- 3. your own code

import_rules = [
    ("One import per line", "import os\nimport sys", "not: import os, sys"),
    ("Absolute over relative", "from pkg.mod import f", "clearer about origin"),
    ("No wildcard imports", "from x import name", "not: from x import *"),
    ("Group with blank lines", "stdlib / third-party / local", "three groups"),
    ("Alphabetical within group", "import os then import sys", "easy to scan"),
    ("At the top of the file", "after the docstring", "never inside a function*"),
]

print("Rule                        Example                      Note")
print("-" * 84)
for rule, example, note in import_rules:
    display = example.replace(chr(10), " / ")
    print(rule.ljust(27), display.ljust(28), note)

print("")
print("* The exception: a deliberately deferred import to avoid a circular")
print("  dependency or a slow startup. Rare, and worth a comment when used.")

print("")
print("Why no wildcard imports: you cannot tell where a name came from,")
print("and they silently overwrite existing names.")
```

## A before-and-after example

Everything above, applied to one piece of code.

```python
# BEFORE - valid Python, but hard to read.
#
#   import os,sys
#   MAX=100
#   def calc( x,y ):
#     if x>y :
#       return x*2+y
#     else :
#       return y*2+x
#   class myClass :
#     def __init__(self,name) :
#       self.Name=name

# AFTER - the same logic, following PEP 8.

import os
import sys

MAX_ITEMS = 100


def calculate_weighted(first, second):
    """Return the larger value doubled, plus the smaller one."""
    # Guard clause keeps the main path unindented.
    if first > second:
        return first*2 + second

    return second*2 + first


class InventoryItem:
    """A single item held in stock."""

    def __init__(self, name):
        # Attribute names use snake_case, like variables.
        self.name = name


# Check the rewritten code behaves.
print("calculate_weighted(5, 3) =", calculate_weighted(5, 3))
print("calculate_weighted(3, 5) =", calculate_weighted(3, 5))
print("InventoryItem('bolt').name =", InventoryItem("bolt").name)

print("")
print("Changes made:")
changes = [
    "split the combined import onto separate lines",
    "renamed MAX to MAX_ITEMS - constants say what they limit",
    "renamed calc to calculate_weighted - a verb phrase",
    "renamed myClass to InventoryItem - PascalCase, and meaningful",
    "renamed self.Name to self.name - attributes are snake_case",
    "4-space indentation instead of 2",
    "removed spaces inside brackets and before colons",
    "two blank lines between top-level definitions",
    "added docstrings",
]
for index, change in enumerate(changes, start=1):
    print("  ", index, "-", change)
```

## Where PEP 8 says to ignore PEP 8

The guide has a section titled *A Foolish Consistency is the Hobgoblin of Little
Minds*. It lists cases where following the rule makes things worse.

```python
exceptions = [
    ("Breaking it makes code less readable", "the rule serves readability, not the reverse"),
    ("Surrounding code already breaks it", "match the file you are editing"),
    ("The code predates the rule", "do not reformat unrelated code in a bug fix"),
    ("Compatibility with an older Python", "practicality beats purity"),
]

print("Ignore PEP 8 when...")
print("-" * 76)
for when, why in exceptions:
    print("  " + when.ljust(42), why)

print("")
print("The guide's own priority order:")
priorities = [
    "Consistency within one function - most important",
    "Consistency within one module",
    "Consistency with the project",
    "Consistency with PEP 8 - least important of the four",
]
for index, priority in enumerate(priorities, start=1):
    print("  ", index, "-", priority)

print("")
print("This surprises people. The guide is a default, not an authority.")
```

## Let the tools do it

Nobody applies PEP 8 by hand. Modern projects use `ruff`, which replaces several
older tools and runs fast enough to be invisible.

```python
tools = [
    ("ruff check .", "lint - report problems", "ruff"),
    ("ruff check --fix .", "lint and auto-fix what it safely can", "ruff"),
    ("ruff format .", "reformat the code", "ruff"),
    ("black .", "reformat, the original tool", "black"),
    ("mypy .", "check type hints", "mypy"),
    ("pre-commit run", "run all of the above before each commit", "pre-commit"),
]

print("Command                 Does                                  Tool")
print("-" * 80)
for command, does, tool in tools:
    print(command.ljust(23), does.ljust(38), tool)

print("")
print("A typical pyproject.toml section:")
print("")
config_lines = [
    "[tool.ruff]",
    "line-length = 88",
    "",
    "[tool.ruff.lint]",
    'select = ["E", "F", "I", "N", "UP"]',
    "#   E = pycodestyle errors      F = pyflakes",
    "#   I = import sorting          N = naming conventions",
    "#   UP = modernise old syntax",
]
for line in config_lines:
    print("   " + line)

print("")
print("Chapter 41 covers tooling in full.")
```

## Error codes worth recognising

Linters report short codes rather than sentences. These are the ones you will see
most.

```python
codes = [
    ("E501", "line too long", "wrap it, or raise line-length"),
    ("E225", "missing whitespace around operator", "x=1 should be x = 1"),
    ("E302", "expected 2 blank lines", "between top-level definitions"),
    ("E303", "too many blank lines", "more than two in a row"),
    ("E711", "comparison to None", "use `is None`, not `== None`"),
    ("E712", "comparison to True", "use `if x:`, not `if x == True:`"),
    ("F401", "imported but unused", "delete the import"),
    ("F841", "local variable assigned but never used", "delete it, or use _"),
    ("N802", "function name should be lowercase", "rename to snake_case"),
    ("N806", "variable in function should be lowercase", "rename to snake_case"),
]

print("Code   Means                                      Fix")
print("-" * 84)
for code, means, fix in codes:
    print(code.ljust(6), means.ljust(42), fix)

print("")
print("E711 and E712 catch a real bug class, not just style:")

value = None
print("   value == None  ->", value == None, "  (works, but fragile)")
print("   value is None  ->", value is None, "  (correct - identity, not equality)")
print("")
print("Chapter 06 explains why `is` and `==` differ.")
```

## Takeaways

1. **PEP 8** is Python's style guide — not enforced at runtime, but universally
   followed.
2. Four-space indentation, 79 characters (or 88 with `black`), two blank lines
   between top-level definitions.
3. Spaces around assignment and comparison; **no** spaces around `=` for default
   values and keyword arguments — unless there is an annotation.
4. Imports: one per line, at the top, grouped stdlib / third-party / local.
5. Never use wildcard imports — you lose track of where names came from.
6. PEP 8 explicitly says **consistency within a file beats consistency with the
   guide**.
7. Do not apply it manually. `ruff` lints and formats in one tool.
8. Learn the common codes — `E501`, `E711`, `F401` — so linter output reads
   instantly.
