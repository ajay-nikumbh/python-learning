# 02.6 pip Basics

pip installs packages written by other people. This is the single
biggest reason Python is so productive: whatever you need, someone has
probably already built it.

This file shows what is installed in your environment right now, and
the commands for managing packages.


## EXAMPLE 1: What is installed right now?


```python
print("EXAMPLE 1: Listing installed packages")

# importlib.metadata reads information about installed packages.
# This is the same data `pip list` displays.
from importlib import metadata

# distributions() gives one entry per installed package.
installed = []
for distribution in metadata.distributions():
    # Some odd installs have missing metadata, so read defensively.
    package_name = distribution.metadata["Name"]
    package_version = distribution.version
    if package_name:
        installed.append((package_name, package_version))

# Sort alphabetically, ignoring case, so the list is easy to scan.
installed.sort(key=lambda entry: entry[0].lower())

print("This environment has", len(installed), "packages installed.")

# Show the first ten so the output stays readable.
print("The first ten:")
for package_name, package_version in installed[:10]:
    print("   ", package_name.ljust(28), package_version)
```

## EXAMPLE 2: Is a specific package available?


```python
print("EXAMPLE 2: Checking for a package")

# Real code often needs to know whether an optional package is present.
packages_to_check = ["pip", "jupyterlab", "requests", "numpy", "definitely_not_real"]

for package_name in packages_to_check:
    try:
        # version() raises if the package is not installed.
        found_version = metadata.version(package_name)
        print("   ", package_name.ljust(22), "installed, version", found_version)
    except metadata.PackageNotFoundError:
        print("   ", package_name.ljust(22), "not installed")
```

## EXAMPLE 3: Standard library vs third party


```python
print("EXAMPLE 3: What needs installing, and what does not")

import sys

# The standard library ships with Python - no pip needed.
standard_library_examples = ["json", "csv", "datetime", "pathlib", "re", "sqlite3"]

# These must be installed before you can import them.
third_party_examples = ["requests", "numpy", "pandas", "django", "pytest"]

print("Ships with Python (no install needed):")
for module_name in standard_library_examples:
    # sys.stdlib_module_names lists every standard library module.
    in_stdlib = module_name in sys.stdlib_module_names
    marker = "yes" if in_stdlib else "no"
    print("   ", module_name.ljust(12), "standard library:", marker)

print("Must be installed with pip:")
for module_name in third_party_examples:
    print("   ", module_name)

print("Before reaching for pip, check whether the standard library")
print("already solves your problem. Chapter 21 tours what is included.")
```

## EXAMPLE 4: The commands you will actually use


```python
print("EXAMPLE 4: Everyday pip commands")

# Each entry is (command, what it does).
commands = [
    ("pip install requests", "install the latest version"),
    ("pip install requests==2.31.0", "install one exact version"),
    ("pip install 'requests>=2.28'", "install at least that version"),
    ("pip install -r requirements.txt", "install everything in a file"),
    ("pip install --upgrade requests", "upgrade to the latest"),
    ("pip uninstall requests", "remove it"),
    ("pip list", "show everything installed"),
    ("pip list --outdated", "show what has newer versions"),
    ("pip show requests", "details, including dependencies"),
    ("pip freeze > requirements.txt", "record exact versions to a file"),
]

print("Command                              Does")
print("-" * 72)

for command, description in commands:
    print(command.ljust(37), description)
```

## EXAMPLE 5: Version specifiers explained


```python
print("EXAMPLE 5: Pinning versions")

specifiers = [
    ("requests", "any version - whatever is newest today"),
    ("requests==2.31.0", "exactly this version, nothing else"),
    ("requests>=2.28", "this version or newer"),
    ("requests<3.0", "anything before the next major version"),
    ("requests>=2.28,<3.0", "a range - the usual choice for libraries"),
    ("requests~=2.31.0", "2.31.x but not 2.32 - compatible releases"),
]

print("Specifier                Meaning")
print("-" * 62)

for specifier, meaning in specifiers:
    print(specifier.ljust(25), meaning)

print("Applications usually pin exactly (==) for reproducible installs.")
print("Libraries usually allow a range, so they can coexist with others.")
print("Semantic versioning is covered in Chapter 40.")
```

## EXAMPLE 6: The command that avoids the wrong-pip problem


```python
print("EXAMPLE 6: Making sure pip matches your Python")

print("With several Pythons installed, plain `pip` can install into the")
print("wrong one. You then get ModuleNotFoundError for a package you")
print("just watched install successfully.")
print("The fix is to run pip THROUGH the interpreter you mean:")
print("   python3 -m pip install requests")
print("This guarantees pip installs into the Python that runs your code.")
print("Your current interpreter is:")
print("   ", sys.executable)
print("So the safest install command for you is:")
print(f"    {sys.executable} -m pip install <package>")
```

## EXAMPLE 7: Reading a dependency tree


```python
print("EXAMPLE 7: Packages depend on other packages")

# Installing one package often pulls in several others.
# Let's look at a real one from this environment if we can.
print("Installing a package usually installs its dependencies too.")
print("For example, `pip install requests` also installs:")
for dependency in ["urllib3", "certifi", "charset-normalizer", "idna"]:
    print("   ", dependency)

print("Check what something requires with:")
print("   pip show requests")

# Show real dependency data for a package we know exists.
try:
    pip_requires = metadata.requires("pip")
    if pip_requires:
        print("pip itself declares these requirements:")
        for requirement in pip_requires[:5]:
            print("   ", requirement)
    else:
        print("pip itself declares no external requirements.")
except metadata.PackageNotFoundError:
    print("pip metadata not available in this environment.")

print("This is why requirements.txt matters: it records the whole tree,")
print("not just the packages you asked for directly.")
```

## EXAMPLE 8: Safety notes


```python
print("EXAMPLE 8: Installing safely")

safety_rules = [
    ("Check the name carefully", "typo-squatting packages are a real attack"),
    ("Activate your venv first", "otherwise you install system-wide"),
    ("Prefer python3 -m pip", "guarantees the right interpreter"),
    ("Avoid sudo pip install", "it can break your operating system's Python"),
    ("Pin versions in production", "an unexpected upgrade can break a deploy"),
    ("Review new dependencies", "each one is code you are choosing to trust"),
]

print("Rule                          Why")
print("-" * 72)

for rule, reason in safety_rules:
    print(rule.ljust(30), reason)

print("The typo-squatting point is not theoretical. Packages with names")
print("one character away from popular ones have shipped malware.")
print("Security is covered properly in Chapter 42.")
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. pip installs third-party packages from PyPI.")
print("2. The standard library needs no installation - check there first.")
print("3. Use `python3 -m pip` so pip matches your interpreter.")
print("4. == pins an exact version; >= allows newer ones.")
print("5. pip freeze records what is installed; -r installs from that file.")
print("6. Activate your virtual environment before installing anything.")
print("7. Never use sudo pip - it can damage the system Python.")
```
