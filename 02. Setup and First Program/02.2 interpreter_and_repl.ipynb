# 02.2 Python Interpreter and REPL

The REPL - Read, Evaluate, Print, Loop - is an interactive Python
session. You type an expression, press Enter, and see the result
immediately.

It is the single best tool for answering "what does this do?" in five
seconds instead of writing a whole file.

This file demonstrates what the REPL does, and the built-in functions
you use to explore inside it.


## EXAMPLE 1: What the REPL actually does


```python
print("EXAMPLE 1: The Read-Evaluate-Print Loop")

# In a REPL, typing an expression displays its value automatically.
# In a FILE, you must print() it - this is the difference that confuses
# beginners most often.

# Here is the loop, made explicit. Each string is something you might type.
typed_lines = [
    "2 + 3",
    "len('hello')",
    "'py' * 3",
    "10 / 4",
    "sorted([3, 1, 2])",
]

print("If you typed these into the REPL, you would see:")

for line in typed_lines:
    # READ: take the text the user typed.
    # EVALUATE: eval() computes the value of an expression string.
    # We use it here only to simulate the REPL - you rarely need eval().
    result = eval(line)

    # PRINT: show the result, the way the REPL would.
    print("   >>>", line)
    print("   " + repr(result))
    print("")

# LOOP: the REPL goes back and waits for the next line.
print("Then it loops and waits for your next line.")
```

## EXAMPLE 2: Expression vs statement - why some lines show nothing


```python
print("EXAMPLE 2: Why some REPL lines print nothing")

# An EXPRESSION produces a value, so the REPL displays it.
# A STATEMENT performs an action, so there is nothing to display.

print("These are EXPRESSIONS - the REPL shows a result:")
print("   >>> 2 + 3")
print("   5")
print("   >>> 'hello'.upper()")
print("   'HELLO'")

print("These are STATEMENTS - the REPL shows nothing:")
print("   >>> x = 5          (assignment does not produce a value)")
print("   >>> import math    (importing does not produce a value)")

# We can demonstrate the difference concretely.
value_of_expression = 2 + 3

# An assignment evaluates to nothing at all. Functions with no explicit
# return give back None, which the REPL deliberately hides.
def a_function_with_no_return():
    """Do something, but return nothing."""
    # No return statement, so Python returns None automatically.
    pass

print("2 + 3 evaluates to:", repr(value_of_expression))
print("A function with no return gives:", repr(a_function_with_no_return()))
print("The REPL hides None, which is why those lines look empty.")
```

## EXAMPLE 3: The underscore - the last result


```python
print("EXAMPLE 3: The _ shortcut")

print("Inside the REPL, _ holds the value of the last expression:")
print("   >>> 10 * 5")
print("   50")
print("   >>> _ + 2        # _ is 50")
print("   52")
print("   >>> _ * 2        # _ is now 52")
print("   104")
print("This only works in the REPL - _ is not special inside a file.")
print("It is handy for chaining quick calculations without variables.")
```

## EXAMPLE 4: help() - reading documentation without leaving Python


```python
print("EXAMPLE 4: help()")

# help() prints the documentation for whatever you pass it.
# Normally it opens a pager; here we read a docstring directly so the
# output stays short.

# Every built-in function carries its documentation as a __doc__ string.
print("The first line of help(len) says:")

# splitlines() breaks the text into a list of lines; [0] takes the first.
print("   ", len.__doc__.splitlines()[0])

print("The first line of help(sorted) says:")
print("   ", sorted.__doc__.splitlines()[0])

print("In a real REPL you would type:")
print("   >>> help(len)")
print("   >>> help(str.upper)")
print("   >>> help(list)")
print("Press q to leave the help pager and return to the prompt.")
```

## EXAMPLE 5: dir() - what can this thing do?


```python
print("EXAMPLE 5: dir()")

# dir() lists the names available on an object - its methods and
# attributes. This is how you discover what is possible.

# Take a string and ask what it can do.
example_text = "hello"

# Most objects have many dunder names (__like_this__) that are internal.
# Filter them out to see the methods you would actually call.
all_names = dir(example_text)

# A list comprehension keeps only names that do not start with underscore.
# Comprehensions are covered fully in Chapter 15.
public_methods = [name for name in all_names if not name.startswith("_")]

print("A string has", len(all_names), "total names,")
print("of which", len(public_methods), "are public methods.")

print("The first fifteen:")

# Show them in rows of five so the output stays readable.
for start_index in range(0, 15, 5):
    # Slice out five names at a time.
    row = public_methods[start_index:start_index + 5]
    # join() glues the list into one string separated by commas.
    print("   ", ", ".join(row))

print("So you can discover .upper() without looking anything up:")
print("   'hello'.upper() gives", repr(example_text.upper()))
```

## EXAMPLE 6: type() - what am I holding?


```python
print("EXAMPLE 6: type()")

# type() answers "what kind of value is this?" - the fastest way to
# diagnose a confusing bug.

# A list of sample values to inspect.
sample_values = [42, 3.14, "text", True, None, [1, 2], (1, 2), {"a": 1}, {1, 2}]

print("Value                 Type")
print("-" * 40)

for value in sample_values:
    # repr() shows the value as you would type it in code.
    value_text = repr(value)

    # __name__ gives the short type name instead of "<class 'int'>".
    type_name = type(value).__name__

    # ljust() pads the text so the columns line up.
    print(value_text.ljust(22), type_name)

print("Notice True is a 'bool' and None is a 'NoneType'.")
print("Both are covered properly in Chapter 05.")
```

## EXAMPLE 7: Useful REPL commands


```python
print("EXAMPLE 7: Getting in and out")

print("Starting the REPL:")
print("   python3            start an interactive session")
print("   python3 -q         start it without the version banner")
print("Running code without a file:")
print("   python3 -c \"print(2 + 3)\"       run one line and exit")
print("   python3 -m this                 run a module")
print("Leaving the REPL:")
print("   exit()             or  quit()")
print("   Ctrl-D             macOS and Linux")
print("   Ctrl-Z then Enter  Windows")
print("Interrupting a stuck program:")
print("   Ctrl-C             stops whatever is running")
```
