# 02.1 Installing Python

You cannot run Python until it is installed, and "is it installed?" is
a surprisingly common source of confusion - most machines have several
Pythons on them at once.

This file inspects the Python that is running it, so you can tell
exactly which installation you are using.


## EXAMPLE 1: Which Python is running this file?


```python
print("EXAMPLE 1: Identifying this interpreter")

# sys holds information about the interpreter currently running.
import sys

# sys.executable is the full path to the python program running this code.
# This is the single most useful line when something confusing happens.
print("Executable path:")
print("   ", sys.executable)

# version_info is a structured tuple - use this when comparing in code.
print("Version numbers:")
print("    major:", sys.version_info.major)
print("    minor:", sys.version_info.minor)
print("    micro:", sys.version_info.micro)

# The full version string, including the compiler that built it.
print("Full version string:")
print("   ", sys.version)
```

## EXAMPLE 2: Checking you meet a minimum version


```python
print("EXAMPLE 2: Requiring a minimum version")

# Real projects check the version before using newer features.
# Comparing tuples works element by element, so this reads naturally.
required_version = (3, 12)

# sys.version_info compares directly against a plain tuple.
if sys.version_info >= required_version:
    print("This Python meets the 3.12+ requirement.")
else:
    print("This Python is older than 3.12 - some chapters will not run.")

# Show the comparison explicitly so the logic is visible.
print("Required:", required_version)
print("Actual:  ", (sys.version_info.major, sys.version_info.minor))
print("Good enough?", sys.version_info >= required_version)
```

## EXAMPLE 3: Which operating system are you on?


```python
print("EXAMPLE 3: Your platform")

# platform reports facts about the machine and OS.
import platform

# system() gives the short OS name: Windows, Darwin (macOS), or Linux.
operating_system = platform.system()
print("Operating system:", operating_system)

# Note that macOS reports as "Darwin" - its underlying Unix name.
if operating_system == "Darwin":
    print("   (Darwin is the internal name for macOS)")

# A longer description including the version.
print("Platform detail:", platform.platform())

# The processor architecture - arm64 on Apple Silicon, x86_64 on Intel.
print("Machine type:", platform.machine())
```

## EXAMPLE 4: Install commands for each platform


```python
print("EXAMPLE 4: How Python is installed on each platform")

# A dictionary mapping each OS to its recommended install method.
# Dictionaries are covered fully in Chapter 14.
install_methods = {
    "Windows": [
        "Download the installer from python.org",
        "TICK 'Add python.exe to PATH' on the first screen",
        "Verify with:  python --version",
    ],
    "Darwin": [
        "Download the installer from python.org, or use Homebrew",
        "Homebrew:  brew install python@3.12",
        "Verify with:  python3 --version",
    ],
    "Linux": [
        "Most distributions ship Python already",
        "Debian/Ubuntu:  sudo apt install python3 python3-venv python3-pip",
        "Verify with:  python3 --version",
    ],
}

# Show the steps for the platform we are actually on.
print("Steps for your platform (" + operating_system + "):")

# .get() returns a fallback when the key is missing, instead of crashing.
steps = install_methods.get(operating_system, ["Unknown platform"])

# enumerate() gives us a counter alongside each item, starting at 1.
for step_number, step in enumerate(steps, start=1):
    print("   ", step_number, "-", step)
```

## EXAMPLE 5: python vs python3 - the classic confusion


```python
print("EXAMPLE 5: python vs python3")

print("On macOS and Linux, the command is usually 'python3'.")
print("On Windows, it is usually 'python'.")
print("The reason is historical: 'python' used to mean Python 2, so")
print("Unix systems kept 'python3' explicit to avoid breaking scripts.")
print("If a command is not found, try the other one. To be certain")
print("which interpreter a command maps to, run:")
print("    python3 -c \"import sys; print(sys.executable)\"")
print("That prints exactly what this file printed in Example 1.")
```

## EXAMPLE 6: Where Python looks for installed packages


```python
print("EXAMPLE 6: Where this Python finds its files")

# sys.prefix is the root folder of this Python installation.
print("Installation root (sys.prefix):")
print("   ", sys.prefix)

# sys.path is the list of folders searched when you import something.
# Covered fully in Chapter 20.
print("First few folders searched on import:")

# Show only the first three, to keep the output readable.
for folder in sys.path[:3]:
    # An empty string means "the current directory".
    if folder == "":
        print("     (the current directory)")
    else:
        print("    ", folder)
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. sys.executable tells you exactly which Python is running.")
print("2. sys.version_info compares cleanly against a tuple like (3, 12).")
print("3. macOS reports its platform name as 'Darwin'.")
print("4. Use 'python3' on macOS/Linux, 'python' on Windows.")
print("5. On Windows, ticking 'Add to PATH' avoids most setup problems.")
print("6. Several Pythons can coexist - always check which one you have.")
```
