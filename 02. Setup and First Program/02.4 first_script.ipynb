# 02.4 Writing and Running Your First Script

A script is a file of Python that runs top to bottom. This is where the
feedback loop starts: write, run, read the output, change it, run again.

This file builds up from the smallest possible program to one that takes
input, makes decisions, and reports results.


## EXAMPLE 1: The smallest possible program


```python
print("EXAMPLE 1: Hello, world")

# The traditional first program in any language. One line, one function.
print("Hello, world!")

# print() sends text to the terminal. That is its whole job.
# The brackets are how you call a function, and the text inside is the
# argument you are passing to it.

print("That one line is a complete, valid Python program.")
```

## EXAMPLE 2: Storing values in variables


```python
print("EXAMPLE 2: Variables")

# A variable gives a name to a value so you can reuse it.
name = "Ajay"
year_of_birth = 1995
favourite_number = 7.5
is_learning_python = True

# print() accepts several arguments and separates them with a space.
print("Name:", name)
print("Born:", year_of_birth)
print("Favourite number:", favourite_number)
print("Learning Python:", is_learning_python)

# You can use variables in calculations.
current_year = 2026
age = current_year - year_of_birth

print("Age this year:", age)
```

## EXAMPLE 3: f-strings - putting values inside text


```python
print("EXAMPLE 3: Building sentences with f-strings")

# Passing many arguments to print() gets clumsy. An f-string embeds
# values directly in the text. The f before the quote is what enables it.
print(f"{name} is {age} years old.")

# Any expression works inside the braces, not just a plain name.
print(f"Next year {name} will be {age + 1}.")

# You can format numbers as you insert them.
price = 1234.5678

# :.2f means "show this as a number with two decimal places".
print(f"Price rounded: {price:.2f}")

# :, adds thousands separators.
print(f"Price with separators: {price:,.2f}")

print("f-strings are covered fully in Chapter 07.")
```

## EXAMPLE 4: Doing arithmetic


```python
print("EXAMPLE 4: Arithmetic")

first = 17
second = 5

# Each operator gets its own line so you can see the difference clearly.
print(f"{first} + {second}  =", first + second)
print(f"{first} - {second}  =", first - second)
print(f"{first} * {second}  =", first * second)
print(f"{first} / {second}  =", first / second, " <- true division, a float")
print(f"{first} // {second} =", first // second, " <- floor division, rounds down")
print(f"{first} % {second}  =", first % second, " <- remainder")
print(f"{first} ** 2 =", first ** 2, "<- power")

print("Note that / always gives a float, even when it divides evenly:")
print("   10 / 5 =", 10 / 5)
print("   10 // 5 =", 10 // 5)
```

## EXAMPLE 5: Getting input from the user


```python
print("EXAMPLE 5: Reading input")

# input() pauses the program and waits for the user to type something.
# It ALWAYS returns text, even when the user types a number.
#
# We do not call it here, because this file needs to run unattended.
# Instead we simulate what it returns, so the behaviour is still visible.

# This is what input("Your age: ") would give if the user typed 30.
typed_by_user = "30"

print("If the user typed 30, input() returns:", repr(typed_by_user))
print("Its type is:", type(typed_by_user).__name__)

# Because it is text, arithmetic fails until you convert it.
try:
    # This looks reasonable but does not do what you want.
    wrong = typed_by_user + 1
except TypeError as error:
    print("")
    print("Trying to add 1 to it raises TypeError:")
    print("   ", error)

# int() converts text into a whole number.
age_as_number = int(typed_by_user)

print("After int(), the value is:", age_as_number)
print("Its type is now:", type(age_as_number).__name__)
print("And arithmetic works:", age_as_number + 1)

print("The real pattern you would write:")
print("   age = int(input('Your age: '))")
```

## EXAMPLE 6: Making a decision


```python
print("EXAMPLE 6: Conditionals")

# A program becomes useful when it can react to different values.
# We test several ages so you can see every branch run.

test_ages = [8, 16, 25, 70]

for test_age in test_ages:
    # Check each condition in order; the first true one wins.
    if test_age < 13:
        category = "child"
    elif test_age < 20:
        category = "teenager"
    elif test_age < 65:
        category = "adult"
    else:
        category = "senior"

    print(f"   age {str(test_age).rjust(2)} -> {category}")

print("Conditionals are covered fully in Chapter 09.")
```

## EXAMPLE 7: Repeating work with a loop


```python
print("EXAMPLE 7: Loops")

# A loop repeats a block of code once per item.
print("Counting to five:")

# range(1, 6) produces 1, 2, 3, 4, 5 - the end value is excluded.
for number in range(1, 6):
    print("   ", number)

print("The five times table:")

for multiplier in range(1, 6):
    # Work out the answer, then display the whole sum.
    answer = 5 * multiplier
    print(f"    5 x {multiplier} = {answer}")

print("Loops are covered fully in Chapter 10.")
```

## EXAMPLE 8: A complete small program


```python
print("EXAMPLE 8: Putting it together")

# Everything above, combined into something that does a real job:
# summarise a shopping basket.

# The basket: each entry is an item name, its price, and the quantity.
basket = [
    ("Notebook", 45.00, 3),
    ("Pen", 12.50, 10),
    ("Backpack", 1299.00, 1),
    ("Water bottle", 350.00, 2),
]

# Start the total at zero before we begin adding to it.
running_total = 0.0

print("Item            Qty     Price      Line total")
print("-" * 50)

# Look at each line of the basket in turn.
for item_name, unit_price, quantity in basket:
    # Work out what this line costs.
    line_total = unit_price * quantity

    # Add it to the running total.
    running_total = running_total + line_total

    # Print the line, with columns aligned.
    # :<15 means left-aligned in 15 characters, :>3 right-aligned in 3.
    print(f"{item_name:<15} {quantity:>3}  {unit_price:>8,.2f}  {line_total:>10,.2f}")

# Work out the tax and the final amount.
tax_rate = 0.18
tax_amount = running_total * tax_rate
final_amount = running_total + tax_amount

print("-" * 50)
print(f"{'Subtotal':<15} {'':>3}  {'':>8}  {running_total:>10,.2f}")
print(f"{'Tax (18%)':<15} {'':>3}  {'':>8}  {tax_amount:>10,.2f}")
print(f"{'TOTAL':<15} {'':>3}  {'':>8}  {final_amount:>10,.2f}")

# Report the most expensive line, as a small extra.
# max() with a key tells Python what to compare on.
most_expensive = max(basket, key=lambda entry: entry[1] * entry[2])

print(f"Largest line: {most_expensive[0]} at {most_expensive[1] * most_expensive[2]:,.2f}")
```

## EXAMPLE 9: Comments, and what they are for


```python
print("EXAMPLE 9: Comments")

# Anything after a # is ignored by Python. It is there for humans.

# A GOOD comment explains WHY something is done.
# Apply the standard rate, which finance confirmed on 2026-01-15.
standard_rate = 0.18

# A POOR comment restates WHAT the code already says.
# Set standard_rate to 0.18
another_rate = 0.18

print("Both rates are", standard_rate, "and", another_rate)
print("The first comment tells you something the code cannot.")
print("The second just repeats it, and will rot when the code changes.")
print("In this course every meaningful line carries a comment, because")
print("it is teaching material. Production code needs fewer.")
```
