## SECTION A: Concept checks


```python
concept_questions = [
    (
        "Q1. How do you find out exactly which Python is running your code?",
        "Print sys.executable. It gives the full path to the interpreter, "
        "which settles any confusion when several Pythons are installed.",
    ),
    (
        "Q2. In the REPL, why does `x = 5` display nothing?",
        "Assignment is a statement, not an expression - it produces no "
        "value. The REPL only displays values, and it hides None.",
    ),
    (
        "Q3. What does input() always return, and why does that matter?",
        "Always a string, even when the user types digits. You must convert "
        "with int() or float() before doing arithmetic, or you get a "
        "TypeError.",
    ),
    (
        "Q4. What is the difference between / and // ?",
        "/ is true division and always returns a float. // is floor "
        "division and rounds down to a whole number. 7/2 is 3.5, 7//2 is 3.",
    ),
    (
        "Q5. How do you tell whether you are inside a virtual environment?",
        "Compare sys.prefix with sys.base_prefix. They differ inside a "
        "virtual environment and match outside one.",
    ),
    (
        "Q6. Why use `python3 -m pip install` instead of `pip install`?",
        "It guarantees pip installs into the interpreter you are actually "
        "running. Plain `pip` can belong to a different Python, giving you "
        "ModuleNotFoundError for a package that just installed fine.",
    ),
    (
        "Q7. What does `if __name__ == '__main__':` actually do?",
        "__name__ is '__main__' when a file is run directly, and the module "
        "name when it is imported. The guard means that block runs only on "
        "direct execution, so one file can be both script and library.",
    ),
    (
        "Q8. Why should you never name a file json.py?",
        "It shadows the standard library module. Your file gets imported "
        "instead of the real one, producing errors that are very hard to "
        "diagnose.",
    ),
    (
        "Q9. What belongs in .gitignore - .venv or requirements.txt?",
        ".venv is ignored: it is large, machine-specific and rebuildable. "
        "requirements.txt is committed, because it is what lets anyone "
        "else recreate the environment.",
    ),
    (
        "Q10. Why are spaces-not-tabs a syntax matter in Python?",
        "Indentation defines block structure. Mixing tabs and spaces makes "
        "Python unable to work out the nesting, raising IndentationError "
        "or TabError.",
    ),
]

for question, answer in concept_questions:
    print("")
    print(question)
    print("   ANSWER:", answer)
```

## SECTION B: Predict the output


```python
# B1
print("B1.  print(7 / 2)  and  print(7 // 2)  and  print(7 % 2)")
print("     RESULT:", 7 / 2, "|", 7 // 2, "|", 7 % 2)
print("     WHY: true division, floor division, remainder.")

# B2
print("B2.  print('5' + '3')  and  print(5 + 3)")
print("     RESULT:", repr("5" + "3"), "|", 5 + 3)
print("     WHY: + joins strings but adds numbers. Same symbol, two jobs.")

# B3
print("B3.  print(type(input_value)) where input_value came from input()")
simulated_input = "42"
print("     RESULT:", type(simulated_input).__name__)
print("     WHY: input() always returns str, never int.")

# B4
print("B4.  print(int('42') + 8)")
print("     RESULT:", int("42") + 8)
print("     WHY: int() converts the text first, so arithmetic works.")

# B5
print("B5.  print(bool(0), bool(''), bool([]), bool('no'))")
print("     RESULT:", bool(0), bool(""), bool([]), bool("no"))
print("     WHY: zero, empty string and empty list are falsy.")
print("          Any non-empty string is truthy - even 'no' and 'False'.")

# B6
print("B6.  print(0.1 + 0.2 == 0.3)")
print("     RESULT:", 0.1 + 0.2 == 0.3)
print("     ACTUAL SUM:", 0.1 + 0.2)
print("     WHY: floats are binary approximations. Chapter 05 explains it.")
```

## SECTION C: Fix the bug


```python
# C1 - the input conversion bug.
print("C1. This code is meant to add 10 to the user's number:")
print("       number = input('Enter a number: ')")
print("       print(number + 10)")
print("    BUG: input() returns text, so this raises TypeError.")

# Demonstrate the error safely.
user_text = "5"
try:
    broken = user_text + 10
except TypeError as error:
    print("    ERROR:", error)

print("    FIX: number = int(input('Enter a number: '))")
print("    WORKS:", int(user_text) + 10)

# C2 - the shadowing bug.
print("C2. This code is meant to sum a list:")
print("       sum = 0")
print("       print(sum([1, 2, 3]))")
print("    BUG: assigning to `sum` hides the built-in function.")


def demonstrate_c2():
    """Show the shadowing failure inside a contained scope."""
    # This assignment hides the built-in sum() within this function.
    sum = 0
    try:
        sum([1, 2, 3])
    except TypeError as error:
        print("    ERROR:", error)


demonstrate_c2()
print("    FIX: name the variable total, not sum.")
print("    WORKS:", sum([1, 2, 3]))

# C3 - the indentation bug.
print("C3. This code will not even compile:")
print("       for i in range(3):")
print("       print(i)")

bad_code = "for i in range(3):\nprint(i)\n"
try:
    compile(bad_code, "<exercise>", "exec")
except IndentationError as error:
    print("    ERROR: IndentationError -", error.msg)

print("    FIX: indent the body of the loop by four spaces.")
```
