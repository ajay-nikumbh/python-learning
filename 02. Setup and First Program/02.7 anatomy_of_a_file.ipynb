# 02.7 Anatomy of a Python File

Every real Python file follows a conventional layout. Learning it now means your
code looks professional from the first project, and means you can read other
people's files without confusion.

The standard order is:

1. Shebang line (Unix only, optional)
2. Module docstring
3. Imports — standard library, third-party, then local
4. Constants
5. Functions and classes
6. The main guard

This notebook walks through each part.


## 1. Imports

PEP 8 specifies three groups, separated by a blank line: standard library,
third-party packages, then your own modules.


````python
# Group 1: standard library.
import sys
from pathlib import Path

# Group 2 would be third-party imports, for example:
#   import requests

# Group 3 would be imports from your own project, for example:
#   from myproject.helpers import clean_text

print("Imports done. sys and Path are now available.")
````

## 2. Module-level constants

Constants use `UPPER_CASE` names. Python does not enforce this, but every Python
programmer reads it as "do not change this at runtime".


````python
# Values used throughout the file, named so their role is obvious.
DEFAULT_TAX_RATE = 0.18
MAXIMUM_RETRIES = 3
APPLICATION_NAME = "Anatomy Demo"

print("Tax rate:", DEFAULT_TAX_RATE)
print("Max retries:", MAXIMUM_RETRIES)
print("App name:", APPLICATION_NAME)
````

## 3. Functions

Functions come after the constants. Each carries a docstring: the first line is a
short summary written as an instruction, and editors read this text.


````python
def calculate_total(price: float, quantity: int, tax_rate: float = DEFAULT_TAX_RATE) -> float:
    """Work out a total cost including tax.

    Args:
        price: Cost of a single item.
        quantity: Number of items.
        tax_rate: Tax rate as a decimal. Defaults to DEFAULT_TAX_RATE.

    Returns:
        The total cost with tax included.
    """
    # Work out the cost before tax.
    subtotal = price * quantity

    # Apply the tax and hand the answer back.
    return subtotal * (1 + tax_rate)


# Call it with the default tax rate.
print("4 items at 250.00, default tax:", round(calculate_total(250.0, 4), 2))

# Call it again, overriding the default.
print("4 items at 250.00, 5% tax:     ", round(calculate_total(250.0, 4, 0.05), 2))
````

## 4. The special variables Python provides

Python sets several dunder (double-underscore) variables in every module.


````python
# __name__ is the module's name. In a notebook it is "__main__".
print("__name__ is:", repr(__name__))

# __doc__ is the module docstring - the triple-quoted text at the top of a file.
# A notebook has no module docstring, so this is None here.
print("__doc__ is:", repr(__doc__))
````

## 5. The main guard

Python sets `__name__` to `"__main__"` when a file is **run directly**, and to the
module's name when it is **imported**.

```python
if __name__ == "__main__":
    # Runs only on direct execution, never on import.
```

This is what lets one file be both a runnable script *and* an importable library.
Without it, importing the file would execute everything inside it.


````python
# In a notebook __name__ is always "__main__", so a guard like this always runs.
if __name__ == "__main__":
    print("This block ran, because __name__ is", repr(__name__))

print("")
print("If another file did `import anatomy_of_a_file`, then:")
print("   - the imports, constants and function definitions would run")
print("   - the guarded block would NOT run")
print("   - __name__ would be 'anatomy_of_a_file' instead")
````

## 6. The shebang line

On macOS and Linux, a file can start with:

```python
#!/usr/bin/env python3
```

That must be the **very first line**, before the docstring. Combined with
`chmod +x file.py`, it lets you run `./file.py` instead of `python3 file.py`.

Windows ignores it.


## 7. File naming rules

````python
# Each entry is (example, verdict, reason).
naming = [
    ("my_script.py", "good", "lowercase with underscores"),
    ("data_cleaner.py", "good", "describes what it does"),
    ("MyScript.py", "avoid", "capitals are for class names"),
    ("my-script.py", "broken", "cannot be imported - hyphen is minus"),
    ("2024_report.py", "broken", "cannot start with a digit"),
    ("json.py", "dangerous", "shadows the standard library module"),
]

print("Filename              Verdict     Why")
print("-" * 68)

for filename, verdict, reason in naming:
    # ljust() pads each column so the output lines up.
    print(filename.ljust(22), verdict.ljust(11), reason)

print("")
print("The last one causes baffling bugs: your own file gets imported")
print("instead of the real library.")
````

## 8. Inspecting a file on disk

`pathlib` turns a filename into an object with useful properties. Covered fully
in Chapter 21.


````python
# Point at the notebook's own folder, since a notebook has no __file__.
this_folder = Path.cwd()

print("Current folder:", this_folder.name)
print("Full path:     ", this_folder)
print("Exists?        ", this_folder.exists())

print("")
print("Python files in this folder:")

# glob() finds files matching a pattern; sorted() puts them in order.
for found in sorted(this_folder.glob("*.ipynb"))[:5]:
    print("   ", found.name)
````

## Takeaways

1. Order: shebang, docstring, imports, constants, code, main guard.
2. Imports go in three groups: standard library, third-party, local.
3. Constants are `UPPER_CASE` by convention.
4. `__name__` is `"__main__"` when run, the module name when imported.
5. The main guard lets one file be both script and library.
6. Name files `lowercase_with_underscores.py`.
7. Never name a file after a standard library module.
