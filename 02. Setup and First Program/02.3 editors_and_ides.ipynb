# 02.3 Editors and IDEs

You can write Python in Notepad. You should not.

A good editor catches mistakes as you type, shows you what a function
expects, and lets you run code without leaving the window. That feedback
loop is worth more than any tutorial.

This file shows what editor features actually do for you, using real
code you can inspect.


## EXAMPLE 1: What an editor reads to help you


```python
print("EXAMPLE 1: Where autocomplete information comes from")

# Editors are not magic. They read the same information you can read
# from Python itself: docstrings, type hints, and signatures.


def calculate_total(price: float, quantity: int, tax_rate: float = 0.18) -> float:
    """Work out the total cost including tax.

    Args:
        price: The cost of one item.
        quantity: How many items are being bought.
        tax_rate: The tax rate as a decimal, defaulting to 18 percent.

    Returns:
        The total cost with tax applied.
    """
    # Multiply out the items, then add the tax on top.
    subtotal = price * quantity
    return subtotal * (1 + tax_rate)


# The inspect module reads a function the way an editor does.
import inspect

# signature() shows the parameters, their types, and their defaults.
# This is exactly what your editor displays in its popup hint.
print("What your editor shows when you type 'calculate_total(':")
print("   calculate_total" + str(inspect.signature(calculate_total)))

print("The docstring it shows alongside:")

# Print the first three lines of the docstring, indented.
for line in calculate_total.__doc__.splitlines()[:3]:
    print("   ", line)

print("Call it to check the maths:")
total = calculate_total(price=100.0, quantity=3)
print("   100.0 x 3 with 18% tax =", round(total, 2))
```

## EXAMPLE 2: Type hints power the editor's warnings


```python
print("EXAMPLE 2: How type hints help before you run anything")

# __annotations__ holds the type hints as a dictionary.
print("The hints your editor reads from calculate_total:")

# .items() gives us each key and value together.
for parameter_name, expected_type in calculate_total.__annotations__.items():
    # __name__ gives the short type name.
    print("   ", parameter_name.ljust(10), "->", expected_type.__name__)

print("Because of these, a good editor underlines this in red:")
print('   calculate_total("100", "3")     <- strings, not numbers')
print("It catches the mistake BEFORE you run the program.")
print("Type hints are covered fully in Chapter 32.")
```

## EXAMPLE 3: What a linter catches


```python
print("EXAMPLE 3: Problems a linter finds automatically")

# A linter reads your code and reports likely mistakes. Here are the
# kinds of thing it catches, written out as examples rather than run.

linter_catches = [
    ("unused variable", "result = compute()   # result never used again"),
    ("unused import", "import json          # json never referenced"),
    ("undefined name", "print(totl)          # typo for 'total'"),
    ("shadowing a built-in", "list = [1, 2, 3]     # hides the built-in list"),
    ("mutable default", "def f(items=[]):     # the Chapter 16 trap"),
    ("unreachable code", "return x             # code after this never runs"),
]

print("Problem                Example")
print("-" * 60)

for problem, example in linter_catches:
    # ljust() pads the first column so everything lines up.
    print(problem.ljust(22), example)

print("Every one of these is a real bug that costs time to find by hand.")
```

## EXAMPLE 4: Shadowing a built-in, demonstrated safely


```python
print("EXAMPLE 4: Why shadowing matters")

# This is the mistake from Example 3, shown concretely.

# Normally, list() builds a list from something iterable.
print("Normally, list('abc') gives:", list("abc"))

# Now we shadow it - using 'list' as a variable name hides the built-in.
# We do this inside a function so the damage is contained to this scope.
def demonstrate_shadowing():
    """Show what happens when a built-in name is reused."""
    # This assignment hides the built-in list() inside this function.
    list = [1, 2, 3]

    # The name now refers to our data, not the function.
    print("   After 'list = [1, 2, 3]', list is:", list)

    # Trying to call it as a function now fails.
    try:
        list("abc")
    except TypeError as error:
        print("   Calling list('abc') now raises TypeError:")
        print("     ", error)


demonstrate_shadowing()

# Outside that function, the built-in is untouched.
print("Outside the function, list('xyz') still works:", list("xyz"))
print("An editor would have warned you the moment you typed it.")
```

## EXAMPLE 6: Settings that prevent real problems


```python
print("EXAMPLE 6: Editor settings worth changing on day one")

# Each entry is (setting, why it matters).
settings = [
    ("Insert spaces, not tabs", "mixing them is a syntax error in Python"),
    ("Tab size = 4", "PEP 8 requires four spaces per level"),
    ("Show whitespace", "makes invisible indentation errors visible"),
    ("Trim trailing whitespace", "keeps diffs clean in version control"),
    ("Format on save", "ends every argument about style"),
    ("Rulers at column 88", "keeps lines readable side by side"),
]

print("Setting                     Why")
print("-" * 70)

for setting, reason in settings:
    print(setting.ljust(28), reason)

print("The tabs-versus-spaces one is not a style preference in Python.")
print("Mixing them genuinely stops your program from running.")
```

## EXAMPLE 7: Proof that indentation is syntax


```python
print("EXAMPLE 7: Python reads your indentation")

# In most languages, indentation is decoration. In Python it defines
# the structure, which is why editor settings matter so much.

# Two versions of the same code, differing only in indentation.
correctly_indented = """
for number in [1, 2, 3]:
    print(number)
"""

badly_indented = """
for number in [1, 2, 3]:
print(number)
"""

# compile() checks the syntax without running anything.
# We catch the error rather than letting it stop the file.
for label, source in [("Correct", correctly_indented), ("Bad", badly_indented)]:
    try:
        compile(source, "<example>", "exec")
        print(label.ljust(9), "- compiles fine")
    except IndentationError as error:
        print(label.ljust(9), "- IndentationError:", error.msg)

print("This is why 'show whitespace' is a genuinely useful setting.")
```
