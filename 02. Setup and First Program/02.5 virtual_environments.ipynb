# 02.5 Virtual Environments

A virtual environment is a private copy of Python for one project, with
its own installed packages.

Without them, every project shares one set of packages. Project A needs
a library at version 1, project B needs version 2, and installing one
breaks the other. This is called dependency hell, and venv is how you
avoid it entirely.

This file inspects the environment it is running in, so you can tell
whether you are inside one.


## EXAMPLE 1: Am I inside a virtual environment right now?


```python
print("EXAMPLE 1: Detecting a virtual environment")

import sys

# sys.prefix points at the Python installation currently in use.
# sys.base_prefix points at the original system Python.
# Inside a virtual environment these two differ. Outside, they match.
print("sys.prefix      (active Python):")
print("   ", sys.prefix)
print("sys.base_prefix (system Python):")
print("   ", sys.base_prefix)

# This comparison is the standard, reliable check.
inside_venv = sys.prefix != sys.base_prefix

if inside_venv:
    print("You ARE inside a virtual environment.")
else:
    print("You are NOT inside a virtual environment.")
    print("That is fine for this course, but real projects should use one.")
```

## EXAMPLE 2: Where packages get installed


```python
print("EXAMPLE 2: Where pip would install a package")

# site tells us the folder where third-party packages live.
import site

# getsitepackages() returns the folders searched for installed packages.
# Some environments return several; we show what is available.
try:
    package_folders = site.getsitepackages()
except AttributeError:
    # Very unusual environments may not provide this function.
    package_folders = []

if package_folders:
    print("Packages install into:")
    for folder in package_folders:
        print("   ", folder)
else:
    print("This environment does not report its site-packages folders.")

print("Inside a virtual environment this path sits within the project")
print("folder, so the packages belong to that project alone.")
```

## EXAMPLE 3: The commands, in order


```python
print("EXAMPLE 3: Creating and using a virtual environment")

# Each entry is (step, command, what it does).
workflow = [
    ("1. Create",
     "python3 -m venv .venv",
     "makes a .venv folder holding a private Python"),
    ("2. Activate (macOS/Linux)",
     "source .venv/bin/activate",
     "points your shell at that private Python"),
    ("2. Activate (Windows)",
     r".venv\Scripts\activate",
     "the same thing, Windows syntax"),
    ("3. Install",
     "pip install jupyterlab",
     "installs into .venv only, not system-wide"),
    ("4. Record",
     "pip freeze > requirements.txt",
     "writes down exactly what is installed"),
    ("5. Leave",
     "deactivate",
     "returns your shell to the system Python"),
]

for step, command, explanation in workflow:
    print("")
    print("  " + step)
    print("     $ " + command)
    print("       " + explanation)
```

## EXAMPLE 4: How to tell it worked


```python
print("EXAMPLE 4: Confirming activation")

print("Once activated, your shell prompt gains a prefix:")
print("   before:  ajay@laptop ~/project $")
print("   after:   (.venv) ajay@laptop ~/project $")
print("And the python command now points inside the project:")
print("   $ which python          # macOS and Linux")
print("   /Users/you/project/.venv/bin/python")
print("   > where python          # Windows")
print("   C:\\Users\\you\\project\\.venv\\Scripts\\python.exe")
print("If the path does not contain .venv, activation did not work.")
```

## EXAMPLE 5: The problem virtual environments solve


```python
print("EXAMPLE 5: Why this matters")

# Two imaginary projects with conflicting requirements.
project_requirements = {
    "website-2024": {"django": "3.2", "requests": "2.25"},
    "website-2026": {"django": "5.0", "requests": "2.31"},
}

print("Two projects on one machine:")

for project, packages in project_requirements.items():
    print("  " + project)
    for package_name, version in packages.items():
        print(f"     needs {package_name} {version}")

print("WITHOUT virtual environments:")
print("   installing django 5.0 for the new project breaks the old one,")
print("   because there is only one shared copy of each package.")
print("WITH virtual environments:")
print("   each project has its own .venv folder and its own django,")
print("   so both work, and neither knows about the other.")
```

## EXAMPLE 6: requirements.txt - making it reproducible


```python
print("EXAMPLE 6: Recording what you installed")

# A requirements file is a plain list of packages and exact versions.
example_requirements = """jupyterlab==4.2.0
ruff==0.5.0
pytest==8.2.0"""

print("A typical requirements.txt:")
for line in example_requirements.splitlines():
    print("   ", line)

print("Generate it from your current environment:")
print("   $ pip freeze > requirements.txt")
print("Recreate it on another machine:")
print("   $ python3 -m venv .venv")
print("   $ source .venv/bin/activate")
print("   $ pip install -r requirements.txt")
print("The == pins an exact version, so everyone gets identical packages.")
print("That is what makes a build reproducible.")
```

## EXAMPLE 7: Rules worth following


```python
print("EXAMPLE 7: Practical rules")

rules = [
    ("One environment per project", "never share one between projects"),
    ("Name it .venv", "tools and editors look for this name"),
    ("Never commit .venv", "add it to .gitignore - it is large and rebuildable"),
    ("Always commit requirements.txt", "this is what others actually need"),
    ("Activate before installing", "otherwise pip installs system-wide"),
    ("Delete freely", "rm -rf .venv then recreate - nothing is lost"),
]

print("Rule                             Why")
print("-" * 72)

for rule, reason in rules:
    print(rule.ljust(33), reason)

print("The last one surprises people: a virtual environment is disposable.")
print("Everything needed to rebuild it lives in requirements.txt.")
```

## TAKEAWAYS


```python
print("TAKEAWAYS")
print("1. A virtual environment is a private Python for one project.")
print("2. sys.prefix differs from sys.base_prefix when you are inside one.")
print("3. Create with: python3 -m venv .venv")
print("4. Activate with: source .venv/bin/activate  (or Scripts on Windows)")
print("5. pip install then affects only that project.")
print("6. Commit requirements.txt, never commit .venv itself.")
print("7. Environments are disposable - delete and rebuild at will.")
```
