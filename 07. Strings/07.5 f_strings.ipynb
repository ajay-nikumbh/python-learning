# 07.5 f-strings

An **f-string** puts values directly inside text. Write `f` before the quote,
then put any expression in `{braces}`.

This is the way to build strings in modern Python. Everything else is legacy.

## Easy — Putting values into text

The basic idea, one small step at a time.

```python
# EXAMPLE 1: Your first f-string
# The f before the quote is what makes the braces work.
name = "Asha"

print(f"Hello, {name}")
```

```python
# EXAMPLE 2: Without the f, nothing happens
# Forgetting the f is the most common f-string mistake.
name = "Asha"

print(f"With f:    Hello, {name}")
print("Without f: Hello, {name}")
```

```python
# EXAMPLE 3: Using several values
name = "Asha"
age = 31

print(f"{name} is {age} years old")
```

```python
# EXAMPLE 4: Doing arithmetic inside the braces
# Any expression works, not just a plain name.
age = 31

print(f"Next year you will be {age + 1}")
```

```python
# EXAMPLE 5: Calling a function inside the braces
name = "asha"

print(f"Hello, {name.upper()}")
```

```python
# EXAMPLE 6: Using a list item
scores = [90, 85, 77]

print(f"The first score is {scores[0]}")
```

```python
# EXAMPLE 7: Using a dictionary value
# Use a different quote style inside the braces.
person = {"name": "Asha", "city": "Mumbai"}

print(f"{person['name']} lives in {person['city']}")
```

```python
# EXAMPLE 8: Showing a literal brace
# Double the brace to print an actual brace.
print(f"Use {{braces}} like this")
```

```python
# EXAMPLE 9: Multi-line f-strings
name = "Asha"
age = 31

message = f"""Name: {name}
Age: {age}"""

print(message)
```

## Medium — Formatting numbers and alignment

Add a colon after the expression to control how it is displayed.

```python
# EXAMPLE 10: Rounding to two decimal places
# :.2f means "a float with 2 decimal places".
price = 1234.5678

print(f"Price: {price:.2f}")
```

```python
# EXAMPLE 11: Different decimal places
value = 3.14159

print(f"0 places: {value:.0f}")
print(f"1 place:  {value:.1f}")
print(f"3 places: {value:.3f}")
```

```python
# EXAMPLE 12: Adding thousands separators
# :, inserts commas between thousands.
population = 1400000000

print(f"Population: {population:,}")
```

```python
# EXAMPLE 13: Combining separators and decimals
amount = 1234567.891

print(f"Amount: {amount:,.2f}")
```

```python
# EXAMPLE 14: Showing a percentage
# :.1% multiplies by 100 and adds a percent sign.
share = 0.4567

print(f"Market share: {share:.1%}")
```

```python
# EXAMPLE 15: Aligning text to a width
# < is left, > is right, ^ is centre.
name = "Asha"

print(f"[{name:<10}]")
print(f"[{name:>10}]")
print(f"[{name:^10}]")
```

```python
# EXAMPLE 16: Padding with a chosen character
name = "Asha"

print(f"{name:.<10}")
print(f"{name:.>10}")
print(f"{name:*^10}")
```

```python
# EXAMPLE 17: Building an aligned table
rows = [("Apple", 10, 1.5), ("Banana", 5, 0.75), ("Cherry", 120, 12.0)]

print(f"{'Item':<10}{'Qty':>5}{'Price':>10}")
print("-" * 25)
for name, quantity, price in rows:
    print(f"{name:<10}{quantity:>5}{price:>10.2f}")
```

```python
# EXAMPLE 18: Zero-padding a number
# Useful for invoice numbers, IDs and times.
for number in [7, 42, 512]:
    print(f"INV-{number:05d}")
```

```python
# EXAMPLE 19: Formatting in different bases
value = 255

print(f"decimal: {value:d}")
print(f"binary:  {value:b}")
print(f"octal:   {value:o}")
print(f"hex:     {value:x}")
print(f"HEX:     {value:X}")
```

```python
# EXAMPLE 20: Scientific notation
value = 1234567.891

print(f"scientific: {value:e}")
print(f"with 2 dp:  {value:.2e}")
```

```python
# EXAMPLE 21: Formatting a date
# Date objects accept their own format codes after the colon.
from datetime import date

today = date(2026, 9, 13)

print(f"ISO:   {today}")
print(f"Long:  {today:%d %B %Y}")
print(f"Short: {today:%d/%m/%y}")
```

## Hard — Debugging, nesting and the details

The features that make f-strings genuinely powerful.

```python
# EXAMPLE 22: The = suffix for debugging
# Adding = prints the expression AND its value.
count = 42
name = "Asha"

print(f"{count=}")
print(f"{name=}")
print(f"{count * 2=}")
```

```python
# EXAMPLE 23: Why = is so useful while debugging
# Compare the two styles.
width = 10
height = 3

print("Manual:", "width =", width, "height =", height)
print(f"With =: {width=} {height=}")
print("")
print("The second cannot drift out of sync with the variable names.")
```

```python
# EXAMPLE 24: Choosing repr instead of str
# !r calls repr() instead of str(), showing quotes and escapes.
text = "  hello  "

print(f"with str:  {text}")
print(f"with repr: {text!r}")
```

```python
# EXAMPLE 25: Nesting a width from a variable
# Braces can appear inside the format spec itself.
name = "Asha"
column_width = 15

print(f"[{name:<{column_width}}]")
```

```python
# EXAMPLE 26: Nesting the precision too
value = 3.14159
places = 3

print(f"{value:.{places}f}")
```

```python
# EXAMPLE 27: Conditional expressions inside braces
count = 1

print(f"Found {count} item{'' if count == 1 else 's'}")

count = 3
print(f"Found {count} item{'' if count == 1 else 's'}")
```

```python
# EXAMPLE 28: What you cannot do inside braces
# Statements are not allowed - only expressions (see 03.1).
value = 5

# This works - a conditional EXPRESSION.
print(f"{'big' if value > 3 else 'small'}")

# These would be SyntaxErrors, so we test them with compile().
for source in ['f"{x = 5}"', 'f"{import os}"', 'f"{return 5}"']:
    try:
        compile(source, "<demo>", "eval")
        print(f"{source} compiled")
    except SyntaxError:
        print(f"{source:<16} SyntaxError - statements are not allowed")
```

```python
# EXAMPLE 29: f-strings are evaluated immediately
# The value is captured when the f-string is created, not when printed.
name = "Asha"
message = f"Hello, {name}"

name = "Ben"

print(message)
print("The message kept the ORIGINAL value.")
```

```python
# EXAMPLE 30: f-strings are faster than the alternatives
import time

name = "Asha"
age = 31
repetitions = 200000


def time_it(operation):
    """Return the milliseconds taken to repeat an operation."""
    start = time.perf_counter()
    for _ in range(repetitions):
        operation()
    return (time.perf_counter() - start) * 1000


fstring = time_it(lambda: f"{name} is {age}")
format_method = time_it(lambda: "{} is {}".format(name, age))
percent = time_it(lambda: "%s is %s" % (name, age))
concat = time_it(lambda: name + " is " + str(age))

print(f"f-string:   {fstring:.1f} ms")
print(f".format():  {format_method:.1f} ms")
print(f"% operator: {percent:.1f} ms")
print(f"+ concat:   {concat:.1f} ms")
```

```python
# EXAMPLE 31: Never use an f-string for SQL
# This is the most important security rule involving f-strings.
user_input = "'; DROP TABLE users; --"

# NEVER do this - the input becomes part of the query.
unsafe = f"SELECT * FROM users WHERE name = '{user_input}'"
print("UNSAFE query:")
print("  ", unsafe)

print("")
print("SAFE - let the database driver handle the value:")
print("   cursor.execute('SELECT * FROM users WHERE name = ?', (user_input,))")
print("")
print("The same applies to shell commands and file paths.")
print("Covered fully in Chapter 39 and Chapter 42.")
```

## Takeaways

1. `f"..."` embeds any **expression** in `{braces}`. Forgetting the `f` is the
   most common mistake.
2. Add `:spec` to format: `.2f` for decimals, `,` for thousands, `<>^` for
   alignment, `05d` for zero-padding.
3. `{value=}` prints both the expression and its value — excellent for debugging.
4. `!r` uses `repr()`, revealing quotes and whitespace.
5. Width and precision can themselves be **nested variables**: `{v:.{places}f}`.
6. Only **expressions** are allowed inside braces, never statements.
7. f-strings are evaluated **immediately** and are the **fastest** formatting
   option.
8. **Never build SQL or shell commands with f-strings** — use parameterised
   queries.
