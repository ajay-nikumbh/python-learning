# 07.6–07.7 Older Formatting and the Format Spec

f-strings are the modern choice, but you will meet two older styles in existing
code, and both are still occasionally the right tool.

This notebook also covers the **format mini-language** — the part after the colon
— in full, since it is shared by all three styles.

## Easy — The two older styles

Enough to read them when you meet them in existing code.

```python
# EXAMPLE 1: The .format() method
# Braces mark the slots; .format() supplies the values in order.
name = "Asha"
age = 31

print("{} is {} years old".format(name, age))
```

```python
# EXAMPLE 2: format() with numbered slots
# Numbers let you reuse or reorder values.
print("{0} {1} {0}".format("a", "b"))
```

```python
# EXAMPLE 3: format() with named slots
# Names are clearer than numbers when there are several values.
print("{name} is {age}".format(name="Asha", age=31))
```

```python
# EXAMPLE 4: The % operator
# The oldest style, borrowed from C.
name = "Asha"
age = 31

print("%s is %d years old" % (name, age))
```

```python
# EXAMPLE 5: The % placeholders
value = 42
text = "hello"
decimal = 3.14159

print("%s  - any value as text" % text)
print("%d  - a whole number" % value)
print("%f  - a float" % decimal)
print("%.2f - a float, 2 places" % decimal)
```

```python
# EXAMPLE 6: All three styles side by side
name = "Asha"
age = 31

print("f-string:  ", f"{name} is {age}")
print(".format(): ", "{} is {}".format(name, age))
print("% operator:", "%s is %d" % (name, age))
```

## Medium — When the older styles are still right

Three cases where f-strings are not the best answer.

```python
# EXAMPLE 7: A reusable template
# The template is defined once and filled many times.
template = "Dear {name}, your order {order} has shipped."

print(template.format(name="Asha", order="A123"))
print(template.format(name="Ben", order="B456"))
```

```python
# EXAMPLE 8: Filling from a dictionary
# format_map takes the values from a mapping.
person = {"name": "Asha", "city": "Mumbai"}

print("{name} lives in {city}".format_map(person))
```

```python
# EXAMPLE 9: Logging uses % style deliberately
# The logging module only formats the message if it is actually emitted.
import logging

logging.basicConfig(level=logging.WARNING, format="%(message)s")
logger = logging.getLogger("demo")

name = "Asha"

# CORRECT - the formatting is deferred until needed.
logger.warning("User %s logged in", name)

print("")
print("With an f-string the text is built even when the level is off,")
print("which wastes work in hot code paths. Chapter 37 covers logging.")
```

```python
# EXAMPLE 10: Unpacking a list into format()
# The star unpacks a sequence into the slots.
values = ["Asha", 31, "Mumbai"]

print("{} is {} and lives in {}".format(*values))
```

```python
# EXAMPLE 11: Choosing between the three
guidance = [
    ("Everyday string building", "f-string", "clearest and fastest"),
    ("A template reused later", ".format()", "the template can be a variable"),
    ("Values in a dictionary", ".format_map()", "no unpacking needed"),
    ("Logging messages", "% style", "formatting is deferred"),
    ("Very old codebases", "% style", "you will have to read it"),
]

for situation, choice, why in guidance:
    print(f"{situation:<26} {choice:<14} {why}")
```

## Hard — The format mini-language in full

Everything that can go after the colon. This applies to all three styles.

```python
# EXAMPLE 12: The full grammar
# Every part is optional, and they appear in this order.
print("[[fill]align][sign][#][0][width][grouping][.precision][type]")
print("")
print("fill      any character to pad with")
print("align     < left   > right   ^ centre   = after the sign")
print("sign      + always  - negative only  space for a leading space")
print("#         prefix 0b, 0o or 0x for binary, octal and hex")
print("0         zero-pad, shorthand for fill=0 align==")
print("width     minimum total characters")
print("grouping  , or _ as a thousands separator")
print(".prec     decimal places, or maximum characters for a string")
print("type      d b o x e f % s and others")
```

```python
# EXAMPLE 13: Alignment options
value = 42

print(f"[{value:<10}]  left")
print(f"[{value:>10}]  right")
print(f"[{value:^10}]  centre")
print(f"[{value:=10}]  after the sign")
```

```python
# EXAMPLE 14: Sign options
for value in [42, -42]:
    print(f"{value:+}  always show the sign")
    print(f"{value:-}  only for negatives (the default)")
    print(f"{value: }  space for positives, aligns columns")
    print("")
```

```python
# EXAMPLE 15: The # flag adds the base prefix
value = 255

print(f"without #: {value:b} {value:o} {value:x}")
print(f"with #:    {value:#b} {value:#o} {value:#x}")
```

```python
# EXAMPLE 16: Grouping with commas or underscores
value = 1234567

print(f"commas:      {value:,}")
print(f"underscores: {value:_}")
print(f"binary with underscores: {value:_b}")
```

```python
# EXAMPLE 17: Precision means different things
# For floats it is decimal places. For strings it is a maximum length.
number = 3.14159
text = "abcdefghij"

print(f"float with .3:  {number:.3f}")
print(f"string with .3: {text:.3}", "<- truncated to 3 characters")
```

```python
# EXAMPLE 18: Every numeric type code
value = 255

codes = [
    ("d", "decimal"),
    ("b", "binary"),
    ("o", "octal"),
    ("x", "hex lowercase"),
    ("X", "hex uppercase"),
    ("c", "the character with that code point"),
    ("n", "decimal, locale aware"),
]

for code, description in codes:
    print(f"{code}: {value:{code}}".ljust(18), description)
```

```python
# EXAMPLE 19: Every float type code
value = 1234.5678

codes = [
    ("f", "fixed point"),
    ("e", "scientific, lowercase e"),
    ("E", "scientific, uppercase E"),
    ("g", "general - picks f or e"),
    ("%", "percentage, multiplies by 100"),
]

for code, description in codes:
    print(f"{code}: {value:{code}}".ljust(30), description)
```

```python
# EXAMPLE 20: Combining many options at once
value = 1234567.891

print(f"{value:>20,.2f}   right, commas, 2 places")
print(f"{value:<20,.2f}   left")
print(f"{value:^20,.2f}   centre")
print(f"{value:*>20,.2f}   padded with stars")
print(f"{value:+20,.2f}   with a sign")
```

```python
# EXAMPLE 21: A realistic invoice
# Everything combined into something you might actually ship.
items = [
    ("Widget", 3, 19.99),
    ("Gadget", 12, 5.49),
    ("Doohickey", 1, 1299.00),
]

WIDTH = 42

print("INVOICE".center(WIDTH))
print("=" * WIDTH)
print(f"{'Item':<18}{'Qty':>5}{'Price':>9}{'Total':>10}")
print("-" * WIDTH)

subtotal = 0.0
for name, quantity, price in items:
    line_total = quantity * price
    subtotal += line_total
    print(f"{name:<18}{quantity:>5}{price:>9,.2f}{line_total:>10,.2f}")

tax = subtotal * 0.18
print("-" * WIDTH)
print(f"{'Subtotal':<32}{subtotal:>10,.2f}")
print(f"{'Tax (18%)':<32}{tax:>10,.2f}")
print("=" * WIDTH)
print(f"{'TOTAL':<32}{subtotal + tax:>10,.2f}")
```

```python
# EXAMPLE 22: Custom classes can define their own format
# __format__ lets your objects respond to a format spec.
class Temperature:
    """A temperature that can display in C or F."""

    def __init__(self, celsius):
        self.celsius = celsius

    def __format__(self, spec):
        # The spec is whatever appears after the colon.
        if spec == "f":
            return f"{self.celsius * 9 / 5 + 32:.1f}F"
        if spec == "c" or spec == "":
            return f"{self.celsius:.1f}C"
        raise ValueError(f"Unknown format spec: {spec}")


reading = Temperature(25)

print(f"default: {reading}")
print(f"celsius: {reading:c}")
print(f"fahrenheit: {reading:f}")
```

## Takeaways

1. **f-strings** are the default choice. The older styles exist for specific
   cases.
2. `.format()` is right for a **reusable template** stored in a variable, and
   `format_map()` for filling from a dictionary.
3. **`%` style is correct for logging**, because the formatting is deferred until
   the message is actually emitted.
4. The format spec is shared by all three:
   `[[fill]align][sign][#][0][width][,][.precision][type]`.
5. `.precision` means **decimal places** for numbers but **maximum length** for
   strings.
6. `#` adds the `0b`, `0o` or `0x` prefix; `,` and `_` group thousands.
7. Classes control their own formatting with `__format__`.
