# 07.4 String Methods

Python strings come with about 45 built-in methods. You do not need to memorise
them — but knowing which **families** exist means you will reach for the right
one instead of writing a loop.

## The families

| Family | Examples | For |
|---|---|---|
| **Case** | `upper`, `lower`, `title` | Changing capitalisation |
| **Whitespace** | `strip`, `lstrip`, `rstrip` | Cleaning input |
| **Search** | `find`, `index`, `count`, `startswith` | Locating text |
| **Split/join** | `split`, `join`, `splitlines`, `partition` | Breaking up and rebuilding |
| **Replace** | `replace`, `removeprefix`, `translate` | Changing content |
| **Test** | `isdigit`, `isalpha`, `isspace` | Validating |
| **Pad** | `ljust`, `rjust`, `center`, `zfill` | Aligning output |

Remember: **every one returns a new string** (07.3).

## Easy — The methods you will use daily

These ten cover most everyday string work.

```python
# EXAMPLE 1: Making text uppercase
name = "asha"

print(name.upper())
```

```python
# EXAMPLE 2: Making text lowercase
name = "ASHA"

print(name.lower())
```

```python
# EXAMPLE 3: Capitalising the first letter
name = "asha"

print(name.capitalize())
```

```python
# EXAMPLE 4: Title case for names
full_name = "asha nikumbh"

print(full_name.title())
```

```python
# EXAMPLE 5: Removing spaces from both ends
# strip() is the most-used cleaning method.
messy = "   hello   "

print(repr(messy))
print(repr(messy.strip()))
```

```python
# EXAMPLE 6: Removing spaces from one end
messy = "   hello   "

print("lstrip:", repr(messy.lstrip()))
print("rstrip:", repr(messy.rstrip()))
```

```python
# EXAMPLE 7: Splitting a sentence into words
# split() with no argument splits on any whitespace.
sentence = "the quick brown fox"

words = sentence.split()
print(words)
print("Number of words:", len(words))
```

```python
# EXAMPLE 8: Splitting on a specific character
# Pass the separator you want to split on.
csv_line = "name,age,city"

fields = csv_line.split(",")
print(fields)
```

```python
# EXAMPLE 9: Joining a list back into a string
# join() is called ON the separator, which surprises people at first.
words = ["the", "quick", "fox"]

print(" ".join(words))
print("-".join(words))
```

```python
# EXAMPLE 10: Replacing text
sentence = "I like cats"

print(sentence.replace("cats", "dogs"))
```

```python
# EXAMPLE 11: Counting occurrences
sentence = "the cat sat on the mat"

print("'the' appears", sentence.count("the"), "times")
print("'at' appears", sentence.count("at"), "times")
```

```python
# EXAMPLE 12: Checking how text starts and ends
filename = "report.pdf"

print("Starts with 'report'?", filename.startswith("report"))
print("Ends with '.pdf'?    ", filename.endswith(".pdf"))
```

## Medium — Searching, testing and padding

The next layer - still everyday, slightly more specific.

```python
# EXAMPLE 13: Finding the position of text
# find() returns the index, or -1 when not present.
sentence = "hello world"

print("Position of 'world':", sentence.find("world"))
print("Position of 'xyz':  ", sentence.find("xyz"))
```

```python
# EXAMPLE 14: find() versus index()
# index() raises instead of returning -1.
sentence = "hello world"

print("find:", sentence.find("xyz"))

try:
    sentence.index("xyz")
except ValueError as error:
    print("index raises:", error)
```

```python
# EXAMPLE 15: Searching from the right
# rfind finds the LAST occurrence.
path = "folder/subfolder/file.txt"

print("First slash:", path.find("/"))
print("Last slash: ", path.rfind("/"))
print("Filename:   ", path[path.rfind("/") + 1:])
```

```python
# EXAMPLE 16: Testing whether text is a number
# isdigit() checks every character is a digit.
for candidate in ["42", "3.14", "abc", ""]:
    print(f"{candidate!r:<8} isdigit -> {candidate.isdigit()}")
```

```python
# EXAMPLE 17: The full family of is-methods
samples = ["abc", "123", "abc123", "  ", "ABC", "Hello World"]

print("value          alpha  digit  alnum  space  upper  title")
print("-" * 60)
for text in samples:
    print(f"{text!r:<14} {text.isalpha():<6} {text.isdigit():<6} "
          f"{text.isalnum():<6} {text.isspace():<6} "
          f"{text.isupper():<6} {text.istitle()}")
```

```python
# EXAMPLE 18: Padding text to a fixed width
# ljust, rjust and center pad with spaces by default.
name = "Asha"

print("[" + name.ljust(10) + "]")
print("[" + name.rjust(10) + "]")
print("[" + name.center(10) + "]")
```

```python
# EXAMPLE 19: Padding with a custom character
name = "Asha"

print(name.ljust(10, "."))
print(name.rjust(10, "."))
print(name.center(10, "*"))
```

```python
# EXAMPLE 20: Zero-padding numbers
# zfill is made for numbers - it handles the minus sign correctly.
for number in ["7", "42", "-7"]:
    print(f"{number!r:<6} -> {number.zfill(5)}")
```

```python
# EXAMPLE 21: Building an aligned table
# Padding methods make simple reports readable.
rows = [("Apple", 10), ("Banana", 5), ("Cherry", 120)]

print("Item".ljust(10) + "Qty".rjust(5))
print("-" * 15)
for name, quantity in rows:
    print(name.ljust(10) + str(quantity).rjust(5))
```

```python
# EXAMPLE 22: Splitting only a limited number of times
# The second argument caps how many splits happen.
line = "key=value=extra"

print("all splits: ", line.split("="))
print("one split:  ", line.split("=", 1))
```

```python
# EXAMPLE 23: Splitting lines of text
# splitlines() handles all the different newline conventions.
document = "first line\nsecond line\nthird line"

for line in document.splitlines():
    print(line)
```

```python
# EXAMPLE 24: Partitioning around a separator
# partition() always returns three parts: before, separator, after.
setting = "timeout=30"

before, separator, after = setting.partition("=")
print("before:   ", repr(before))
print("separator:", repr(separator))
print("after:    ", repr(after))
```

```python
# EXAMPLE 25: Partition when the separator is missing
# Unlike split, partition always returns three items.
setting = "no separator here"

before, separator, after = setting.partition("=")
print("before:   ", repr(before))
print("separator:", repr(separator))
print("after:    ", repr(after))
```

## Hard — The methods with sharp edges

Correct behaviour that is easy to misread.

```python
# EXAMPLE 26: strip() removes CHARACTERS, not a prefix
# This is the most misunderstood string method.
filename = "example.txt"

print("Expected 'example':", filename.strip(".txt"))
print("")
print("strip() removed every '.', 't' and 'x' from BOTH ends.")
print("It treats the argument as a SET of characters.")
```

```python
# EXAMPLE 27: The correct way to remove a suffix
# removeprefix and removesuffix do what you actually meant.
filename = "example.txt"

print("removesuffix:", filename.removesuffix(".txt"))
print("removeprefix:", "prefix_name".removeprefix("prefix_"))
```

```python
# EXAMPLE 28: Removing a suffix before Python 3.9
# The older idiom, still common in existing code.
filename = "example.txt"
suffix = ".txt"

if filename.endswith(suffix):
    trimmed = filename[:-len(suffix)]
else:
    trimmed = filename

print(trimmed)
```

```python
# EXAMPLE 29: split() with no argument versus split(' ')
# They behave differently with repeated or leading spaces.
messy = "  a   b  "

print("split():    ", messy.split())
print("split(' '): ", messy.split(" "))
print("")
print("split() with no argument collapses runs of whitespace")
print("and ignores leading and trailing spaces. Usually what you want.")
```

```python
# EXAMPLE 30: title() gets apostrophes wrong
# title() capitalises after any non-letter, including an apostrophe.
name = "o'brien mcdonald"

print("title():", name.title())
print("")
print("For real names, capitalise word by word instead:")
fixed = " ".join(word[:1].upper() + word[1:] for word in name.split())
print("manual: ", fixed)
```

```python
# EXAMPLE 31: Case conversion is not always reversible
# Some languages have characters that do not round-trip.
german = "straße"

print("original:  ", german)
print("upper:     ", german.upper())
print("back down: ", german.upper().lower())
print("Round trip equal?", german.upper().lower() == german)
```

```python
# EXAMPLE 32: casefold() for proper case-insensitive comparison
# lower() is not enough for every language. casefold() is stricter.
first = "straße"
second = "STRASSE"

print("lower comparison:   ", first.lower() == second.lower())
print("casefold comparison:", first.casefold() == second.casefold())
```

```python
# EXAMPLE 33: translate() for bulk character replacement
# Faster than chaining several replace() calls.
table = str.maketrans({"a": "4", "e": "3", "i": "1"})

text = "alphabet size"
print("original:  ", text)
print("translated:", text.translate(table))
```

```python
# EXAMPLE 34: Removing characters with translate()
# Mapping to None deletes the character.
table = str.maketrans("", "", "aeiou")

text = "remove the vowels"
print(text.translate(table))
```

```python
# EXAMPLE 35: The complete method list
# Everything a string can do, grouped by family.
families = {
    "case": ["upper", "lower", "capitalize", "title", "swapcase", "casefold"],
    "whitespace": ["strip", "lstrip", "rstrip", "expandtabs"],
    "search": ["find", "rfind", "index", "rindex", "count",
               "startswith", "endswith"],
    "split/join": ["split", "rsplit", "splitlines", "join",
                   "partition", "rpartition"],
    "replace": ["replace", "removeprefix", "removesuffix", "translate",
                "maketrans"],
    "test": ["isalpha", "isdigit", "isalnum", "isspace", "isupper",
             "islower", "istitle", "isnumeric", "isdecimal",
             "isidentifier", "isprintable", "isascii"],
    "pad": ["ljust", "rjust", "center", "zfill"],
    "format": ["format", "format_map", "encode"],
}

total = 0
for family, methods in families.items():
    total += len(methods)
    print(f"{family:<12} ({len(methods):>2}) {', '.join(methods)}")

print("")
print("Total listed:", total)
print("Use help(str.method_name) to read about any of them.")
```

## Takeaways

1. Strings have ~45 methods in seven families. Learn the families, look up the
   details.
2. **Every method returns a new string** — assign the result.
3. `split()` with no argument collapses runs of whitespace; `split(" ")` does
   not.
4. `join()` is called **on the separator**: `", ".join(items)`.
5. **`strip(".txt")` removes characters, not a suffix.** Use `removesuffix()`.
6. `find()` returns `-1` when absent; `index()` raises `ValueError`.
7. `partition()` always returns three parts, even when the separator is missing.
8. Use `casefold()` rather than `lower()` for case-insensitive comparison.
9. `translate()` beats chained `replace()` calls for bulk changes.
