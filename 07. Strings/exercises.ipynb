```python
# EXAMPLE 25: Predict: which is faster, and why?
import time

count = 20000

start = time.perf_counter()
result = ""
for number in range(count):
    result += str(number)
plus_ms = (time.perf_counter() - start) * 1000

start = time.perf_counter()
result = "".join(str(number) for number in range(count))
join_ms = (time.perf_counter() - start) * 1000

print(f"+= :   {plus_ms:7.1f} ms")
print(f"join:  {join_ms:7.1f} ms")
```

```python
# EXAMPLE 28: Build it: a safe truncation
def truncate(text, limit):
    """Shorten text to limit characters, adding an ellipsis if cut."""
    if len(text) <= limit:
        return text
    # Leave room for the three dots.
    return text[:limit - 3] + "..."


for candidate in ["short", "a much longer piece of text"]:
    print(f"{candidate!r:<30} -> {truncate(candidate, 15)!r}")
```

```python
# EXAMPLE 29: Build it: strip accents
import unicodedata


def remove_accents(text):
    """Return the text with accent marks removed."""
    decomposed = unicodedata.normalize("NFD", text)
    return "".join(character for character in decomposed
                   if unicodedata.category(character) != "Mn")


for name in ["café", "naïve", "Zürich"]:
    print(f"{name:<10} -> {remove_accents(name)}")
```

```python
# EXAMPLE 30: Build it: a palindrome checker
import unicodedata


def is_palindrome(text):
    """Check whether text reads the same backwards, ignoring case and spaces."""
    # Keep only letters and digits, lowercased.
    cleaned = "".join(character.lower() for character in text
                      if character.isalnum())
    return cleaned == cleaned[::-1]


for candidate in ["racecar", "A man, a plan, a canal: Panama", "hello"]:
    print(f"{candidate!r:<36} {is_palindrome(candidate)}")
```

```python
# EXAMPLE 31: Build it: parse a log line
line = "2026-09-13 14:22:05 ERROR Disk full on /dev/sda1"

# Split off the fixed-width parts, keep the rest as the message.
date, time_part, level, message = line.split(" ", 3)

print("date:   ", date)
print("time:   ", time_part)
print("level:  ", level)
print("message:", message)
```

```python
# EXAMPLE 32: Build it: word wrap by hand
# Then compare with textwrap.
import textwrap


def wrap_manually(text, width):
    """Wrap text to a given width without using textwrap."""
    lines = []
    current = []
    length = 0

    for word in text.split():
        # +1 for the space that will join them.
        if length + len(word) + len(current) > width and current:
            lines.append(" ".join(current))
            current = []
            length = 0
        current.append(word)
        length += len(word)

    if current:
        lines.append(" ".join(current))

    return lines


paragraph = "The quick brown fox jumps over the lazy dog near the river"

print("Manual:")
for line in wrap_manually(paragraph, 25):
    print("  ", line)

print("")
print("textwrap:")
for line in textwrap.wrap(paragraph, 25):
    print("  ", line)
```
