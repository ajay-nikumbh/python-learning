# 07.13 Replace, Strip and Pad — Deep Dive

Eleven methods that change or align text. This family contains **the single most
misunderstood string method in Python**: `strip()`.

| Method | Signature | Does |
|---|---|---|
| `replace` | `replace(old, new, count=-1)` | swap text |
| `strip` | `strip(chars=None)` | remove **characters** from both ends |
| `lstrip` / `rstrip` | `lstrip(chars=None)` | one end only |
| `removeprefix` | `removeprefix(prefix)` | remove an exact prefix |
| `removesuffix` | `removesuffix(suffix)` | remove an exact suffix |
| `ljust` / `rjust` | `ljust(width, fill=" ")` | pad to a width |
| `center` | `center(width, fill=" ")` | pad both sides |
| `zfill` | `zfill(width)` | zero-pad, sign-aware |
| `expandtabs` | `expandtabs(tabsize=8)` | tabs to spaces |
| `translate` | `translate(table)` | bulk character mapping |

## Easy — Changing and cleaning text

The everyday operations.

```python
# EXAMPLE 1: replace() swaps text
sentence = "I like cats"

print(sentence.replace("cats", "dogs"))
```

```python
# EXAMPLE 2: replace() changes EVERY occurrence
sentence = "the cat sat on the mat"

print(sentence.replace("the", "THE"))
```

```python
# EXAMPLE 3: strip() removes whitespace from both ends
messy = "   hello   "

print(repr(messy))
print(repr(messy.strip()))
```

```python
# EXAMPLE 4: lstrip() and rstrip() do one end
messy = "   hello   "

print("lstrip:", repr(messy.lstrip()))
print("rstrip:", repr(messy.rstrip()))
```

```python
# EXAMPLE 5: strip() does not touch the middle
messy = "   hello   world   "

print(repr(messy.strip()))
print("The inner spaces are untouched.")
```

```python
# EXAMPLE 6: ljust, rjust and center pad to a width
name = "Asha"

print("[" + name.ljust(10) + "]")
print("[" + name.rjust(10) + "]")
print("[" + name.center(10) + "]")
```

```python
# EXAMPLE 7: zfill() pads numbers with zeros
for number in ["7", "42", "512"]:
    print(number.zfill(5))
```

```python
# EXAMPLE 8: Padding does nothing if already wide enough
name = "Alexandra"

print(repr(name.ljust(5)))
print("The string is longer than 5, so it is returned unchanged.")
```

```python
# EXAMPLE 9: All of these return NEW strings
text = "  hello  "

text.strip()

print("after calling strip():", repr(text))
print("you must assign:", repr(text.strip()))
```

## Medium — The arguments, and the strip() trap

This section contains the most important cell in the chapter.

```python
# EXAMPLE 10: THE TRAP: strip() takes CHARACTERS, not a prefix
# This is the most misunderstood method in Python.
filename = "example.txt"

print("Expected 'example':", filename.strip(".txt"))
```

```python
# EXAMPLE 11: Why that happened
filename = "example.txt"

print("strip('.txt') treats the argument as a SET of characters:")
print("   remove any of: . t x")
print("")
print("From the left:  'e' is not in the set, so it stops immediately")
print("From the right: 't', 'x', 't', '.' are all in the set - removed")
print("                then 'e' stops it")
print("")
print("result:", repr(filename.strip(".txt")))
```

```python
# EXAMPLE 12: A clearer demonstration of the character set
text = "xxhelloxx"

print("strip('x'):  ", repr(text.strip("x")))
print("strip('xh'): ", repr(text.strip("xh")), "<- also removed the h")
print("")
print("Order does not matter - it is a set, not a sequence.")
print("strip('hx') gives:", repr(text.strip("hx")))
```

```python
# EXAMPLE 13: The correct way: removeprefix and removesuffix
filename = "example.txt"

print("strip('.txt'):      ", repr(filename.strip(".txt")))
print("removesuffix('.txt'):", repr(filename.removesuffix(".txt")))
```

```python
# EXAMPLE 14: removeprefix and removesuffix are exact
# They remove the string only if it is actually there.
print("'test_file'.removeprefix('test_') ->", "test_file".removeprefix("test_"))
print("'test_file'.removeprefix('xyz_')  ->", "test_file".removeprefix("xyz_"))
print("")
print("No match means no change - and no error.")
```

```python
# EXAMPLE 15: They only remove ONE occurrence
text = "aaabbb"

print("removeprefix('a'):", text.removeprefix("a"))
print("Only the first 'a' went. strip would remove all three:")
print("strip('a'):       ", text.strip("a"))
```

```python
# EXAMPLE 16: The pre-3.9 idiom
# Still common in existing code.
filename = "example.txt"
suffix = ".txt"

if filename.endswith(suffix):
    trimmed = filename[:-len(suffix)]
else:
    trimmed = filename

print(trimmed)
print("")
print("The endswith check matters - without it, a file with no")
print("extension would lose its last four characters.")
```

```python
# EXAMPLE 17: replace() with a count limit
sentence = "the cat sat on the mat"

print("all:      ", sentence.replace("at", "AT"))
print("first one:", sentence.replace("at", "AT", 1))
print("first two:", sentence.replace("at", "AT", 2))
```

```python
# EXAMPLE 18: Using replace() to delete text
# Replacing with an empty string removes it.
text = "hello world"

print(text.replace("l", ""))
print(text.replace(" ", ""))
```

```python
# EXAMPLE 19: Padding with a custom character
name = "Asha"

print(name.ljust(12, "."))
print(name.rjust(12, "."))
print(name.center(12, "*"))
```

```python
# EXAMPLE 20: center() favours the right when uneven
# With an odd amount of padding, the extra space goes on the right.
name = "abc"

print(repr(name.center(8, "-")))
print("2 on the left, 3 on the right.")
```

```python
# EXAMPLE 21: zfill() handles the minus sign correctly
# This is why zfill exists rather than just using rjust.
for number in ["42", "-42"]:
    print(f"{number!r:<7} zfill(6)={number.zfill(6)!r:<10} "
          f"rjust(6,'0')={number.rjust(6, '0')!r}")

print("")
print("zfill puts the zeros AFTER the sign. rjust does not.")
```

```python
# EXAMPLE 22: expandtabs() converts tabs to spaces
row = "name\tage\tcity"

print("raw:", repr(row))
print("")
print("default (8):", row.expandtabs())
print("size 4:     ", row.expandtabs(4))
```

```python
# EXAMPLE 23: expandtabs() aligns to tab stops, not fixed widths
# The number of spaces depends on the current column.
lines = ["a\tb", "abcdefg\tb"]

for line in lines:
    print(repr(line), "->", repr(line.expandtabs(8)))

print("")
print("Both 'b' characters land at column 8.")
```

## Hard — translate, chaining and performance

Bulk operations and the cases that catch people out.

```python
# EXAMPLE 24: translate() maps many characters at once
# maketrans builds the mapping table.
table = str.maketrans({"a": "4", "e": "3", "i": "1"})

text = "alphabet size"

print("original:  ", text)
print("translated:", text.translate(table))
```

```python
# EXAMPLE 25: maketrans with two strings
# Equal-length strings map position by position.
table = str.maketrans("aeiou", "12345")

print("hello world".translate(table))
```

```python
# EXAMPLE 26: maketrans with a third argument deletes
# The third string lists characters to remove entirely.
table = str.maketrans("", "", "aeiou")

print("remove the vowels".translate(table))
```

```python
# EXAMPLE 27: translate() versus chained replace()
import time

text = "the quick brown fox jumps over the lazy dog " * 200
repetitions = 2000

table = str.maketrans({"a": "4", "e": "3", "i": "1", "o": "0", "u": "5"})


def with_replace():
    result = text
    for old, new in [("a", "4"), ("e", "3"), ("i", "1"),
                     ("o", "0"), ("u", "5")]:
        result = result.replace(old, new)
    return result


def with_translate():
    return text.translate(table)


start = time.perf_counter()
for _ in range(repetitions):
    with_replace()
replace_ms = (time.perf_counter() - start) * 1000

start = time.perf_counter()
for _ in range(repetitions):
    with_translate()
translate_ms = (time.perf_counter() - start) * 1000

print(f"chained replace: {replace_ms:7.1f} ms")
print(f"translate:       {translate_ms:7.1f} ms")
print("Same result?", with_replace() == with_translate())
```

```python
# EXAMPLE 28: Chained replace can corrupt data
# Each replace sees the output of the previous one.
text = "ab"

# Intending to swap a and b.
broken = text.replace("a", "b").replace("b", "a")
print("chained replace:", broken, "<- both became 'a'")

# translate does it in one pass.
table = str.maketrans("ab", "ba")
print("translate:      ", text.translate(table), "<- correct")
```

```python
# EXAMPLE 29: strip() with no argument strips ALL whitespace types
messy = "\t\n  hello  \r\n"

print("raw:     ", repr(messy))
print("stripped:", repr(messy.strip()))
print("")
print("Tabs, newlines and carriage returns all go.")
```

```python
# EXAMPLE 30: strip() and the non-breaking space
# A non-breaking space IS stripped by default in Python 3.
normal = " hello "
nbsp = "\u00a0hello\u00a0"

print("normal stripped:", repr(normal.strip()))
print("nbsp stripped:  ", repr(nbsp.strip()))
```

```python
# EXAMPLE 31: Chaining these methods
# Each returns a string, so they chain naturally.
messy = "  HELLO, World!  "

cleaned = messy.strip().lower().replace(",", "").replace("!", "")

print("original:", repr(messy))
print("cleaned: ", repr(cleaned))
```

```python
# EXAMPLE 32: A reusable cleaning function
import unicodedata


def clean_name(raw):
    """Normalise a name from user input."""
    # Remove surrounding whitespace.
    text = raw.strip()

    # Collapse internal runs of whitespace to single spaces.
    text = " ".join(text.split())

    # Normalise the Unicode encoding.
    text = unicodedata.normalize("NFC", text)

    # Capitalise each word without breaking apostrophes.
    return " ".join(word[:1].upper() + word[1:].lower()
                    for word in text.split())


for raw in ["  asha   NIKUMBH ", "o'BRIEN", "  jean-luc  "]:
    print(f"{raw!r:<22} -> {clean_name(raw)!r}")
```

```python
# EXAMPLE 33: Padding is not enough for real tables
# Padding counts CHARACTERS, but some render wider.
rows = [("Asha", 10), ("北京", 5), ("café", 120)]

print("Naive padding:")
for name, quantity in rows:
    print("   " + name.ljust(10) + str(quantity).rjust(5))

print("")
print("The CJK row misaligns because those characters are double-width.")
print("For terminal tables with mixed scripts, use a library such as wcwidth.")
```

```python
# EXAMPLE 34: f-strings can do the padding instead
# Often clearer than calling ljust and rjust.
name = "Asha"
quantity = 42

print("with methods: " + name.ljust(10) + str(quantity).rjust(5))
print(f"with f-string: {name:<10}{quantity:>5}")
print("")
print("f-strings also handle the number formatting in the same place.")
```

```python
# EXAMPLE 35: Complete reference for this family
text = "  Hello World  "

operations = [
    ("text.strip()", text.strip()),
    ("text.lstrip()", text.lstrip()),
    ("text.rstrip()", text.rstrip()),
    ("text.strip('Hd ')", text.strip("Hd ")),
    ("text.replace('o', '0')", text.replace("o", "0")),
    ("text.replace('o', '0', 1)", text.replace("o", "0", 1)),
    ("'x.txt'.removesuffix('.txt')", "x.txt".removesuffix(".txt")),
    ("'a_x'.removeprefix('a_')", "a_x".removeprefix("a_")),
    ("'5'.zfill(3)", "5".zfill(3)),
    ("'-5'.zfill(4)", "-5".zfill(4)),
    ("'ab'.center(6, '-')", "ab".center(6, "-")),
    ("'a\tb'.expandtabs(4)", "a\tb".expandtabs(4)),
]

for expression, result in operations:
    print(f"{expression:<32} -> {result!r}")
```

## Takeaways

1. **`strip(chars)` removes CHARACTERS, not a prefix or suffix.** It treats the
   argument as a *set* and removes any of them from both ends.
   `"example.txt".strip(".txt")` gives `"exampl"`, not `"example"`.
2. Use **`removeprefix()`** and **`removesuffix()`** for exact removal. They
   remove one occurrence, or nothing.
3. `replace(old, new, count)` — the third argument limits how many are changed.
4. **Chained `replace()` calls can corrupt data**, because each sees the previous
   result. Use `translate()` for simultaneous mapping.
5. `translate()` with `maketrans()` is both **correct and faster** for bulk
   character changes.
6. **`zfill()` is sign-aware**; `rjust(width, "0")` is not.
7. `expandtabs()` aligns to **tab stops**, so the number of spaces depends on the
   current column.
8. `strip()` with no argument removes **all whitespace types**, including tabs,
   newlines and non-breaking spaces.
9. Padding counts **characters**, not display width — CJK text misaligns.
