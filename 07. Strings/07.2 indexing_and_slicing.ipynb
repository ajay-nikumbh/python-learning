# 07.2 Indexing and Slicing

A string is a **sequence** of characters. You can reach any single character by
its position, or pull out a whole section.

This is one of the most useful skills in Python — and everything you learn here
works identically on lists and tuples later.

## The mental model

```
  P    y    t    h    o    n
  0    1    2    3    4    5      <- index from the start
 -6   -5   -4   -3   -2   -1      <- index from the end
```

Positions start at **0**, not 1. Negative positions count back from the end.

## Easy — Getting single characters

One character at a time. Nothing here is complicated.

````python
# EXAMPLE 1: The first character
# Positions start at 0, so the first character is at index 0.
word = "Python"

print(word[0])
````

````python
# EXAMPLE 2: The second character
# Index 1 is the SECOND character, not the first.
word = "Python"

print(word[1])
````

````python
# EXAMPLE 3: Walking through the first few
# Each index gives one character.
word = "Python"

print("index 0:", word[0])
print("index 1:", word[1])
print("index 2:", word[2])
````

````python
# EXAMPLE 4: The last character, the hard way
# len() gives the count, so the last index is len - 1.
word = "Python"

last_index = len(word) - 1
print("Length:", len(word))
print("Last index:", last_index)
print("Last character:", word[last_index])
````

````python
# EXAMPLE 5: The last character, the easy way
# Index -1 always means "the last one".
word = "Python"

print(word[-1])
````

````python
# EXAMPLE 6: Counting backwards
# Negative indices count from the end.
word = "Python"

print("word[-1]:", word[-1])
print("word[-2]:", word[-2])
print("word[-3]:", word[-3])
````

````python
# EXAMPLE 7: Going too far raises an error
# There is no index 10 in a 6-character string.
word = "Python"

try:
    print(word[10])
except IndexError as error:
    print("IndexError:", error)
````

````python
# EXAMPLE 8: Checking before indexing
# Test the length first when the index might be out of range.
word = "Python"
wanted = 10

if wanted < len(word):
    print(word[wanted])
else:
    print("Index", wanted, "is too large for a string of length", len(word))
````

````python
# EXAMPLE 9: Looping over every character
# A for loop gives you each character in turn - no indexing needed.
word = "Python"

for character in word:
    print(character)
````

````python
# EXAMPLE 10: Looping with the position
# enumerate() gives the position alongside the character.
word = "Python"

for position, character in enumerate(word):
    print(position, character)
````

## Medium — Slicing out sections

Slicing takes a RANGE of characters. The syntax is text[start:stop].

````python
# EXAMPLE 11: Your first slice
# text[start:stop] takes from start UP TO but NOT INCLUDING stop.
word = "Python"

print(word[0:2])
````

````python
# EXAMPLE 12: Why the stop is excluded
# The stop index is where slicing STOPS, not the last item taken.
word = "Python"

print("word[0:2] ->", word[0:2], "  (positions 0 and 1)")
print("word[0:3] ->", word[0:3], " (positions 0, 1 and 2)")
print("")
print("Handy result: the length of the slice is stop - start.")
print("3 - 0 =", 3 - 0, "and len(word[0:3]) =", len(word[0:3]))
````

````python
# EXAMPLE 13: Leaving out the start
# Omitting start means "from the beginning".
word = "Python"

print(word[:3])
````

````python
# EXAMPLE 14: Leaving out the stop
# Omitting stop means "to the end".
word = "Python"

print(word[3:])
````

````python
# EXAMPLE 15: Leaving out both
# [:] takes everything - a full copy.
word = "Python"

print(word[:])
````

````python
# EXAMPLE 16: Slicing with negative numbers
# Negative positions work in slices too.
word = "Python"

print("Last two characters:", word[-2:])
print("Everything except the last two:", word[:-2])
````

````python
# EXAMPLE 17: Splitting a string in half
# Use // to get a whole number for the midpoint.
word = "Python"

middle = len(word) // 2
print("First half: ", word[:middle])
print("Second half:", word[middle:])
````

````python
# EXAMPLE 18: A slice that is out of range does NOT error
# Unlike indexing, slicing clamps to what exists.
word = "Python"

print("word[0:100] ->", word[0:100])
print("word[10:20] ->", repr(word[10:20]), "- an empty string, not an error")
````

````python
# EXAMPLE 19: Using a step
# text[start:stop:step] takes every step-th character.
alphabet = "abcdefghij"

print("Every character: ", alphabet[::1])
print("Every 2nd:       ", alphabet[::2])
print("Every 3rd:       ", alphabet[::3])
````

````python
# EXAMPLE 20: Reversing a string
# A step of -1 walks backwards through the whole string.
word = "Python"

print(word[::-1])
````

````python
# EXAMPLE 21: Step with a start and stop
# All three parts together.
alphabet = "abcdefghij"

# From position 1, up to position 8, taking every 2nd character.
print(alphabet[1:8:2])
````

````python
# EXAMPLE 22: Taking the first and last few
# A common pattern for previewing long text.
text = "The quick brown fox jumps over the lazy dog"

print("First 10: ", text[:10])
print("Last 10:  ", text[-10:])
print("Preview:  ", text[:10] + "..." + text[-10:])
````

## Hard — Slice details and performance

The corners that matter once you use slicing heavily.

````python
# EXAMPLE 23: Slices always return a new string
# Slicing never modifies the original - strings are immutable.
word = "Python"

part = word[0:3]
print("original:", word)
print("slice:   ", part)
print("Same object?", part is word)
````

````python
# EXAMPLE 24: A full slice of a string returns the SAME object
# Because strings are immutable, Python can safely reuse the original.
word = "Python"
copy_attempt = word[:]

print("Same object?", copy_attempt is word)
print("")
print("For a LIST, [:] genuinely copies, because lists are mutable:")
numbers = [1, 2, 3]
print("list copy is the same object?", numbers[:] is numbers)
````

````python
# EXAMPLE 25: Negative steps need the start and stop swapped
# With a negative step, slicing runs from HIGH to LOW.
alphabet = "abcdefghij"

print("alphabet[2:6]      ->", alphabet[2:6])
print("alphabet[6:2:-1]   ->", alphabet[6:2:-1], "- note the order")
print("alphabet[2:6:-1]   ->", repr(alphabet[2:6:-1]), "- empty, wrong direction")
````

````python
# EXAMPLE 26: The slice object
# The [a:b:c] syntax builds a slice object behind the scenes.
word = "Python"

manual = slice(0, 3)
print("word[0:3]        ->", word[0:3])
print("word[slice(0,3)] ->", word[manual])
print("")
print("The slice object itself:", manual)
print("Its parts:", manual.start, manual.stop, manual.step)
````

````python
# EXAMPLE 27: Naming a slice makes repeated use readable
# A named slice documents what the positions mean.
record = "2026-09-13ERROR  disk full"

DATE = slice(0, 10)
LEVEL = slice(10, 15)
MESSAGE = slice(17, None)

print("date:   ", record[DATE])
print("level:  ", record[LEVEL])
print("message:", record[MESSAGE])
````

````python
# EXAMPLE 28: Slicing does not raise, so typos hide
# An out-of-range index shouts. An out-of-range slice stays silent.
word = "Python"

try:
    word[99]
except IndexError as error:
    print("Indexing:", error)

print("Slicing: ", repr(word[99:200]), "- no error at all")
print("")
print("If a slice unexpectedly returns '', check your positions.")
````

````python
# EXAMPLE 29: Reversing with a slice versus reversed()
# Both reverse, but they return different things.
word = "Python"

by_slice = word[::-1]
by_function = reversed(word)

print("Slice gives a string:  ", repr(by_slice))
print("reversed() gives:      ", by_function)
print("Which you must join:   ", repr("".join(by_function)))
````

````python
# EXAMPLE 30: Slicing is O(k), not O(1)
# A slice COPIES the characters, so it costs time proportional to its size.
import time

large = "x" * 1_000_000


def time_it(operation, repetitions=1000):
    """Return the milliseconds taken to repeat an operation."""
    start = time.perf_counter()
    for _ in range(repetitions):
        operation()
    return (time.perf_counter() - start) * 1000


small_slice = time_it(lambda: large[:10])
large_slice = time_it(lambda: large[:500_000])

print(f"slicing 10 characters:      {small_slice:.2f} ms")
print(f"slicing 500,000 characters: {large_slice:.2f} ms")
print("")
print("Indexing one character is O(1). Slicing copies, so it is O(k).")
````

````python
# EXAMPLE 31: Slicing a string with multi-byte characters
# Python slices by CHARACTER, not by byte - which is what you want.
text = "cafe" + chr(769) + " menu"
emoji_text = "hello " + chr(128512)

print("text:", text, "length:", len(text))
print("first 4:", text[:4])
print("")
print("emoji_text:", emoji_text, "length:", len(emoji_text))
print("last character:", emoji_text[-1])
print("")
print("In many languages this would split a character in half.")
print("Unicode is covered fully in 07.9.")
````

## Takeaways

1. Indexing starts at **0**. The last character is at `-1`.
2. `text[start:stop]` takes from `start` **up to but not including** `stop` — so
   the slice length is `stop - start`.
3. Omitting a part means "from the beginning" or "to the end". `[:]` is
   everything.
4. `text[::-1]` reverses a string.
5. Indexing out of range raises `IndexError`; **slicing out of range does not** —
   it clamps silently.
6. Slicing a string returns a **new** string, except `[:]` which may return the
   same object because strings are immutable.
7. With a negative step, the start and stop swap roles.
8. Indexing is O(1); slicing is **O(k)** because it copies.
