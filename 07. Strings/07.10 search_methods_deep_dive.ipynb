# 07.10 Search Methods — Deep Dive

Seven methods find text inside text. Every one of them accepts optional
**`start` and `end`** arguments that most people never learn — and those
arguments are the difference between clean code and slicing everything by hand.

| Method | Signature | Returns | When not found |
|---|---|---|---|
| `find` | `find(sub, start, end)` | index | `-1` |
| `rfind` | `rfind(sub, start, end)` | index of **last** | `-1` |
| `index` | `index(sub, start, end)` | index | **raises `ValueError`** |
| `rindex` | `rindex(sub, start, end)` | index of **last** | **raises `ValueError`** |
| `count` | `count(sub, start, end)` | how many times | `0` |
| `startswith` | `startswith(prefix, start, end)` | `True`/`False` | — |
| `endswith` | `endswith(suffix, start, end)` | `True`/`False` | — |

Plus the `in` operator, which answers "is it there?" without telling you where.

## Easy — Finding text

The basic question: is it there, and where?

```python
# EXAMPLE 1: Is the text present at all?
# The `in` operator is the simplest check.
sentence = "hello world"

print("world" in sentence)
print("xyz" in sentence)
```

```python
# EXAMPLE 2: find() gives the position
sentence = "hello world"

print(sentence.find("world"))
```

```python
# EXAMPLE 3: Position 0 means it is at the start
sentence = "hello world"

print(sentence.find("hello"))
```

```python
# EXAMPLE 4: find() returns -1 when absent
# -1 is the "not found" signal, not an error.
sentence = "hello world"

print(sentence.find("xyz"))
```

```python
# EXAMPLE 5: Testing a find() result correctly
# -1 is truthy, so always compare against -1 explicitly.
sentence = "hello world"

position = sentence.find("xyz")

if position != -1:
    print("Found at", position)
else:
    print("Not found")
```

```python
# EXAMPLE 6: The bug: treating -1 as falsy
sentence = "hello world"

# WRONG - position 0 is falsy, and -1 is truthy.
if sentence.find("hello"):
    print("This does NOT run, even though 'hello' is at position 0")
else:
    print("Wrong branch taken - position 0 is falsy")

print("")
print("Always write: if text.find(x) != -1")
```

```python
# EXAMPLE 7: count() counts occurrences
sentence = "the cat sat on the mat"

print("'the' appears", sentence.count("the"), "times")
print("'at' appears", sentence.count("at"), "times")
```

```python
# EXAMPLE 8: startswith() and endswith()
filename = "report_2026.pdf"

print("starts with 'report'?", filename.startswith("report"))
print("ends with '.pdf'?    ", filename.endswith(".pdf"))
```

```python
# EXAMPLE 9: Counting something that is not there
sentence = "hello world"

print(sentence.count("xyz"))
print("count() returns 0, never an error.")
```

## Medium — The arguments nobody teaches

Every search method takes start and end. This is where they earn their keep.

```python
# EXAMPLE 10: Searching from a position with find(start)
# The second argument says where to begin looking.
sentence = "the cat sat on the mat"

print("first 'the': ", sentence.find("the"))
print("next 'the':  ", sentence.find("the", 1))
```

```python
# EXAMPLE 11: Limiting the search with find(start, end)
# The third argument is where to stop - exclusive, like slicing.
sentence = "the cat sat on the mat"

print("search everywhere:   ", sentence.find("the"))
print("search only [10:22]: ", sentence.find("the", 10, 22))
print("search only [0:10]:  ", sentence.find("the", 0, 10))
```

```python
# EXAMPLE 12: find(start, end) matches slice positions
# The returned index is into the ORIGINAL string, not the slice.
sentence = "the cat sat on the mat"

found = sentence.find("the", 10)

print("index returned:", found)
print("that position in the original:", repr(sentence[found:found + 3]))
print("")
print("This is why start/end beats slicing - no index arithmetic.")
```

```python
# EXAMPLE 13: Finding every occurrence with a loop
# Move the start position past each match.
sentence = "the cat sat on the mat"
target = "at"

positions = []
position = sentence.find(target)

while position != -1:
    positions.append(position)
    # Start the next search just after this match.
    position = sentence.find(target, position + 1)

print("found at:", positions)
```

```python
# EXAMPLE 14: rfind() searches from the right
path = "folder/subfolder/file.txt"

print("first slash:", path.find("/"))
print("last slash: ", path.rfind("/"))
```

```python
# EXAMPLE 15: Using rfind() to split a path
# A common real use - getting the filename.
path = "folder/subfolder/file.txt"

last_slash = path.rfind("/")

print("directory:", path[:last_slash])
print("filename: ", path[last_slash + 1:])
```

```python
# EXAMPLE 16: Using rfind() to get a file extension
filename = "archive.tar.gz"

last_dot = filename.rfind(".")

print("name:     ", filename[:last_dot])
print("extension:", filename[last_dot:])
print("")
print("rfind gets the LAST dot, which is what you want here.")
```

```python
# EXAMPLE 17: index() raises instead of returning -1
sentence = "hello world"

print("find:", sentence.find("xyz"))

try:
    sentence.index("xyz")
except ValueError as error:
    print("index raises:", error)
```

```python
# EXAMPLE 18: When to use index() over find()
# Use index() when absence is a genuine error you want to catch.
config_line = "timeout=30"

try:
    separator = config_line.index("=")
    print("key:  ", config_line[:separator])
    print("value:", config_line[separator + 1:])
except ValueError:
    print("Malformed config line - no '=' found")
```

```python
# EXAMPLE 19: rindex() is the right-hand version
path = "a/b/c/d.txt"

print("rfind: ", path.rfind("/"))
print("rindex:", path.rindex("/"))

try:
    path.rindex("|")
except ValueError as error:
    print("rindex on a missing character:", error)
```

```python
# EXAMPLE 20: count() also takes start and end
sentence = "the cat sat on the mat"

print("all 'at':        ", sentence.count("at"))
print("'at' in [0:10]:  ", sentence.count("at", 0, 10))
print("'at' in [10:]:   ", sentence.count("at", 10))
```

```python
# EXAMPLE 21: startswith() accepts a TUPLE of options
# This is the feature that replaces a chain of `or`.
filename = "report.pdf"

print("single:", filename.endswith(".pdf"))
print("tuple: ", filename.endswith((".pdf", ".doc", ".txt")))
print("")
print("A tuple means 'any of these'. A list will NOT work.")
```

```python
# EXAMPLE 22: Why the tuple form matters
filenames = ["a.pdf", "b.doc", "c.png", "d.txt"]

documents = [name for name in filenames
             if name.endswith((".pdf", ".doc", ".txt"))]

print("documents:", documents)
```

```python
# EXAMPLE 23: startswith() with start and end
# Check whether a substring starts with something, without slicing.
line = "    indented text"

print("whole line starts with 'indented'?", line.startswith("indented"))
print("from position 4?                  ", line.startswith("indented", 4))
```

## Hard — Edge cases and performance

The behaviour that surprises people, and how to search efficiently.

```python
# EXAMPLE 24: Searching for an empty string
# An empty string is found everywhere, at position 0.
sentence = "hello"

print("find(''):     ", sentence.find(""))
print("'' in 'hello':", "" in sentence)
print("count(''):    ", sentence.count(""))
print("")
print("count('') returns len + 1 - one gap before each character, plus the end.")
```

```python
# EXAMPLE 25: Overlapping matches are NOT counted
# count() moves past each match, so overlaps are missed.
text = "aaaa"

print("count('aa') in 'aaaa':", text.count("aa"))
print("")
print("There are 3 overlapping 'aa' pairs, but count finds 2.")
print("For overlapping matches you need a manual loop or regex.")
```

```python
# EXAMPLE 26: Counting overlapping matches correctly
def count_overlapping(text, target):
    """Count matches including overlaps."""
    total = 0
    position = text.find(target)

    while position != -1:
        total += 1
        # Advance by ONE, not by the length of the match.
        position = text.find(target, position + 1)

    return total


text = "aaaa"
print("built-in count:", text.count("aa"))
print("overlapping:   ", count_overlapping(text, "aa"))
```

```python
# EXAMPLE 27: Negative start and end positions
# Like slicing, negative positions count from the end.
sentence = "the cat sat on the mat"

print("search the last 10 characters:", sentence.find("the", -10))
print("that slice is:", repr(sentence[-10:]))
```

```python
# EXAMPLE 28: Out-of-range positions do not raise
sentence = "hello"

print("find('l', 100):  ", sentence.find("l", 100))
print("find('l', 0, 100):", sentence.find("l", 0, 100))
print("")
print("Out-of-range start returns -1. Out-of-range end is clamped.")
```

```python
# EXAMPLE 29: Case sensitivity
sentence = "Hello World"

print("find('hello'):", sentence.find("hello"))
print("find('Hello'):", sentence.find("Hello"))
print("")
print("For case-insensitive searching, lower both sides:")
print("casefolded:   ", sentence.casefold().find("hello"))
```

```python
# EXAMPLE 30: A reusable case-insensitive find
def find_ignoring_case(text, target, start=0):
    """Find target in text, ignoring case."""
    # Casefold both sides so the positions still line up for ASCII.
    return text.casefold().find(target.casefold(), start)


sentence = "Hello World"
position = find_ignoring_case(sentence, "WORLD")

print("position:", position)
print("original text there:", repr(sentence[position:position + 5]))
```

```python
# EXAMPLE 31: in versus find for a presence check
import time

text = "the quick brown fox " * 5000
repetitions = 2000


def time_it(operation):
    """Return milliseconds for repeated calls."""
    start = time.perf_counter()
    for _ in range(repetitions):
        operation()
    return (time.perf_counter() - start) * 1000


in_time = time_it(lambda: "fox" in text)
find_time = time_it(lambda: text.find("fox") != -1)

print(f"'fox' in text:        {in_time:6.1f} ms")
print(f"text.find('fox') != -1: {find_time:6.1f} ms")
print("")
print("Use `in` when you only need yes/no. It reads better and is no slower.")
```

```python
# EXAMPLE 32: Searching from the wrong end is slower
import time

# The target is at the very end.
text = "x" * 200000 + "needle"
repetitions = 500


def time_it(operation):
    start = time.perf_counter()
    for _ in range(repetitions):
        operation()
    return (time.perf_counter() - start) * 1000


forward = time_it(lambda: text.find("needle"))
backward = time_it(lambda: text.rfind("needle"))

print(f"find  (scans forward):  {forward:6.1f} ms")
print(f"rfind (scans backward): {backward:6.1f} ms")
print("")
print("Choose the direction that reaches your target first.")
```

```python
# EXAMPLE 33: When to use a regular expression instead
# These methods find LITERAL text. Patterns need the re module.
import re

log = "ERROR 2026-09-14 disk full"

# Literal search - fine with find().
print("literal 'ERROR':", log.find("ERROR"))

# Pattern search - needs regex.
match = re.search(r"\d{4}-\d{2}-\d{2}", log)
print("date pattern:   ", match.group() if match else "not found")

print("")
print("Use string methods for literals - they are faster and clearer.")
print("Use re when you need a pattern. Chapter 30 covers regex.")
```

```python
# EXAMPLE 34: A complete search reference
sentence = "the cat sat on the mat"

print("sentence:", repr(sentence))
print("")

checks = [
    ("'cat' in sentence", "cat" in sentence),
    ("sentence.find('at')", sentence.find("at")),
    ("sentence.rfind('at')", sentence.rfind("at")),
    ("sentence.find('at', 10)", sentence.find("at", 10)),
    ("sentence.find('at', 0, 5)", sentence.find("at", 0, 5)),
    ("sentence.count('at')", sentence.count("at")),
    ("sentence.count('at', 10)", sentence.count("at", 10)),
    ("sentence.startswith('the')", sentence.startswith("the")),
    ("sentence.endswith('mat')", sentence.endswith("mat")),
    ("sentence.startswith('cat', 4)", sentence.startswith("cat", 4)),
]

for expression, result in checks:
    print(f"{expression:<32} -> {result}")
```

## Takeaways

1. **Every search method takes optional `start` and `end`** — they save you from
   slicing and index arithmetic.
2. The returned index is into the **original** string, not the slice.
3. `find()` returns **`-1`** when absent; `index()` **raises `ValueError`**. Use
   `index()` when absence is a real error.
4. **Never test a `find()` result for truthiness** — position `0` is falsy and
   `-1` is truthy. Always compare `!= -1`.
5. `rfind()` and `rindex()` search from the **right** — ideal for file paths and
   extensions.
6. **`startswith()` and `endswith()` accept a tuple** meaning "any of these". A
   list does not work.
7. `count()` does **not** count overlapping matches.
8. Searching for `""` succeeds everywhere; `count("")` returns `len + 1`.
9. Use `in` for a presence check, `find` when you need the position, and `re`
   only when you need a pattern.
