# 07.9 Case Methods — Deep Dive

Six methods change capitalisation. They look interchangeable and are not.

| Method | Signature | Returns |
|---|---|---|
| `upper()` | no arguments | every cased character uppercased |
| `lower()` | no arguments | every cased character lowercased |
| `capitalize()` | no arguments | first character up, **rest forced down** |
| `title()` | no arguments | first letter of each word up |
| `swapcase()` | no arguments | upper becomes lower and vice versa |
| `casefold()` | no arguments | aggressive lowercase for **comparison** |

None take arguments. All return a **new string**. All leave non-letters alone.

The differences only show up in edge cases — which is exactly where bugs live.

## Easy — What each method does

One method per cell, on the same input, so the differences are visible.

```python
# EXAMPLE 1: upper() — everything uppercase
text = "hello World 123"

print(text.upper())
```

```python
# EXAMPLE 2: lower() — everything lowercase
text = "hello World 123"

print(text.lower())
```

```python
# EXAMPLE 3: capitalize() — first letter up, REST DOWN
# Note what happens to the existing capital W.
text = "hello World 123"

print(text.capitalize())
```

```python
# EXAMPLE 4: title() — first letter of every word
text = "hello World 123"

print(text.title())
```

```python
# EXAMPLE 5: swapcase() — flip every letter
text = "hello World 123"

print(text.swapcase())
```

```python
# EXAMPLE 6: casefold() — aggressive lowercase
# For plain English it looks identical to lower().
text = "hello World 123"

print(text.casefold())
```

```python
# EXAMPLE 7: All six on one line, compared
text = "hello World 123"

print("original:  ", text)
print("upper:     ", text.upper())
print("lower:     ", text.lower())
print("capitalize:", text.capitalize())
print("title:     ", text.title())
print("swapcase:  ", text.swapcase())
print("casefold:  ", text.casefold())
```

```python
# EXAMPLE 8: Numbers and symbols are never changed
text = "abc 123 !@#"

print("upper:", text.upper())
print("lower:", text.lower())
print("Only letters are affected.")
```

```python
# EXAMPLE 9: The original is never modified
text = "hello"

text.upper()

print("after calling upper():", text)
print("You must assign:", text.upper())
```

## Medium — The differences that matter

Where capitalize and title behave differently from what you expect.

```python
# EXAMPLE 10: capitalize() destroys existing capitals
# This is the most common capitalize() surprise.
text = "hello WORLD"

print("capitalize:", text.capitalize())
print("")
print("It lowercased WORLD. capitalize() forces the rest down.")
```

```python
# EXAMPLE 11: Keeping the rest unchanged
# If you only want the first letter changed, do it manually.
text = "hello WORLD"

manual = text[:1].upper() + text[1:]

print("capitalize():", text.capitalize())
print("manual:      ", manual)
```

```python
# EXAMPLE 12: title() splits on any non-letter
# This is why title() gets names wrong.
text = "o'brien-smith"

print("title():", text.title())
print("")
print("It capitalised after the apostrophe and the hyphen.")
```

```python
# EXAMPLE 13: title() with a correct alternative
# Split on spaces only, and capitalise each word manually.
name = "o'brien mcdonald"

print("title(): ", name.title())

fixed = " ".join(word[:1].upper() + word[1:] for word in name.split())
print("manual:  ", fixed)
```

```python
# EXAMPLE 14: istitle() checks the same rule
# istitle() agrees with title(), including the odd cases.
candidates = ["Hello World", "Hello world", "O'Brien", "O'brien"]

for text in candidates:
    print(f"{text!r:<16} istitle={text.istitle()}")
```

```python
# EXAMPLE 15: title() on text with numbers
# Digits also count as word boundaries.
text = "abc123def"

print("title():", text.title())
```

```python
# EXAMPLE 16: swapcase() is not always reversible
# For simple ASCII it round-trips fine.
text = "Hello World"

print("original:   ", text)
print("swapped:    ", text.swapcase())
print("swapped x2: ", text.swapcase().swapcase())
print("Round trip equal?", text.swapcase().swapcase() == text)
```

```python
# EXAMPLE 17: Where swapcase() does NOT round-trip
# A character whose uppercase form is two letters breaks it.
text = "ß"

print("original:   ", repr(text))
print("swapped:    ", repr(text.swapcase()))
print("swapped x2: ", repr(text.swapcase().swapcase()))
print("Round trip equal?", text.swapcase().swapcase() == text)
```

```python
# EXAMPLE 18: Empty strings are safe
# Every case method handles an empty string without error.
text = ""

print("upper:     ", repr(text.upper()))
print("capitalize:", repr(text.capitalize()))
print("title:     ", repr(text.title()))
print("No errors - they all return an empty string.")
```

```python
# EXAMPLE 19: Single characters
for character in ["a", "A", "1", " "]:
    print(f"{character!r:<5} upper={character.upper()!r:<5} "
          f"capitalize={character.capitalize()!r}")
```

## Hard — Unicode, comparison and performance

Why casefold exists, and when case conversion changes length.

```python
# EXAMPLE 20: casefold() versus lower() — the German sharp s
# This is the textbook case where lower() is not enough.
first = "straße"
second = "STRASSE"

print("lower comparison:   ", first.lower() == second.lower())
print("casefold comparison:", first.casefold() == second.casefold())
print("")
print("casefold maps ß to ss, so the comparison succeeds.")
```

```python
# EXAMPLE 21: What casefold actually does to ß
text = "ß"

print("original: ", repr(text), "length", len(text))
print("lower():  ", repr(text.lower()), "length", len(text.lower()))
print("casefold():", repr(text.casefold()), "length", len(text.casefold()))
```

```python
# EXAMPLE 22: Case conversion can change the length
# Some characters expand when uppercased.
samples = ["ß", "ﬁ", "ǰ"]

for text in samples:
    upper = text.upper()
    print(f"{text!r:<6} len {len(text)}  ->  upper {upper!r:<8} len {len(upper)}")
```

```python
# EXAMPLE 23: The Turkish dotless i problem
# In Turkish, uppercase i is not I. Python uses the default locale rules.
turkish_dotless = "\u0131"
turkish_dotted = "i"

print("dotless i:", repr(turkish_dotless), "-> upper:",
      repr(turkish_dotless.upper()))
print("normal i: ", repr(turkish_dotted), "-> upper:",
      repr(turkish_dotted.upper()))
print("")
print("Python does not apply Turkish rules. For locale-correct casing,")
print("use a library such as PyICU.")
```

```python
# EXAMPLE 24: Greek final sigma
# Greek sigma has two lowercase forms depending on position.
greek = "ΟΔΟΣ"

print("original:", greek)
print("lower():  ", greek.lower())
print("")
print("Python correctly picks the final form for the last letter.")
```

```python
# EXAMPLE 25: The right way to compare text case-insensitively
# Normalise the encoding, then casefold.
import unicodedata


def same_text(first, second):
    """Compare two strings the way a human reader would."""
    normalised_first = unicodedata.normalize("NFC", first).casefold()
    normalised_second = unicodedata.normalize("NFC", second).casefold()
    return normalised_first == normalised_second


tests = [
    ("Hello", "HELLO"),
    ("straße", "STRASSE"),
    ("caf\u00e9", "CAFE\u0301"),
    ("hello", "world"),
]

for first, second in tests:
    print(f"{first!r:<14} vs {second!r:<14} -> {same_text(first, second)}")
```

```python
# EXAMPLE 26: Never use case methods for security comparisons
# Case folding is not a substitute for proper comparison.
import unicodedata

print("These look identical but are different characters:")
latin_a = "a"
cyrillic_a = "\u0430"

print("latin a:   ", repr(latin_a), unicodedata.name(latin_a))
print("cyrillic a:", repr(cyrillic_a), unicodedata.name(cyrillic_a))
print("Equal after casefold?", latin_a.casefold() == cyrillic_a.casefold())
print("")
print("This is a homograph attack. For usernames and domains you need")
print("explicit script checks, not just case folding.")
```

```python
# EXAMPLE 27: Performance of the case methods
import time

text = "The Quick Brown Fox Jumps Over The Lazy Dog" * 100
repetitions = 20000


def time_it(operation):
    """Return the milliseconds taken to repeat an operation."""
    start = time.perf_counter()
    for _ in range(repetitions):
        operation()
    return (time.perf_counter() - start) * 1000


methods = [
    ("upper", text.upper),
    ("lower", text.lower),
    ("casefold", text.casefold),
    ("capitalize", text.capitalize),
    ("title", text.title),
    ("swapcase", text.swapcase),
]

for name, method in methods:
    print(f"{name:<12} {time_it(method):7.1f} ms")

print("")
print("title() and swapcase() do more work per character.")
```

```python
# EXAMPLE 28: Choosing the right method
guidance = [
    ("Display a heading", "upper()", "shouty but simple"),
    ("Normalise for storage", "lower()", "consistent keys"),
    ("Compare user input", "casefold()", "handles ß and more"),
    ("First letter of a sentence", "text[:1].upper() + text[1:]",
     "capitalize() would lowercase the rest"),
    ("Format a person's name", "manual per-word", "title() breaks O'Brien"),
    ("Format a book title", "title()", "acceptable for simple titles"),
    ("Toggle a display", "swapcase()", "rarely useful in practice"),
]

for situation, choice, why in guidance:
    print(f"{situation:<28} {choice:<30} {why}")
```

## Takeaways

1. All six case methods take **no arguments** and return a **new string**.
2. **`capitalize()` forces the rest of the string lowercase** — use
   `text[:1].upper() + text[1:]` to keep it.
3. **`title()` capitalises after any non-letter**, so it breaks `O'Brien` and
   `mother-in-law`.
4. **`casefold()` is for comparison**, `lower()` is for display. Casefold maps
   `ß` to `ss`.
5. Case conversion **can change the length** of a string.
6. Python does not apply **locale-specific** rules (Turkish dotless i) — use
   PyICU if you need them.
7. For case-insensitive comparison: **normalise, then casefold**.
8. Case folding is **not** a security control — homograph attacks survive it.
