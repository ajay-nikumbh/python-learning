# 07.1 String Literals and Quoting

A string is text. Python gives you several ways to write one, and choosing the
right quote style saves you from a lot of backslashes.

## What you will learn

- The four ways to write a string literal
- How to include quotes inside quotes
- Multi-line strings
- When strings are joined automatically

Work through the **Easy** section first. Stop there if that is enough for today —
Medium and Hard build on it but nothing later depends on finishing them now.

## Easy — Writing your first strings

Start here. Each example is one small idea.

```python
# EXAMPLE 1: Single quotes
# The simplest string - just text between single quotes.
greeting = 'Hello'

print(greeting)
```

```python
# EXAMPLE 2: Double quotes work identically
# Double quotes produce exactly the same kind of string.
greeting = "Hello"

print(greeting)
```

```python
# EXAMPLE 3: Proving both styles are the same
# Python does not care which quotes you used.
with_single = 'Hello'
with_double = "Hello"

print("Are they equal?", with_single == with_double)
print("Same type?", type(with_single) == type(with_double))
```

```python
# EXAMPLE 4: An empty string
# A string with nothing in it is still a valid string.
nothing = ""

print("Length:", len(nothing))
print("Is it falsy?", bool(nothing))
```

```python
# EXAMPLE 5: A string containing spaces
# Spaces are ordinary characters inside a string.
spaced = "a b c"

print(spaced)
print("Length:", len(spaced))
```

```python
# EXAMPLE 6: Putting an apostrophe in text
# Use DOUBLE quotes when the text contains a single quote.
sentence = "It's a sunny day"

print(sentence)
```

```python
# EXAMPLE 7: Putting double quotes in text
# Use SINGLE quotes when the text contains a double quote.
sentence = 'She said "hello" to me'

print(sentence)
```

```python
# EXAMPLE 8: Checking the length of a string
# len() counts the characters.
word = "Python"

print("The word:", word)
print("Number of characters:", len(word))
```

```python
# EXAMPLE 9: Joining two strings with +
# The + operator sticks two strings together.
first = "Hello"
second = "World"

joined = first + second
print(joined)
```

```python
# EXAMPLE 10: Adding a space when joining
# Nothing is added automatically - include the space yourself.
first = "Hello"
second = "World"

joined = first + " " + second
print(joined)
```

```python
# EXAMPLE 11: Repeating a string with *
# Multiplying a string repeats it.
line = "-" * 20

print(line)
```

```python
# EXAMPLE 12: Repeating a word
# The same works for any string.
chant = "go " * 3

print(chant)
```

## Medium — Multi-line strings and escaping

Now that the basics are comfortable, here are the tools for longer text.

```python
# EXAMPLE 13: A multi-line string with triple quotes
# Three quotes let the text span several lines.
message = """Dear customer,

Your order has shipped.

Thank you."""

print(message)
```

```python
# EXAMPLE 14: Triple quotes work with single quotes too
# Both ''' and """ behave the same way.
message = '''Line one
Line two'''

print(message)
```

```python
# EXAMPLE 15: Counting the newlines in a multi-line string
# The line breaks are real characters stored in the string.
message = """one
two
three"""

print("Characters:", len(message))
print("Newlines:", message.count(chr(10)))
```

```python
# EXAMPLE 16: Avoiding a leading blank line
# A triple-quoted string starts immediately after the quotes,
# so this version begins with a newline.
with_gap = """
first line"""

# A backslash at the end joins the first line to the opening quotes.
without_gap = """\
first line"""

print("with_gap starts with a newline?", with_gap.startswith(chr(10)))
print("without_gap starts with a newline?", without_gap.startswith(chr(10)))
```

```python
# EXAMPLE 17: Escaping a newline with backslash-n
# \n inside a string means "start a new line here".
two_lines = "first line\nsecond line"

print(two_lines)
```

```python
# EXAMPLE 18: Escaping a tab
# \t inserts a tab, useful for rough column alignment.
row = "name\tage\tcity"

print(row)
```

```python
# EXAMPLE 19: Escaping a backslash itself
# To get one literal backslash, write two.
path = "C:\\Users\\Asha"

print(path)
print("Actual length:", len(path))
```

```python
# EXAMPLE 20: Escaping a quote inside the same quote style
# If you must use the same quote, escape it with a backslash.
sentence = 'It\'s fine'

print(sentence)
```

```python
# EXAMPLE 21: Seeing escapes with repr()
# print() shows the EFFECT of an escape.
# repr() shows what is actually written in the source.
text = "line one\nline two"

print("With print():")
print(text)
print("")
print("With repr():", repr(text))
```

```python
# EXAMPLE 22: Raw strings ignore escapes
# An r before the quote turns off escape processing entirely.
normal = "C:\name"
raw = r"C:\name"

print("normal:", repr(normal))
print("raw:   ", repr(raw))
```

```python
# EXAMPLE 23: Why raw strings help with Windows paths
# Without r, every backslash must be doubled.
awkward = "C:\\Users\\Asha\\Documents"
easy = r"C:\Users\Asha\Documents"

print("Are they the same?", awkward == easy)
print(easy)
```

```python
# EXAMPLE 24: Adjacent literals join automatically
# Two string literals side by side are merged by the compiler.
message = "Hello, " "World"

print(message)
```

```python
# EXAMPLE 25: Using adjacent literals for long text
# Brackets plus adjacent literals keep long text readable.
message = (
    "This is a long message "
    "split across several lines "
    "for readability."
)

print(message)
```

## Hard — The details that cause bugs

These are the corners worth knowing about before they surprise you.

```python
# EXAMPLE 26: The missing comma trap
# A missing comma between strings silently JOINS them.
with_commas = ["alpha", "beta", "gamma"]
missing_comma = ["alpha", "beta" "gamma"]

print("with commas:   ", with_commas, "->", len(with_commas), "items")
print("missing comma: ", missing_comma, "->", len(missing_comma), "items")
print("")
print("No error was raised. This is why linters matter.")
```

```python
# EXAMPLE 27: Joining happens at compile time
# Python merges adjacent literals before the program runs,
# so there is no concatenation cost at runtime.
import dis

dis.dis(compile('x = "abc" "def"', "<demo>", "exec"))
```

```python
# EXAMPLE 28: Adjacent joining only works for literals
# Two literals join. A literal and a variable do not.
first = "Hello"

# This works - both are literals.
literal_join = "Hello" " World"
print("literal join:", literal_join)

# This is a syntax error, so we test it with compile().
try:
    compile('first " World"', "<demo>", "eval")
except SyntaxError:
    print("variable + literal: SyntaxError - you need the + operator")

print("with +:", first + " World")
```

```python
# EXAMPLE 29: Raw strings cannot end with a backslash
# The backslash still escapes the closing quote, even in a raw string.
try:
    compile(r'x = r"C:\"', "<demo>", "exec")
    print("compiled")
except SyntaxError as error:
    print("SyntaxError:", error.msg)
    print("")
    print("Workaround: use r'C:' + chr(92), or a normal string.")
```

```python
# EXAMPLE 30: Every common escape sequence
# The full set you will actually meet.
escapes = [
    ("\\n", "newline", "a\nb"),
    ("\\t", "tab", "a\tb"),
    ("\\r", "carriage return", "a\rb"),
    ("\\\\", "backslash", "a\\b"),
    ("\\'", "single quote", "a\'b"),
    ('\\"', "double quote", 'a\"b'),
    ("\\0", "null character", "a\0b"),
]

for sequence, name, example in escapes:
    print(f"{sequence:<6} {name:<18} {example!r}")
```

```python
# EXAMPLE 31: Unicode escapes
# You can write any character by its code point.
print("By 4-digit code:  ", "\u00e9")
print("By 8-digit code:  ", "\U0001F600")
print("By name:          ", "\N{GREEK SMALL LETTER ALPHA}")
print("By hex byte:      ", "\x41")
```

```python
# EXAMPLE 32: An unknown escape is left alone
# Python does not error on an unrecognised escape - it keeps the backslash.
# This will become an error in a future version, so avoid relying on it.
import warnings

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    text = eval(r'"\d+"')

print(repr(text))
print("Length:", len(text), "- the backslash was kept")
print("")
print("For regex patterns, always use a raw string: r'\\d+'")
```

```python
# EXAMPLE 33: Which quote style to choose
# A simple decision rule.
SINGLE = chr(39)
DOUBLE = chr(34)

print("Rule 1: no quotes inside  -> use double quotes (PEP 8 default)")
print("   example:", "Hello world")

print("")
print("Rule 2: text contains", SINGLE, "-> use double quotes")
print("   example:", "It" + SINGLE + "s fine")

print("")
print("Rule 3: text contains", DOUBLE, "-> use single quotes")
print("   example:", DOUBLE + "quoted" + DOUBLE)

print("")
print("Rule 4: contains both -> use triple quotes")
print("   example:", "It" + SINGLE + "s " + DOUBLE + "fine" + DOUBLE)
```

## Takeaways

1. Single and double quotes produce **identical** strings — choose whichever
   avoids escaping.
2. Triple quotes span multiple lines and keep the newlines as real characters.
3. `\n` is newline, `\t` is tab, `\\` is one literal backslash.
4. `repr()` shows what is written in the source; `print()` shows the effect.
5. **Raw strings** (`r"..."`) turn escapes off — essential for Windows paths and
   regex patterns.
6. Adjacent string literals **join at compile time**, which is why a missing
   comma in a list silently merges two items.
7. A raw string cannot end with a backslash.
